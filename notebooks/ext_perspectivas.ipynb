## La ecuación de Dirac

<div class="admonition tip">
<p class="admonition-title"><b>[>] Taller — Reescribe la ecuación de Dirac</b></p>

Con los operadores $\hat{p}^\mu = i\partial^\mu$, la ecuación de Dirac es
$(\gamma^\mu \hat{p}_\mu - m) \, \Psi = 0$.

1. Desarrolla la contracción $\gamma^\mu \hat{p}_\mu$ en su parte temporal y espacial, en
   función de $\hat{H}$ y $\hat{\bf p}$. ¿Qué signo lleva el término del momento, y de dónde sale?
2. Multiplica por $\gamma^0$ por la izquierda y despeja $\hat{H}$. Es la forma en la que la
   escribió Dirac en 1928: $\hat{H} = \vec{\alpha} \cdot \hat{\bf p} + \beta \, m$. ¿Qué son
   $\vec{\alpha}$ y $\beta$? ¿Qué les exige que $\hat{H}$ sea hermítico?
3. Aplica $(\gamma^\nu \hat{p}_\nu + m)$ a la ecuación y usa el álgebra
   $\{\gamma^\mu, \gamma^\nu\} = 2 g^{\mu\nu}$. ¿Qué ecuación cumple cada componente de
   $\Psi$? ¿Qué relación entre $E$ y ${\bf p}$ impone sobre una onda plana?
4. Sustituye las dos soluciones, $\Psi_s = u_s \, e^{-i p_\mu x^\mu}$ y
   $\Phi_s = v_s \, e^{+i p_\mu x^\mu}$, y obtén las ecuaciones de los espinores $u_s$ y $v_s$.
   ¿De dónde sale el signo $+m$ de la de $v_s$?

</div>

<details>
<summary><b>Solución</b></summary>

1. $\gamma^\mu \hat{p}_\mu = \gamma^0 \hat{p}_0 + \gamma^k \hat{p}_k = \gamma^0 \hat{H} - \vec{\gamma} \cdot \hat{\bf p}$,
   luego

   $$(\gamma^0 \hat{H} - \vec{\gamma} \cdot \hat{\bf p} - m) \, \Psi = 0$$

   El signo menos es el de la métrica: $\hat{p}_k = -\hat{p}^k$ (índice abajo), y el momento
   físico es el de índice arriba.

2. Con $(\gamma^0)^2 = I$:

   $$\hat{H} \, \Psi = (\gamma^0 \vec{\gamma} \cdot \hat{\bf p} + \gamma^0 m) \, \Psi
   \quad \Rightarrow \quad \alpha^k = \gamma^0 \gamma^k, \;\; \beta = \gamma^0$$

   En Pauli-Dirac, $\alpha^k = \begin{pmatrix} 0 & \sigma_k \\ \sigma_k & 0 \end{pmatrix}$ y
   $\beta = \begin{pmatrix} I & 0 \\ 0 & -I \end{pmatrix}$. Que $\hat{H}$ sea hermítico
   —autovalores reales— exige $\alpha^k$ y $\beta$ hermíticas, y de ahí
   $(\gamma^0)^\dagger = \gamma^0$ y $(\gamma^k)^\dagger = -\gamma^k$.

3. Los términos cruzados en $m$ se cancelan:

   $$(\gamma^\nu \hat{p}_\nu + m)(\gamma^\mu \hat{p}_\mu - m) \, \Psi = (\gamma^\nu \gamma^\mu \, \hat{p}_\nu \hat{p}_\mu - m^2) \, \Psi = 0$$

   Como $\hat{p}_\nu \hat{p}_\mu$ es simétrico en $\mu\nu$, solo cuenta la parte simétrica del
   producto de las $\gamma$: $\gamma^\nu \gamma^\mu \, \hat{p}_\nu \hat{p}_\mu =
   \frac{1}{2} \{\gamma^\nu, \gamma^\mu\} \, \hat{p}_\nu \hat{p}_\mu = \hat{p}^\mu \hat{p}_\mu$.
   Con $\hat{p}^\mu \hat{p}_\mu = -\partial^\mu \partial_\mu$:

   $$(\partial_t^2 - \nabla^2 + m^2) \, \Psi = 0$$

   Cada componente cumple la ecuación de **Klein-Gordon**, y sobre una onda plana,
   $E^2 - {\bf p}^2 = m^2$. La ecuación **lineal** lleva dentro la de Einstein: para eso
   se pidió el álgebra de las $\gamma$.

4. Sobre $\Psi_s$, $\hat{p}_\mu \to +p_\mu$; sobre $\Phi_s$, $\hat{p}_\mu \to -p_\mu$. Las fases
   se simplifican y quedan

   $$(\gamma^\mu p_\mu - m) \, u_s = 0, \qquad (-\gamma^\mu p_\mu - m) \, v_s = 0
   \;\; \Leftrightarrow \;\; (\gamma^\mu p_\mu + m) \, v_s = 0$$

   El $+m$ es el signo menos de la energía y el momento de la antipartícula. Son las dos
   ecuaciones que se comprueban numéricamente en el taller del álgebra de Dirac.

</details>
