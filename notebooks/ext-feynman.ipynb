## Introducción

Esta extensión es el **desarrollo matemático** de la sección de diagramas de Feynman de
[perspectivas](./perspectivas.ipynb), en los puntos que allí se dan sin demostrar. Responde a cuatro
preguntas:

1. **¿De dónde sale el propagador $1/(q^2 - m^2)$?** De la ecuación del campo del mediador.
2. **¿Por qué la masa del mediador fija el alcance?** El potencial de Yukawa, y Coulomb y
   Rutherford como caso sin masa.
3. **¿Qué pasa cuando el mediador se pone *on-shell*?** El polo, la anchura y la Breit-Wigner
   relativista: por qué un $e^+ + e$ a $\sqrt{s} = m_Z$ produce $Z$ a espuertas.
4. **¿Cómo sale la teoría de Fermi del $W$?** El límite de baja energía y los factores de
   $G_F/\sqrt{2} = g_W^2/8m_W^2$.

### Notación

* Unidades naturales, $\hbar = c = 1$; métrica $g_{\mu\nu} = \mathrm{diag}(1, -1, -1, -1)$.
* $q^\mu = (q^0, {\bf q})$ es el cuadrimomento transferido por el mediador; $q^2 = (q^0)^2 - {\bf q}^2$.
* $\Box \equiv \partial_\mu\partial^\mu = \partial_t^2 - \nabla^2$.
* Los factores $i$ y los signos globales de las reglas de Feynman dependen del convenio; aquí
  seguimos la estructura, no los factores de fase. [MT5]

## 1. El propagador

### 1.1. Un campo con fuente

Tomamos el mediador más sencillo, un campo escalar $\phi$ de masa $m$. Libre, cumple la ecuación de
Klein-Gordon (la tabla de ecuaciones de campo de perspectivas). Si hay algo que lo emite —una
**fuente** $j(x)$, que en un diagrama es el vértice—, la ecuación gana un término:

$$
(\Box + m^2)\,\phi(x) = j(x) .
$$

Es la misma estructura que las ecuaciones de Maxwell con cargas: la fuente produce el campo.

**En el espacio de momentos.** Con $\phi(x) = \int \frac{\mathrm{d}^4q}{(2\pi)^4}\, e^{-iq\cdot x}\,\tilde\phi(q)$, cada derivada
se convierte en un factor, $\partial_\mu \to -iq_\mu$, y $\Box \to -q^2$. La ecuación diferencial pasa a ser
**algebraica**:

$$
(-q^2 + m^2)\,\tilde\phi(q) = \tilde{j}(q)
\qquad \Rightarrow \qquad
\boxed{
\tilde\phi(q) = -\,\frac{1}{q^2 - m^2}\;\tilde{j}(q)
}
$$

El campo que produce la fuente es la fuente multiplicada por $1/(q^2 - m^2)$: el **propagador** es la
**inversa del operador de la ecuación del campo**, escrita en momentos.

### 1.2. Del campo al diagrama

En un diagrama de dos vértices, la fuente 1 (un fermión que emite) produce el campo del mediador, y
ese campo actúa sobre la fuente 2 (el fermión que absorbe). La amplitud es

$$
M \;\propto\; \underbrace{g\,\tilde{j}_2}_{\text{vértice 2}}\;\;
\underbrace{\frac{1}{q^2 - m^2}}_{\text{propagador}}\;\;
\underbrace{g\,\tilde{j}_1}_{\text{vértice 1}}
$$

que es exactamente la regla de perspectivas: **dos vértices y un propagador**. El propagador depende
solo de $q^2$ y de la masa del mediador, porque es una propiedad del campo, no de quién lo emite.
Por conservación en cada vértice, $q$ es el cuadrimomento que pierde la fuente 1 y gana la 2.

**Fuera de la capa de masas.** Para una partícula real, $q^2 = m^2$ y el propagador diverge. En un
intercambio, $q^2$ lo fijan los fermiones externos y en general $q^2 \neq m^2$: el mediador es
**virtual** (*off-shell*). En dispersión, $q^2 < 0$, y el denominador nunca se anula.

### 1.3. Mediadores vectoriales

El fotón y los $W^\pm, Z$ tienen espín 1. Su ecuación con fuente es la de Maxwell (sin masa) o la de
Proca (con masa):

$$
\partial_\mu F^{\mu\nu} + m^2 A^\nu = j^\nu .
$$

Con una corriente conservada ($\partial_\nu j^\nu = 0$, sección 4.5 de [ext-dirac](./ext-dirac.ipynb)) se puede
tomar $\partial_\mu A^\mu = 0$, y la ecuación queda $(\Box + m^2)\,A^\nu = j^\nu$: la de antes, componente a
componente. En momentos,

$$
\tilde{A}^\nu(q) = -\,\frac{1}{q^2 - m^2}\;\tilde{j}^\nu(q) ,
$$

y el campo producido por la corriente 1 se acopla a la corriente 2 como $j_2^\mu\, g_{\mu\nu}\, \tilde{A}^\nu$. De ahí la
forma de perspectivas:

$$
\frac{g_{\mu\nu}}{q^2 - m^2}, \qquad \text{y para el fotón,} \qquad \frac{g_{\mu\nu}}{q^2} .
$$

El $g_{\mu\nu}$ contrae los índices de las dos corrientes. (El propagador completo de un vector con masa
lleva además un término $q_\mu q_\nu/m^2$, que se anula al contraerlo con corrientes conservadas.)
[MT5.3]

## 2. El alcance: el potencial de Yukawa

### 2.1. Una fuente en reposo

Una fuente puntual y **estática**, $j = g\,\delta^3({\bf x})$, no depende del tiempo: solo intercambia
momento, no energía, y $q^0 = 0$, $q^2 = -{\bf q}^2$. La ecuación del campo queda

$$
(-\nabla^2 + m^2)\,\phi({\bf x}) = g\,\delta^3({\bf x})
\qquad \Rightarrow \qquad
\tilde\phi({\bf q}) = \frac{g}{{\bf q}^2 + m^2} ,
$$

el propagador con $q^2 = -{\bf q}^2$. Volviendo al espacio de posiciones, con $\int \mathrm{d}\Omega\,e^{i{\bf q}\cdot{\bf r}} = 4\pi \sin(qr)/qr$:

$$
\phi(r) = g\int \frac{\mathrm{d}^3{\bf q}}{(2\pi)^3}\,\frac{e^{i{\bf q}\cdot{\bf r}}}{{\bf q}^2 + m^2}
= \frac{g}{2\pi^2 r}\int_0^\infty \frac{q\,\sin(qr)}{q^2 + m^2}\,\mathrm{d}q
= \boxed{\frac{g}{4\pi}\,\frac{e^{-mr}}{r}}
$$

(la última integral se hace por residuos, con el polo en $q = im$).

### 2.2. Qué se lee

* **El alcance es $1/m$.** El campo cae exponencialmente a distancias mayores que $1/m$ ($\hbar/mc$ en el
  SI). Es la relación de Yukawa de la [introducción](./introduccion.ipynb): el pión, $m_\pi = 140$ MeV, da
  $1.4$ fm, el tamaño del núcleo; el $W$, $2.5 \cdot 10^{-3}$ fm, una interacción prácticamente puntual.
* **Sin masa, Coulomb.** Con $m = 0$ y la carga como fuente, $\phi = e/4\pi r$, y la energía entre dos
  cargas es $e^2/4\pi r = \alpha/r$: el alcance infinito del electromagnetismo.
* **El propagador es el potencial en el espacio de momentos.** Lo que en mecánica cuántica es la
  aproximación de Born —la amplitud de dispersión es la transformada de Fourier del potencial,
  $\tilde{V}({\bf q})$— aquí es el propagador: $\tilde{V}({\bf q}) \propto g^2/({\bf q}^2 + m^2)$.

### 2.3. Rutherford, otra vez

Para el fotón, $\tilde{V}({\bf q}) \propto e^2/{\bf q}^2$. En una dispersión elástica con momento $p$ y ángulo $\theta$,
el momento transferido es

$$
|{\bf q}| = 2p\,\sin\frac{\theta}{2}
$$

(el triángulo isósceles del taller de Rutherford), y la sección eficaz va como el cuadrado de la
amplitud:

$$
\frac{\mathrm{d}\sigma}{\mathrm{d}\Omega} \;\propto\; \left|\frac{e^2}{{\bf q}^2}\right|^2 = \frac{e^4}{16\,p^4 \sin^4(\theta/2)} .
$$

El $1/\sin^4(\theta/2)$ de Rutherford **es el propagador del fotón al cuadrado**. Con los factores completos
(Born, $m$ reducida), se recupera la fórmula de la [introducción](./introduccion.ipynb). Un mediador con
masa cambiaría $1/{\bf q}^4$ por $1/({\bf q}^2 + m^2)^2$, que no diverge en $\theta \to 0$.

*Cuestiones:* ¿cuál es el alcance de un mediador de $m = 1$ GeV? ¿A partir de qué $|{\bf q}|$ se nota su
masa en la sección eficaz?

## 3. El polo y la anchura

### 3.1. Un mediador real

En una aniquilación, el mediador se lleva toda la energía: $q^2 = s > 0$. Cuando $\sqrt{s}$ se acerca a
su masa, el propagador $1/(s - m^2)$ **diverge**: el mediador deja de ser virtual y se produce de
verdad. Pero una partícula real e **inestable** no tiene una masa exacta.

**De dónde sale la anchura.** Un estado de masa $m$ en reposo evoluciona como $e^{-imt}$; si se
desintegra con anchura $\Gamma$, su amplitud decae como $e^{-imt}\,e^{-\Gamma t/2}$ (la probabilidad, como
$e^{-\Gamma t}$). Es como si la masa tuviera una parte imaginaria:

$$
m \;\to\; m - \frac{i\,\Gamma}{2}
\qquad \Rightarrow \qquad
m^2 \;\to\; m^2 - i\,m\,\Gamma \quad (\Gamma \ll m) .
$$

El propagador de un mediador inestable es entonces

$$
\boxed{
\frac{1}{s - m^2 + i\,m\,\Gamma}
}
$$

que ya no diverge: en $s = m^2$ vale $1/(i\,m\Gamma)$, grande pero finito.

### 3.2. La Breit-Wigner relativista

La sección eficaz va como el propagador al cuadrado:

$$
\sigma(s) \;\propto\; \frac{1}{(s - m^2)^2 + m^2\,\Gamma^2} ,
$$

la **Breit-Wigner relativista**. Cerca del pico, $s - m^2 = (\sqrt{s} - m)(\sqrt{s} + m) \simeq 2m\,(\sqrt{s} - m)$, y

$$
\frac{1}{(s - m^2)^2 + m^2\Gamma^2} \simeq \frac{1}{4m^2}\,\frac{1}{(\sqrt{s} - m)^2 + \Gamma^2/4} ,
$$

la Breit-Wigner de la [introducción](./introduccion.ipynb), de anchura total a media altura $\Gamma$. La
anchura del pico de OPAL **es** la anchura del propagador del $Z$.

**El pico del $Z$.** El cálculo completo de $e^+ + e \to Z \to f\bar{f}$ en el pico da

$$
\sigma_{pico} = \frac{12\pi}{m_Z^2}\,\frac{\Gamma_{ee}\,\Gamma_{f\bar f}}{\Gamma_Z^2} .
$$

Para hadrones, con $\Gamma_{ee} = 83.9$ MeV, $\Gamma_{had} = 1744$ MeV y $\Gamma_Z = 2495$ MeV:
$12\pi/m_Z^2 = 4.53 \cdot 10^{-3}$ $\mathrm{GeV}^{-2} = 1765$ nb, por $0.0235$, **$41.5$ nb**. OPAL mide unos $30$ nb en el
máximo: la diferencia es la radiación de los electrones antes de chocar, que ensancha y rebaja el
pico.

### 3.3. «$Z$ a espuertas»

A $\sqrt{s} = m_Z$ compiten dos diagramas: el intercambio de un fotón, con $1/s$, y el de un $Z$, con
$1/(s - m_Z^2 + i\,m_Z\Gamma_Z)$. En el pico, el cociente de propagadores es

$$
\frac{|1/(i\,m_Z\Gamma_Z)|}{|1/m_Z^2|} = \frac{m_Z}{\Gamma_Z} \simeq 37 ,
$$

y en la sección eficaz, que va con el cuadrado, más de mil. Con los acoplos incluidos, en
$e^+ + e \to \mu^+ + \mu$:

* solo el fotón: $\sigma = 4\pi\alpha^2/3s = 0.0104$ nb;
* en el pico del $Z$: $\sigma = 41.5\ \mathrm{nb} \times \Gamma_{\mu\mu}/\Gamma_{had} \simeq 2.0$ nb, **doscientas veces más**.

Es lo que resuelve el taller «De la simetría a un número medible» de perspectivas: la estimación
electromagnética falla por un factor 200 porque el propagador del $Z$ tiene su polo justo ahí. Y
como la anchura total suma todos los canales, incluidos los que no se ven, la forma del pico cuenta
los neutrinos: $N_\nu = 3$ (taller de la introducción).

*Cuestiones:* ¿cuánto vale $m_Z/\Gamma_Z$? ¿Qué pasaría con la altura del pico si hubiera una cuarta familia
con neutrino ligero?

## 4. La teoría de Fermi como límite

### 4.1. El vértice del $W$

La corriente cargada de la interacción débil tiene estructura $V - A$ (tema de leptones), y el vértice
del $W$ con un par de leptones o de quarks aporta

$$
\frac{g_W}{2\sqrt{2}}\;\gamma^\mu\,(1 - \gamma^5) .
$$

El $\frac{1}{2}(1 - \gamma^5)$ es el proyector de quiralidad $P_L$ (sección 1.7 de [ext-dirac](./ext-dirac.ipynb)), y
el $1/\sqrt{2}$ viene de cómo se agrupan los campos $W^\pm$ en la teoría electrodébil: el acoplo a cada
corriente cargada es $g_W/\sqrt{2}$, y con el proyector, $\frac{g_W}{\sqrt{2}}\cdot\frac{1}{2}(1-\gamma^5) = \frac{g_W}{2\sqrt{2}}(1-\gamma^5)$. [MT11.5]

### 4.2. El límite de baja energía

En la desintegración $\beta$, $q^2 \sim \mathcal{O}(\mathrm{MeV}^2) \ll m_W^2$, y el propagador se vuelve una constante:

$$
\frac{g_{\mu\nu}}{q^2 - m_W^2} \;\longrightarrow\; -\,\frac{g_{\mu\nu}}{m_W^2} .
$$

La amplitud, con dos vértices y el propagador, queda (salvo el signo global)

$$
M = \frac{g_W^2}{8\,m_W^2}\;\left[\bar{u}\,\gamma^\mu(1 - \gamma^5)\,u\right]\left[\bar{u}\,\gamma_\mu(1 - \gamma^5)\,v\right] .
$$

Es un acoplo **puntual** de dos corrientes $V - A$: la teoría de Fermi, que en su forma moderna se escribe

$$
M = \frac{G_F}{\sqrt{2}}\;\left[\bar{u}\,\gamma^\mu(1 - \gamma^5)\,u\right]\left[\bar{u}\,\gamma_\mu(1 - \gamma^5)\,v\right] .
$$

Comparando:

$$
\boxed{
\frac{G_F}{\sqrt{2}} = \frac{g_W^2}{8\,m_W^2}
}
$$

El $8 = (2\sqrt{2})^2$ son los dos vértices; el $\sqrt{2}$ de $G_F/\sqrt{2}$ es un convenio, que se mantiene
por razones históricas.

### 4.3. Números

Con $G_F = 1.166 \cdot 10^{-5}$ $\mathrm{GeV}^{-2}$ y $m_W = 80.4$ GeV:

$$
g_W = m_W\sqrt{\frac{8\,G_F}{\sqrt{2}}} = 0.653, \qquad \alpha_W = \frac{g_W^2}{4\pi} = 0.034 \simeq \frac{1}{30} .
$$

**El acoplo débil es mayor que el electromagnético** ($\alpha \simeq 1/137$). La debilidad a baja energía la
pone el propagador: $G_F \propto g_W^2/m_W^2$ es pequeña porque $m_W$ es grande. Con la regla de Sargent
(sección 7 de [ext-cinematica](./ext-cinematica.ipynb)), $\Gamma \propto G_F^2 E_0^5$: el muón da
$\Gamma_\mu = G_F^2\,m_\mu^5/192\pi^3$ y $\tau_\mu = 2.2\,\mu\mathrm{s}$.

**Cuándo deja de valer.** Cuando $q^2$ se acerca a $m_W^2$ el propagador ya no es constante, y la teoría de
Fermi —con $G_F$, de dimensiones $E^{-2}$— da secciones eficaces que crecen como $G_F^2\,s$ sin límite. El $W$
**resuelve** ese crecimiento: por encima de $m_W$ el propagador cae como $1/s$. Es el ejemplo clásico de
una **teoría efectiva** que deja de valer a la escala de la partícula que ha «integrado».

*Cuestiones:* comprueba $g_W = 0.653$. ¿A qué $\sqrt{s}$ el término $q^2/m_W^2$ corrige un $10\,\%$ al propagador
constante?
