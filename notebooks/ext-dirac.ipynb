## 1. ¿De dónde sale la ecuación de Dirac?

### 1.1. Punto de partida

En mecánica cuántica la energía y el momento son operadores:

$$
\hat{H} = i\partial_t, \qquad \hat{\bf p} = -i\nabla .
$$

Si se sustituyen en $E^2 = {\bf p}^2 + m^2$ se obtiene la ecuación de **Klein-Gordon**,
$-\partial_t^2\Psi = (-\nabla^2 + m^2)\Psi$. Es de **segundo orden** en el tiempo, y su densidad de
probabilidad puede ser negativa. [MT4.1]

### 1.2. La idea de Dirac (1928)

* primer orden en $\partial_t$, como la ecuación de Schrödinger;
* la relatividad trata $t$ y ${\bf x}$ por igual: primer orden también en $\nabla$;
* la ecuación de Einstein debe seguir cumpliéndose.

La ecuación más general **lineal** en las derivadas es

$$
(i\gamma^0 \partial_t + i\gamma^k \partial_k - m) \, \Psi = 0 ,
$$

con cuatro coeficientes $\gamma^\mu$ que no conocemos (el de $m$ se toma igual a la unidad).

### 1.3. Al cuadrado, la ecuación de Einstein

**Con los operadores**, $i\partial_t \to E$ e $i\partial_k \to -p^k$, la ecuación se lee

$$
(\gamma^0 E - \gamma^k p^k) \, \Psi = m \, \Psi ,
\qquad \text{es decir,} \qquad (\not{p} - m)\,\Psi = 0 .
$$

**Aplicándola dos veces**, $(\gamma^0 E - \gamma^k p^k)^2\,\Psi = m^2\,\Psi$. Desarrollamos sin cambiar el
orden de las $\gamma$, porque no sabemos si conmutan:

$$
(\gamma^0 E - \gamma^k p^k)^2 = (\gamma^0)^2 E^2 + \sum_k (\gamma^k)^2 (p^k)^2
+ \sum_{i<k} \{\gamma^i, \gamma^k\}\, p^i p^k - \sum_k \{\gamma^0, \gamma^k\}\, E\, p^k .
$$

**Exigimos la ecuación de Einstein**: para que esto sea $E^2 - {\bf p}^2$ para cualquier $E$ y ${\bf p}$,

$$
(\gamma^0)^2 = I, \qquad (\gamma^k)^2 = -I, \qquad \{\gamma^0, \gamma^k\} = 0, \qquad
\{\gamma^i, \gamma^k\} = 0 \;\; (i \neq k),
$$

que se resume en el **álgebra de las $\gamma$**:

$$
\boxed{
\{\gamma^\mu, \gamma^\nu\} = \gamma^\mu\gamma^\nu + \gamma^\nu\gamma^\mu = 2\, g^{\mu\nu}\, I
}
$$

### 1.4. Las $\gamma$ son matrices, y $\Psi$ tiene cuatro componentes

Dos números no nulos no pueden anticonmutar: las $\gamma^\mu$ son **matrices**, y $\Psi$ es una
columna sobre la que actúan. Hacen falta cuatro matrices que anticonmuten dos a dos: con $2\times2$
solo hay tres (las de Pauli); tienen traza nula y autovalores $\pm1$ (o $\pm i$), luego su dimensión
es par. La mínima es $4$: $\Psi$ tiene **cuatro componentes**, es un **espinor de Dirac**.

**Hermiticidad.** Multiplicando por $\gamma^0$ se obtiene la forma hamiltoniana,
$E\,\Psi = H\,\Psi$ con $H = \gamma^0\gamma^k p^k + \gamma^0 m$. Como $H$ debe ser hermítico (energías
reales):

$$
(\gamma^0)^\dagger = \gamma^0, \qquad (\gamma^k)^\dagger = -\gamma^k .
$$

### 1.5. La representación de Pauli-Dirac

Solo el álgebra está fijada; la física no depende de la elección. La habitual, en bloques $2\times2$:

$$
\boxed{
\gamma^0 = \begin{pmatrix} I & 0 \\ 0 & -I \end{pmatrix}, \qquad
\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ -\sigma_k & 0 \end{pmatrix}
}
\qquad
\sigma_1 = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \;
\sigma_2 = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \;
\sigma_3 = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}
$$

con $\{\sigma_i, \sigma_j\} = 2\delta_{ij} I$ y $[\sigma_i, \sigma_j] = 2i\epsilon_{ijk}\sigma_k$. Volviendo a las derivadas:

$$
\boxed{
(i\gamma^\mu \partial_\mu - m)\, \Psi(x) = 0
}
$$

[MT4.2]

### 1.6. ¿Qué son las matrices $\gamma$?

**Como el número $i$.** Con números reales, $x^2 + y^2$ no se puede escribir como producto de dos
factores lineales; hace falta un número nuevo con $i^2 = -1$:

$$
x^2 + y^2 = (x + iy)(x - iy) .
$$

Dirac necesita lo mismo con $E^2 - {\bf p}^2$, y ningún número sirve: hacen falta objetos cuyo
cuadrado dé el signo de la métrica y que **anticonmuten**, para cancelar los términos cruzados. En
tres dimensiones ya lo conocemos: $(\vec\sigma\cdot{\bf p})^2 = {\bf p}^2$, así que $\vec\sigma\cdot{\bf p}$ es la
«raíz cuadrada» de ${\bf p}^2$. Dirac hace lo mismo en el espacio-tiempo:

$$
\not{p}^{\,2} = (\gamma^\mu p_\mu)^2 = p_\mu p^\mu = m^2 .
$$

**La métrica escrita con matrices.** El álgebra $\{\gamma^\mu, \gamma^\nu\} = 2g^{\mu\nu}I$ dice que las
$\gamma^\mu$ se comportan como cuatro vectores unitarios del espacio-tiempo: el temporal de cuadrado
$+1$, los espaciales de cuadrado $-1$, y «ortogonales» entre sí.

### 1.7. La quinta matriz, $\gamma^5$

Con las cuatro $\gamma^\mu$ se construye una quinta:

$$
\boxed{
\gamma^5 \equiv i\,\gamma^0\gamma^1\gamma^2\gamma^3
}
$$

Sus propiedades salen solo del álgebra $\{\gamma^\mu, \gamma^\nu\} = 2g^{\mu\nu}I$:

* **Anticonmuta con las cuatro $\gamma^\mu$**, $\{\gamma^5, \gamma^\mu\} = 0$. Para pasar $\gamma^\mu$ de un lado al otro
  del producto hay que cruzarla con las tres $\gamma$ distintas (un signo menos cada una) y con ella
  misma (sin signo): $(-1)^3 = -1$.
* **Su cuadrado es la identidad**, $(\gamma^5)^2 = I$. Reordenando $\gamma^0\gamma^1\gamma^2\gamma^3\,\gamma^0\gamma^1\gamma^2\gamma^3$
  hasta juntar cada $\gamma$ con su pareja salen $(-1)^{3+2+1} = +1$ y
  $(\gamma^0)^2(\gamma^1)^2(\gamma^2)^2(\gamma^3)^2 = (+1)(-1)^3 = -1$; con el $i^2 = -1$ de delante, $+I$.
* **Es hermítica**, $(\gamma^5)^\dagger = \gamma^5$. Con $(\gamma^0)^\dagger = \gamma^0$ y $(\gamma^k)^\dagger = -\gamma^k$:
  $(\gamma^5)^\dagger = -i\,(\gamma^3)^\dagger(\gamma^2)^\dagger(\gamma^1)^\dagger(\gamma^0)^\dagger = i\,\gamma^3\gamma^2\gamma^1\gamma^0$, y dar la vuelta
  al producto de cuatro cuesta seis permutaciones: $(-1)^6 = +1$.

**En Pauli-Dirac.** Con $\gamma^1\gamma^2 = \mathrm{diag}(-i\sigma_3, -i\sigma_3)$ (porque $\sigma_1\sigma_2 = i\sigma_3$) y multiplicando
por $\gamma^3$ y $\gamma^0$:

$$
\gamma^5 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix} :
$$

intercambia el bloque de arriba con el de abajo.

**No cambia con Lorentz.** $\gamma^5$ conmuta con los generadores: con $\Sigma_k$ (diagonal por bloques,
con el mismo bloque arriba y abajo) y con $\gamma^0\gamma^k$ (anticonmuta con cada factor). Luego
conmuta con cualquier $S(\Lambda)$:

$$
S^{-1}\,\gamma^5\, S = \gamma^5 .
$$

Con ella se definen los proyectores de **quiralidad**, $P_{L,R} = \frac{1}{2}(I \mp \gamma^5)$, y la corriente
axial (sección 4.4). La quiralidad está desarrollada en [ext-dirac-old](./ext-dirac-old.ipynb). [MT4.2, MT6.4]

## 2. ¿Cuáles son las soluciones?

### 2.1. Ondas planas y las ecuaciones de los espinores

Buscamos soluciones de onda plana: una columna de cuatro constantes por una fase,

$$
\Psi_s(x) = u_s(p)\, e^{-i p_\mu x^\mu}, \qquad \Phi_s(x) = v_s(p)\, e^{+i p_\mu x^\mu}, \qquad
E = +\sqrt{{\bf p}^2 + m^2} .
$$

Toda la dependencia en $(t, {\bf x})$ está en la fase; el **espinor** depende solo de $p$. Las derivadas
actúan solo sobre la fase ($i\partial_\mu \to \pm p_\mu$), y la ecuación de Dirac se convierte en una
ecuación **algebraica** para los espinores:

$$
\boxed{
(\not{p} - m)\, u_s(p) = 0, \qquad (\not{p} + m)\, v_s(p) = 0
}
$$

Son las ecuaciones fundamentales de los espinores. Las resolvemos en tres pasos: en reposo, con
el momento en $z$ y con el momento en una dirección cualquiera.

### 2.2. Einstein otra vez, y los dos signos de la energía

La ecuación $(\not{p} - m)\,w = 0$ es un sistema lineal **homogéneo**: solo tiene solución $w \neq 0$ si el
determinante se anula. Multiplicando por $(\not{p} + m)$ se ve en una línea:

$$
(\not{p} + m)(\not{p} - m)\, w = (\not{p}^{\,2} - m^2)\, w = (p^2 - m^2)\, w = 0
\quad \Rightarrow \quad E^2 = {\bf p}^2 + m^2 .
$$

La onda plana solo es solución si $E$ y ${\bf p}$ son los de una partícula de masa $m$. Pero la ecuación
fija $E^2$, no $E$: hay soluciones con $E = \pm\sqrt{{\bf p}^2 + m^2}$. Las de energía negativa se
escriben cambiando el signo de toda la fase, de modo que en las fórmulas $E > 0$: son las $\Phi_s$,
con $v_s$.

**Lectura limpia:** como $\not{p}^{\,2} = m^2$, la matriz $\not{p}$ solo tiene autovalores $\pm m$, cada uno dos
veces (tiene traza nula):

$$
\not{p}\, u = +m\, u, \qquad \not{p}\, v = -m\, v .
$$

Partícula y antipartícula son los **dos subespacios** de la misma matriz.

### 2.3. En reposo, ${\bf p} = 0$

Con ${\bf p} = 0$, $E = m$ y $\not{p} = m\,\gamma^0$. Las dos ecuaciones son diagonales:

$$
\not{p} - m = m\,(\gamma^0 - I) = \begin{pmatrix} 0 & 0 \\ 0 & -2m\, I \end{pmatrix}, \qquad
\not{p} + m = m\,(\gamma^0 + I) = \begin{pmatrix} 2m\, I & 0 \\ 0 & 0 \end{pmatrix} .
$$

* $(\not{p} - m)\,u = 0$ solo exige que el **bloque de abajo** sea cero: el de arriba, $\chi$, es libre.
* $(\not{p} + m)\,v = 0$ exige que el **bloque de arriba** sea cero: el de abajo, $\xi$, es libre.

Dos componentes libres en cada caso: los **dos estados de espín**, con el espinor de Pauli de siempre.
Con la normalización $u^\dagger u = 2E = 2m$ (sección 2.4), $N = \sqrt{2m}$:

$$
u_1(m) = \sqrt{2m} \begin{pmatrix} 1 \\ 0 \\ 0 \\ 0 \end{pmatrix}, \;\;
u_2(m) = \sqrt{2m} \begin{pmatrix} 0 \\ 1 \\ 0 \\ 0 \end{pmatrix}, \;\;
v_1(m) = \sqrt{2m} \begin{pmatrix} 0 \\ 0 \\ 0 \\ 1 \end{pmatrix}, \;\;
v_2(m) = \sqrt{2m} \begin{pmatrix} 0 \\ 0 \\ 1 \\ 0 \end{pmatrix}
$$

La partícula vive arriba y la antipartícula abajo: son los autoespacios de $\gamma^0$ con autovalor
$+1$ y $-1$. En $v$ el orden de las componentes está cruzado para que $s = 1$ sea espín físico arriba
también en la antipartícula (sección 2.6).

### 2.4. Con el momento en $z$

Tomamos ${\bf p} = (0, 0, \mathrm{p})$. Con $\not{p} = \gamma^0 E - \gamma^3 \mathrm{p}$ y $u = (u^a, u^b)$ en bloques:

$$
\not{p} - m =
\begin{pmatrix} (E - m)\, I & -\mathrm{p}\,\sigma_3 \\ \mathrm{p}\,\sigma_3 & -(E + m)\, I \end{pmatrix} .
$$

* $\gamma^0$ (la energía) es diagonal: **no mezcla** arriba con abajo.
* $\gamma^3$ (el momento) está fuera de la diagonal: **liga** los dos bloques, con peso $\mathrm{p}$.

Cada fila de bloques es una ecuación:

$$
(E - m)\, u^a = \mathrm{p}\,\sigma_3\, u^b, \qquad (E + m)\, u^b = \mathrm{p}\,\sigma_3\, u^a .
$$

De la segunda, el bloque de abajo queda **fijado** por el de arriba, $u^b = \kappa\,\sigma_3\,u^a$, con
$\kappa = \mathrm{p}/(E+m)$. La primera no añade nada: con $\sigma_3^2 = I$ solo exige $E^2 = \mathrm{p}^2 + m^2$.
Quedan **dos componentes libres**, las de $u^a$, como en reposo. Para $v$ (con $+m$) el bloque libre es
el de abajo, y $v^a = \kappa\,\sigma_3\,v^b$.

**Normalización.** Se toma $u^\dagger u = v^\dagger v = 2E$. La densidad $\Psi^\dagger\Psi$ no es invariante:
con un *boost* el volumen se contrae y la densidad crece como $E$. La única normalización válida en
todos los sistemas es proporcional a $E$; el 2 es un convenio, el mismo que para los bosones. Así,
$N^2(1 + \kappa^2) = 2E$ da $N = \sqrt{E+m}$, y los cuatro espinores son

$$
\boxed{
u_1 = N \begin{pmatrix} 1 \\ 0 \\ \kappa \\ 0 \end{pmatrix}, \;\;
u_2 = N \begin{pmatrix} 0 \\ 1 \\ 0 \\ -\kappa \end{pmatrix}, \;\;
v_1 = N \begin{pmatrix} 0 \\ -\kappa \\ 0 \\ 1 \end{pmatrix}, \;\;
v_2 = N \begin{pmatrix} \kappa \\ 0 \\ 1 \\ 0 \end{pmatrix}
}
\qquad \kappa = \frac{\mathrm{p}}{E+m}, \;\; N = \sqrt{E+m}
$$

Cómo cambian con $\kappa$ (en reposo, en movimiento, ultrarrelativista): ver [perspectivas](./perspectivas.ipynb), *¿Cuán relativista es el espinor?*

### 2.5. Con el momento en una dirección cualquiera

El único cambio es $\mathrm{p}\,\sigma_3 \to \vec\sigma\cdot{\bf p}$ en los bloques de fuera de la diagonal:

$$
\not{p} - m =
\begin{pmatrix} (E - m)\, I & -\vec\sigma\cdot{\bf p} \\ \vec\sigma\cdot{\bf p} & -(E + m)\, I \end{pmatrix},
\qquad
u^b = \frac{\vec\sigma\cdot{\bf p}}{E+m}\, u^a .
$$

La primera ecuación vuelve a no añadir nada porque $(\vec\sigma\cdot{\bf p})^2 = {\bf p}^2$, que hace el papel de
$\sigma_3^2 = I$. Con los bloques libres $\chi_1 = (1, 0)$, $\chi_2 = (0, 1)$ y, en orden cruzado,
$\xi_1 = (0, 1)$, $\xi_2 = (1, 0)$:

$$
\boxed{
u_s(p) = N \begin{pmatrix} \chi_s \\ \dfrac{\vec\sigma\cdot{\bf p}}{E+m}\,\chi_s \end{pmatrix}, \qquad
v_s(p) = N \begin{pmatrix} \dfrac{\vec\sigma\cdot{\bf p}}{E+m}\,\xi_s \\ \xi_s \end{pmatrix}, \qquad N = \sqrt{E+m}
}
$$

La normalización es la misma: $(\vec\sigma\cdot{\bf p})^2 = {\bf p}^2$ da otra vez $u^\dagger u = 2E$. Con
${\bf p} = \mathrm{p}\,\hat{z}$ se recuperan los espinores de la sección 2.4, y con ${\bf p} = 0$ los de la 2.3.

### 2.6. Energía, momento y antipartícula

Sobre la fase, $i\partial_t\,\Phi_s = -E\,\Phi_s$ y $-i\nabla\,\Phi_s = -{\bf p}\,\Phi_s$: en $\Phi_s$ los operadores dan el
signo cambiado, y la regla práctica es invertirlo en todo (energía, momento, espín y cargas) para leer la
antipartícula. La interpretación, en [perspectivas](./perspectivas.ipynb) (*Antipartículas*) y en
[ext-dirac-old](./ext-dirac-old.ipynb) (*Feynman y Stückelberg*).

## 3. ¿Cómo se transforman los espinores?

### 3.1. La condición sobre $S(\Lambda)$

Bajo una transformación de Lorentz $\Lambda$, $x' = \Lambda x$, un cuadrivector se transforma con $\Lambda$ y
el cuadrigradiente, que es covariante, con la inversa: $\partial'_\mu = (\Lambda^{-1})^\nu{}_\mu\,\partial_\nu$. Un
espinor se transforma con una matriz compleja $4\times4$, $S(\Lambda)$:

$$
\Psi'(x') = S(\Lambda)\, \Psi(x) .
$$

**Lo que exigimos:** la ecuación de Dirac tiene la **misma forma** en todos los sistemas inerciales, con
las **mismas** $\gamma^\mu$. Si $(i\gamma^\nu\partial_\nu - m)\,\Psi = 0$, debe cumplirse

$$
(i\gamma^\mu\partial'_\mu - m)\, S\,\Psi
= \left[ i\,\gamma^\mu (\Lambda^{-1})^\nu{}_\mu\, \partial_\nu - m \right] S\,\Psi = 0 .
$$

Multiplicando por $S^{-1}$ por la izquierda, esto coincide con la ecuación original si
$S^{-1}\gamma^\mu S\,(\Lambda^{-1})^\nu{}_\mu = \gamma^\nu$, es decir:

$$
\boxed{
S^{-1}(\Lambda)\, \gamma^\mu\, S(\Lambda) = \Lambda^\mu{}_\nu\, \gamma^\nu
}
$$

Es la condición que define la **representación espinorial** de la transformación de Lorentz: rotar las
$\gamma$ con $\Lambda$ es lo mismo que conjugarlas con $S$.

**¿Por qué las $\gamma$ pueden ser las mismas en todos los sistemas?** Lo exige el principio de
relatividad (la misma ecuación para todos), y es posible porque lo único que se pide a las $\gamma$ es su
álgebra, $\{\gamma^\mu, \gamma^\nu\} = 2g^{\mu\nu}I$, y la métrica es la misma en todos los sistemas
($\Lambda^\mu{}_\alpha\Lambda^\nu{}_\beta\, g^{\alpha\beta} = g^{\mu\nu}$). Un conjunto rotado, $\Lambda^\mu{}_\nu\gamma^\nu$, cumple
la misma álgebra, y por el **teorema fundamental de Pauli** dos conjuntos que cumplen el álgebra están
relacionados por una semejanza: existe $S$ con $\Lambda^\mu{}_\nu\gamma^\nu = S^{-1}\gamma^\mu S$. Rotar las $\gamma$
equivale a un cambio de base en el espacio de los espinores, que se le pasa a $\Psi$.

**Qué significa para $\not{p}$.** Las $\gamma^\mu$ son matrices fijas, no las componentes de un cuadrivector.
$\gamma^\mu v_\mu$ tiene el índice contraído, pero no es un número invariante: de la condición,

$$
\gamma^\mu v'_\mu = S\, (\gamma^\nu v_\nu)\, S^{-1},
$$

se transforma **por semejanza**, a la vez que el espinor. Eso es la **covariancia** de la ecuación de
Dirac. Los números invariantes aparecen al cerrar con el adjunto, $\bar\Psi = \Psi^\dagger\gamma^0$: lo
vemos en la sección 4.

Para la deducción detallada de los generadores en todas las direcciones, ver
[ext-trans-Lorentz](./ext-trans-Lorentz.ipynb).

### 3.2. Transformaciones infinitesimales: los generadores

Buscamos $S = I + \omega$, con $\omega$ pequeña, que cumpla la condición para una transformación
infinitesimal. Como $S^{-1} = I - \omega$, la condición queda

$$
S^{-1}\gamma^\mu S \simeq \gamma^\mu + [\gamma^\mu, \omega] = \Lambda^\mu{}_\nu\,\gamma^\nu .
$$

**Rotación $\delta\phi$ en torno a $z$** (antihoraria): $\gamma^1 \to \gamma^1 - \delta\phi\,\gamma^2$,
$\gamma^2 \to \gamma^2 + \delta\phi\,\gamma^1$, y $\gamma^0, \gamma^3$ no cambian. $\omega$ debe conmutar con
$\gamma^0$ y $\gamma^3$: probamos $\omega = b\,\gamma^1\gamma^2$. Con $[\gamma^1, \gamma^1\gamma^2] = -2\gamma^2$ sale
$b = \delta\phi/2$, y como $\gamma^1\gamma^2 = -i\,\Sigma_3$:

$$
S = I - i\,\frac{\delta\phi}{2}\,\Sigma_3 = I - i\,\delta\phi\, S_3 .
$$

**El generador de las rotaciones es el espín**, $S_3 = \frac{1}{2}\Sigma_3$.

***Boost* $\delta\eta$ en $z$**: $\gamma^0 \to \gamma^0 + \delta\eta\,\gamma^3$, $\gamma^3 \to \gamma^3 + \delta\eta\,\gamma^0$, y
$\gamma^1, \gamma^2$ no cambian. Probamos $\omega = a\,\gamma^0\gamma^3$. Con $[\gamma^0, \gamma^0\gamma^3] = 2\gamma^3$ sale
$a = \delta\eta/2$:

$$
S = I + \frac{\delta\eta}{2}\,\gamma^0\gamma^3 .
$$

**Transformación finita.** Se encadenan muchas infinitesimales: $S$ es la exponencial del generador.
Los ángulos aparecen **divididos por dos**.

### 3.3. Rotación en $z$: el espín

**En $z$.** Como $\Sigma_3^2 = I$, la serie de la exponencial se suma:

$$
\boxed{
S(\phi\,\hat{z}) = e^{-i\phi\,S_3} = \cos\frac{\phi}{2}\, I - i \sin\frac{\phi}{2}\, \Sigma_3
= \mathrm{diag}\left(e^{-i\phi/2},\, e^{+i\phi/2},\, e^{-i\phi/2},\, e^{+i\phi/2}\right)
}
$$

La rotación no mezcla los bloques de arriba y de abajo. Sobre los espinores con ${\bf p}$ en $z$ (una
rotación en torno a $z$ no cambia ${\bf p}$) solo añade una **fase**:

$$
S(\phi\,\hat{z})\, u_1 = e^{-i\phi/2}\, u_1, \qquad S(\phi\,\hat{z})\, u_2 = e^{+i\phi/2}\, u_2,
\qquad \text{es decir,} \qquad S(\phi\,\hat{z})\, u_s = e^{-i m_s \phi}\, u_s ,
$$

con $m_s = \pm\frac{1}{2}$ los autovalores de $S_3$: $S_3 u_1 = +\frac12 u_1$, $S_3 u_2 = -\frac12 u_2$. Con
$\phi = 2\pi$, $S = -I$: el espinor vuelve **cambiado de signo**; hacen falta dos vueltas, $4\pi$.

**¿Por qué es un espín, un momento angular intrínseco?**

1. **Contar estados.** Las rotaciones actúan sobre las dos componentes libres de la sección 2.3; un
   momento angular $s$ tiene $2s + 1$ estados, y $2s + 1 = 2$ da $s = \frac{1}{2}$.
2. **Quién genera las rotaciones.** El generador sobre el espinor es ${\bf S} = \frac{1}{2}{\bf \Sigma}$, que
   cumple el álgebra del momento angular, $[S_i, S_j] = i\epsilon_{ijk}S_k$. Es **intrínseco**: actúa sobre
   las componentes, no depende de ${\bf x}$. El medio ángulo, $e^{-i m_s\phi}$, es el $\frac{1}{2}$.
3. **La conservación lo exige.** El hamiltoniano de Dirac no conmuta con el momento angular orbital;
   solo con el total: $[H, {\bf L}] \neq 0$, $[H, {\bf J}] = 0$, ${\bf J} = {\bf L} + {\bf S}$ (lo
   demostramos a continuación). [MT4.4]

En la antipartícula el operador da el signo cambiado (sección 2.6): $S_3 v_1 = -\frac{1}{2}v_1$, pero el
espín físico de $v_1$ es $+\frac{1}{2}$.

**$[H, {\bf L}] \neq 0$, pero $[H, {\bf J}] = 0$.** El hamiltoniano de la sección 1.4, con el momento como
operador, $\hat{p}^k = -i\partial_k$, es

$$
H = \alpha_k\, \hat{p}^k + \gamma^0 m, \qquad
\alpha_k \equiv \gamma^0\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ \sigma_k & 0 \end{pmatrix} .
$$

Usamos $[x^j, \hat{p}^k] = i\,\delta_{jk}$, $L_i = \epsilon_{ijk}\, x^j \hat{p}^k$ y $[\sigma_i, \sigma_j] = 2i\,\epsilon_{ijk}\,\sigma_k$.

* **El orbital.** $\gamma^0 m$ y las $\alpha_k$ son matrices constantes: conmutan con $x^j$ y $\hat{p}^k$. Solo
  cuenta el conmutador de $\hat{p}^l$ con $x^j$:

  $$
  [H, L_i] = \alpha_l\, \epsilon_{ijk}\, [\hat{p}^l, x^j]\, \hat{p}^k = -i\, \epsilon_{ijk}\, \alpha_j\, \hat{p}^k
  = -i\, (\vec\alpha \times \hat{\bf p})_i \neq 0 .
  $$

* **El espín.** $S_i = \frac{1}{2}\Sigma_i$ actúa solo sobre las componentes: conmuta con $\hat{p}^k$ y con
  $\gamma^0$ (los dos son diagonales por bloques, con el mismo bloque arriba y abajo). Con las $\alpha_k$,
  bloque a bloque, $[\alpha_k, \Sigma_i] = 2i\,\epsilon_{kil}\,\alpha_l$, y

  $$
  [H, S_i] = \frac{1}{2}\, \hat{p}^k\, [\alpha_k, \Sigma_i] = i\, \epsilon_{ilk}\, \alpha_l\, \hat{p}^k
  = +i\, (\vec\alpha \times \hat{\bf p})_i .
  $$

Los dos se cancelan en la suma:

$$
\boxed{
[H, {\bf J}] = [H, {\bf L}] + [H, {\bf S}] = 0, \qquad {\bf J} = {\bf L} + {\bf S}
}
$$

**Qué se lee:**

* Ni ${\bf L}$ ni ${\bf S}$ se conservan por separado; solo la suma. El espín no se postula: hace falta para
  que se conserve el momento angular.
* **El $\frac{1}{2}$ está fijado.** Con ${\bf S} = \lambda\,{\bf \Sigma}$ saldría $[H, {\bf S}] = 2\lambda\, i\,(\vec\alpha\times\hat{\bf p})$, que
  solo cancela a $[H, {\bf L}]$ si $\lambda = \frac{1}{2}$: la ecuación fija el valor del espín.
* **Con ${\bf p}$ en $z$**, $(\vec\alpha\times{\bf p})_3 = \alpha_1 p^2 - \alpha_2 p^1 = 0$: $[H, S_3] = 0$, y $S_3$ es un buen
  número cuántico de las soluciones de la sección 2.4.
* **La helicidad se conserva.** $[H, {\bf \Sigma}\cdot\hat{\bf p}] = \hat{p}^k \hat{p}^i\, [\alpha_k, \Sigma_i] = 2i\,\epsilon_{kil}\,\hat{p}^k\hat{p}^i\,\alpha_l = 0$
  (simétrico por antisimétrico), así que $h = {\bf S}\cdot\hat{\bf p}/\mathrm{p}$ conmuta con $H$ en cualquier
  dirección.

### 3.4. Rotación en una dirección cualquiera (solo indicado)

Un ángulo $\alpha$ en torno a un eje $\hat{\bf n}$. Como $(\hat{\bf n}\cdot{\bf \Sigma})^2 = I$, igual que en $z$:

$$
\boxed{
S(\alpha\,\hat{\bf n}) = e^{-i\alpha\,\hat{\bf n}\cdot{\bf S}} = \cos\frac{\alpha}{2}\, I - i \sin\frac{\alpha}{2}\, (\hat{\bf n}\cdot{\bf \Sigma})
}
$$

**Llevar el espín a una dirección $\hat{\bf v}$.** Primero un ángulo $\theta$ en torno a $y$, después $\phi$ en
torno a $z$: $S(R) = S(\phi\,\hat{z})\,S(\theta\,\hat{y})$, con
$e^{-i\frac{\theta}{2}\sigma_2} = \begin{pmatrix} \cos\frac{\theta}{2} & -\sin\frac{\theta}{2} \\ \sin\frac{\theta}{2} & \cos\frac{\theta}{2} \end{pmatrix}$.
Sobre los bi-espinores (sin demostración):

$$
\chi_1 \;\to\; e^{-i\phi/2} \begin{pmatrix} \cos\frac{\theta}{2} \\ e^{i\phi}\sin\frac{\theta}{2} \end{pmatrix} \equiv e^{-i\phi/2}\,\chi_+(\hat{\bf v}),
\qquad
\chi_2 \;\to\; e^{-i\phi/2} \begin{pmatrix} -\sin\frac{\theta}{2} \\ e^{i\phi}\cos\frac{\theta}{2} \end{pmatrix} \equiv e^{-i\phi/2}\,\chi_-(\hat{\bf v}) ,
$$

que cumplen $(\hat{\bf v}\cdot\vec\sigma)\,\chi_\pm = \pm\chi_\pm$: el espín apunta ahora en $\hat{\bf v}$.

**El fermión en movimiento.** Aplicada a $u_\pm$ con ${\bf p}$ en $z$, la rotación gira también el momento,
$\mathrm{p}\,\hat{z} \to \mathrm{p}\,\hat{\bf v}$, y $\sigma_3 \to \hat{\bf v}\cdot\vec\sigma$:

$$
S(R)\, u_\pm(\mathrm{p}\,\hat{z}) \propto N \begin{pmatrix} \chi_\pm(\hat{\bf v}) \\ \pm\kappa\,\chi_\pm(\hat{\bf v}) \end{pmatrix} = u_\pm(\mathrm{p}\,\hat{\bf v}) :
$$

espín y momento giran juntos, y la **helicidad** (el espín en la dirección del momento) no cambia.

### 3.5. *Boost* en $z$

 Como $(\gamma^0\gamma^3)^2 = I$ y $\gamma^0\gamma^3 = \begin{pmatrix} 0 & \sigma_3 \\ \sigma_3 & 0 \end{pmatrix}$:

$$
\boxed{
S(\eta\,\hat{z}) = e^{\frac{\eta}{2}\,\gamma^0\gamma^3} = \cosh\frac{\eta}{2}\, I + \sinh\frac{\eta}{2}\,\gamma^0\gamma^3
= \begin{pmatrix} \cosh\frac{\eta}{2}\, I & \sinh\frac{\eta}{2}\,\sigma_3 \\ \sinh\frac{\eta}{2}\,\sigma_3 & \cosh\frac{\eta}{2}\, I \end{pmatrix}
}
$$

A diferencia de la rotación, el *boost* **mezcla** los bloques. Sobre el espinor en reposo,
$u_s(m) = \sqrt{2m}\,(\chi_s, 0)$, con las fórmulas del ángulo mitad
$\sqrt{2m}\cosh\frac{\eta}{2} = \sqrt{E+m}$ y $\tanh\frac{\eta}{2} = \frac{\mathrm{p}}{E+m}$:

$$
S(\eta\,\hat{z})\, u_s(m) = \sqrt{2m} \begin{pmatrix} \cosh\frac{\eta}{2}\,\chi_s \\ \sinh\frac{\eta}{2}\,\sigma_3\,\chi_s \end{pmatrix}
= \sqrt{E+m} \begin{pmatrix} \chi_s \\ \kappa\,\sigma_3\,\chi_s \end{pmatrix} = u_s(p) .
$$

* Recuperamos, **sin resolver la ecuación**, el espinor de la sección 2.4. Y $\kappa = \tanh\frac{\eta}{2}$: la
  tangente de la mitad de la rapidez.
* $N = \sqrt{E+m}$ es lo que da el *boost* partiendo de $\sqrt{2m}$ en reposo.
* El *boost* va a lo largo del eje del espín y no cambia $S_3$: $u_1(m)$ pasa a $u_+(p)$.
* **No es unitario**: $u^\dagger u$ pasa de $2m$ a $2E$, mientras que $\bar u u = N^2(1 - \kappa^2) = 2m$ no cambia.
* La fase no cambia: depende de $p_\mu x^\mu$, invariante Lorentz; tras el *boost* se escribe
  $p'_\mu x'^\mu$, pero vale lo mismo.

### 3.6. *Boost* en una dirección cualquiera (solo indicado)

Como $\gamma^0\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ \sigma_k & 0 \end{pmatrix}$, basta cambiar
$\sigma_3$ por $\hat{\bf n}\cdot\vec\sigma$:

$$
\boxed{
S(\eta\,\hat{\bf n}) = e^{\frac{\eta}{2}\,\gamma^0(\hat{\bf n}\cdot\vec\gamma)}
= \begin{pmatrix} \cosh\frac{\eta}{2}\, I & \sinh\frac{\eta}{2}\,(\hat{\bf n}\cdot\vec\sigma) \\ \sinh\frac{\eta}{2}\,(\hat{\bf n}\cdot\vec\sigma) & \cosh\frac{\eta}{2}\, I \end{pmatrix}
}
$$

Sobre el espinor en reposo, con cualquier $\chi$, da la solución general de la sección 2.5 con
${\bf p} = \mathrm{p}\,\hat{\bf n}$:

$$
S(\eta\,\hat{\bf n})\, \sqrt{2m} \begin{pmatrix} \chi \\ 0 \end{pmatrix}
= N \begin{pmatrix} \chi \\ \kappa\,(\hat{\bf n}\cdot\vec\sigma)\,\chi \end{pmatrix}
= N \begin{pmatrix} \chi \\ \dfrac{\vec\sigma\cdot{\bf p}}{E+m}\,\chi \end{pmatrix} .
$$

El *boost* no toca el bloque de arriba: el espín es el que tenía en reposo. Si en reposo apunta en
$z$ y el *boost* va en otra dirección, el resultado no es autoestado de helicidad; si ya apunta en
$\hat{\bf n}$, $\chi = \chi_\pm(\hat{\bf n})$, el resultado es $u_\pm(\mathrm{p}\,\hat{\bf n})$.

## 4. ¿Qué se construye con los espinores?

$M_{fi}$ es un invariante Lorentz, y se construye con espinores. Los espinores no lo son: se
transforman con $S(\Lambda)$. Esta sección demuestra que $\bar\Psi\Psi$ es un **escalar** y
$\bar\Psi\gamma^\mu\Psi$ un **cuadrivector** (y lo mismo con $\gamma^5$), que la corriente **se conserva**, y cómo se lee como
**corriente eléctrica**.

### 4.1. El adjunto y cómo se transforma

$\Psi^\dagger\Psi$ no sirve: $\Psi'^\dagger\Psi' = \Psi^\dagger S^\dagger S\,\Psi$, y $S$ no es unitaria para los *boosts*
(en reposo $u^\dagger u = 2m$; en movimiento, $2E$). Lo que sí se cumple, para cualquier $S(\Lambda)$, es

$$
\boxed{
\gamma^0\, S^\dagger\, \gamma^0 = S^{-1}
}
$$

* **Rotación.** $S(\phi\,\hat{z}) = \cos\frac{\phi}{2}\,I - i\sin\frac{\phi}{2}\,\Sigma_3$. $\Sigma_3$ es hermítica y diagonal
  por bloques, así que conmuta con $\gamma^0$ y $S^\dagger = \cos\frac{\phi}{2}\,I + i\sin\frac{\phi}{2}\,\Sigma_3 = S^{-1}$
  (es unitaria). Entonces $\gamma^0 S^\dagger \gamma^0 = S^\dagger (\gamma^0)^2 = S^{-1}$.
* ***Boost*.** $S(\eta\,\hat{z}) = \cosh\frac{\eta}{2}\,I + \sinh\frac{\eta}{2}\,\gamma^0\gamma^3$. Como
  $(\gamma^0\gamma^3)^\dagger = \gamma^{3\dagger}\gamma^{0\dagger} = -\gamma^3\gamma^0 = \gamma^0\gamma^3$, es hermítica: $S^\dagger = S$ (**no** es
  unitaria). Pero $\gamma^0$ anticonmuta con $\gamma^0\gamma^3$, y
  $\gamma^0 S^\dagger \gamma^0 = \cosh\frac{\eta}{2}\,I - \sinh\frac{\eta}{2}\,\gamma^0\gamma^3 = S(-\eta\,\hat{z}) = S^{-1}$.

En otra dirección es igual ($\Sigma_k$ y $\gamma^0\gamma^k$ tienen las mismas propiedades), y cualquier
transformación es un producto de rotaciones y *boosts*. Con esto, el **adjunto**
$\bar\Psi \equiv \Psi^\dagger\gamma^0$ se transforma con la inversa:

$$
\bar\Psi' = \Psi^\dagger S^\dagger \gamma^0 = \Psi^\dagger\gamma^0\,(\gamma^0 S^\dagger \gamma^0) = \bar\Psi\, S^{-1} .
$$

(Hemos metido $(\gamma^0)^2 = I$ entre $\Psi^\dagger$ y $S^\dagger$.)

### 4.2. $\bar\Psi\Psi$ es un escalar

$$
\bar\Psi'\Psi' = \bar\Psi\, S^{-1} S\, \Psi = \bar\Psi\Psi
\qquad \Rightarrow \qquad
\boxed{\bar\Psi\Psi \;\text{ es invariante Lorentz}}
$$

La fase se cancela ($e^{+ip_\mu x^\mu}e^{-ip_\mu x^\mu} = 1$), así que para las ondas planas basta con los espinores.
**Comprobación** con $u_1$ en $z$: $\bar u_1 = u_1^\dagger\gamma^0 = N\,(1, 0, -\kappa, 0)$, y

$$
\bar u_1 u_1 = N^2(1 - \kappa^2) = (E+m) - \frac{\mathrm{p}^2}{E+m} = \frac{(E+m)^2 - \mathrm{p}^2}{E+m} = 2m ,
$$

lo mismo que en reposo, $\bar u_1(m)\, u_1(m) = 2m$. En cambio $u_1^\dagger u_1 = N^2(1 + \kappa^2) = 2E$. La
diferencia es el signo menos del bloque de abajo, que pone $\gamma^0$. Con un espinor genérico,
$\bar u_s u_s = N^2\left(1 - \frac{{\bf p}^2}{(E+m)^2}\right) = 2m$, y para la antipartícula $\bar v_s v_s = -2m$.

### 4.3. $\bar\Psi\gamma^\mu\Psi$ es un cuadrivector

Con la condición de la sección 3.1, $S^{-1}\gamma^\mu S = \Lambda^\mu{}_\nu\gamma^\nu$:

$$
\bar\Psi'\gamma^\mu\Psi' = \bar\Psi\, S^{-1}\gamma^\mu S\, \Psi = \Lambda^\mu{}_\nu\, \bar\Psi\gamma^\nu\Psi
\qquad \Rightarrow \qquad
\boxed{j^\mu \equiv \bar\Psi\gamma^\mu\Psi \;\text{ es un cuadrivector}}
$$

Las $\gamma^\mu$ no cambian; lo que se transforma son los espinores, y el resultado es el de un
cuadrivector.

**Comprobación** con el *boost* en $z$. En reposo, $\bar u\gamma^0 u = u^\dagger u = 2m$ y
$\bar u\gamma^k u = u^\dagger\gamma^0\gamma^k u = 0$ ($\gamma^0\gamma^k$ solo tiene bloques fuera de la diagonal, y el de
abajo es cero): $j^\mu = (2m, 0, 0, 0)$. El *boost* de rapidez $\eta$ la lleva a

$$
\Lambda^\mu{}_\nu\, j^\nu = (2m\cosh\eta, 0, 0, 2m\sinh\eta) = (2E, 0, 0, 2\mathrm{p}) = 2p^\mu ,
$$

que es lo que da el cálculo directo con $u_s(p)$ (taller de perspectivas): $j^0 = N^2(1 + \kappa^2) = 2E$ y
$j^3 = 2N^2\kappa = 2\mathrm{p}$. Con un momento cualquiera, $\bar u_s\gamma^\mu u_s = 2p^\mu$.

### 4.4. El pseudoescalar y el axial

Como $\gamma^5$ conmuta con $S$ (sección 1.7), metiendo $S S^{-1} = I$ donde haga falta:

$$
\bar\Psi'\gamma^5\Psi' = \bar\Psi\, S^{-1}\gamma^5 S\, \Psi = \bar\Psi\gamma^5\Psi,
\qquad
\bar\Psi'\gamma^\mu\gamma^5\Psi' = \bar\Psi\, (S^{-1}\gamma^\mu S)(S^{-1}\gamma^5 S)\, \Psi = \Lambda^\mu{}_\nu\, \bar\Psi\gamma^\nu\gamma^5\Psi .
$$

$\bar\Psi\gamma^5\Psi$ es invariante y $\bar\Psi\gamma^\mu\gamma^5\Psi$ un cuadrivector, como S y V. Lo que los distingue
de ellos es la **paridad** (el *pseudo* y el *axial*), que se ve en el tema de leptones.

**Valores para $u_\pm$ con ${\bf p}$ en $z$** (el caso $f = i$). En Pauli-Dirac,
$\gamma^0\gamma^5 = \begin{pmatrix} 0 & I \\ -I & 0 \end{pmatrix}$ y $\gamma^0\gamma^k\gamma^5 = \Sigma_k$. Con $u = (u^a, u^b)$:

* **Pseudoescalar**: $\bar u\gamma^5 u = u^{a\dagger}u^b - u^{b\dagger}u^a = 0$, porque $\kappa$ es real. Se anula
  para una partícula libre; solo sobrevive en una transición, $\bar u_f\gamma^5 u_i$.
* **Axial, parte temporal**: $\bar u\gamma^0\gamma^5 u = u^\dagger\gamma^5 u = 2\,\mathrm{Re}(u^{a\dagger}u^b) = \pm 2N^2\kappa = \pm 2\mathrm{p}$.
* **Axial, parte espacial**: $\bar u\gamma^k\gamma^5 u = u^\dagger\Sigma_k u$. Solo $\Sigma_3$ sobrevive, y
  $u_\pm^\dagger\Sigma_3 u_\pm = \pm N^2(1 + \kappa^2) = \pm 2E$.

$$
\bar u_\pm\gamma^\mu\gamma^5 u_\pm = \pm 2\,(\mathrm{p}, 0, 0, E) = \pm 2m\, s^\mu,
\qquad s^\mu = \left(\frac{\mathrm{p}}{m}, 0, 0, \frac{E}{m}\right) .
$$

$s^\mu$ es el **cuadrivector de espín**: el *boost* de $(0, 0, 0, 1)$, el espín en $z$ en reposo, porque
$\Lambda\,(0, 0, 0, 1) = (\sinh\eta, 0, 0, \cosh\eta)$. Cumple $s_\mu p^\mu = 0$ y $s_\mu s^\mu = -1$. La corriente
axial lleva el espín, y por eso cambia de signo entre $u_+$ y $u_-$; la vectorial, $2p^\mu$, no.

### 4.5. La corriente se conserva

**La ecuación del adjunto.** Tomando el hermítico conjugado de $i\gamma^\mu\partial_\mu\Psi - m\Psi = 0$:

$$
-i\,\partial_\mu\Psi^\dagger\,\gamma^{\mu\dagger} - m\,\Psi^\dagger = 0 .
$$

Multiplicando por $\gamma^0$ por la derecha y usando $\gamma^{\mu\dagger}\gamma^0 = \gamma^0\gamma^\mu$ (es decir,
$\gamma^0\gamma^{\mu\dagger}\gamma^0 = \gamma^\mu$, que se comprueba con las matrices de Pauli-Dirac):

$$
\boxed{
i\,\partial_\mu\bar\Psi\,\gamma^\mu + m\,\bar\Psi = 0
}
$$

**La conservación.** Derivando el producto y usando las dos ecuaciones,
$\gamma^\mu\partial_\mu\Psi = -im\,\Psi$ y $\partial_\mu\bar\Psi\,\gamma^\mu = +im\,\bar\Psi$:

$$
\partial_\mu j^\mu = (\partial_\mu\bar\Psi)\,\gamma^\mu\Psi + \bar\Psi\gamma^\mu(\partial_\mu\Psi)
= im\,\bar\Psi\Psi - im\,\bar\Psi\Psi = 0 .
$$

Con $j^\mu = (\rho, {\bf j})$, es la **ecuación de continuidad**:

$$
\partial_t\rho + \nabla\cdot{\bf j} = 0, \qquad \rho = j^0 = \Psi^\dagger\Psi \geq 0 .
$$

Integrando en un volumen, lo que disminuye dentro es lo que sale por la superficie, y
$Q = \int \rho\,\mathrm{d}^3x$ se conserva. A diferencia de Klein-Gordon (sección 1.1), la densidad es
**siempre positiva**: era uno de los objetivos de Dirac.

### 4.6. La corriente eléctrica

Multiplicada por la carga $q$ del fermión ($q = -e$ para el electrón), la corriente conservada es la
**corriente eléctrica**, y su conservación es la de la carga:

$$
j^\mu_{em} = q\,\bar\Psi\gamma^\mu\Psi = (q\rho,\, q\,{\bf j}) .
$$

Para la onda plana, $j^\mu = 2p^\mu$:

$$
\rho = 2E, \qquad {\bf j} = 2{\bf p} = \rho\,\frac{{\bf p}}{E} = \rho\,{\bf v} ,
$$

densidad por velocidad, como una corriente clásica. Hay $2E$ partículas por unidad de volumen (la
normalización) moviéndose con ${\bf v} = {\bf p}/E$.

**La antipartícula.** $\bar v_s\gamma^\mu v_s = 2p^\mu$ también, pero su carga es la opuesta: su corriente
eléctrica es $-q\,2p^\mu$. En la teoría de una partícula eso lo da la regla de Feynman-Stückelberg
(invertir las cargas en $\Phi_s$); en teoría cuántica de campos sale directamente.

Es a esta corriente a la que se acopla el fotón, como **corriente por campo**, $q\,j^\mu A_\mu$: la
simetría *gauge* de [perspectivas](./perspectivas.ipynb). [MT4.3, MT4.5.1]

## 5. La simetría *gauge* local y el fotón

En [perspectivas](./perspectivas.ipynb) se afirma que la interacción del fermión con el fotón **sale**
de exigir que la física no cambie con una fase local. Aquí está la demostración.

### 5.1. Fase global y fase local

**Global.** Con una fase constante, $\Psi' = e^{iq\theta}\,\Psi$, la ecuación de Dirac no cambia: la fase sale
de la derivada y se cancela. Tampoco cambian los bilineales, $\bar\Psi' = \bar\Psi\,e^{-iq\theta}$, y
$\bar\Psi'\Gamma\Psi' = \bar\Psi\Gamma\Psi$. La fase global de $\Psi$ no es observable.

**Local.** Con una fase que depende del punto, $\theta = \theta(x)$,

$$
\Psi'(x) = e^{iq\theta(x)}\,\Psi(x),
$$

la derivada actúa también sobre la fase, $\partial_\mu\Psi' = e^{iq\theta}\left(\partial_\mu\Psi + iq\,(\partial_\mu\theta)\,\Psi\right)$, y
sobra un término:

$$
(i\gamma^\mu\partial_\mu - m)\,\Psi' = e^{iq\theta}\left[(i\gamma^\mu\partial_\mu - m)\,\Psi - q\,\gamma^\mu(\partial_\mu\theta)\,\Psi\right]
= -q\,e^{iq\theta}\,\gamma^\mu(\partial_\mu\theta)\,\Psi \neq 0 .
$$

Si $\Psi$ es solución, $\Psi'$ ya no lo es: la ecuación libre **no** es invariante bajo una fase local.

### 5.2. La derivada covariante

Para absorber el término que sobra hace falta un campo vectorial, $A_\mu(x)$, que se transforme a la vez
que $\Psi$ y compense $\partial_\mu\theta$:

$$
\boxed{
D_\mu \equiv \partial_\mu + iq\,A_\mu(x), \qquad
\Psi' = e^{iq\theta}\,\Psi, \qquad A'_\mu = A_\mu - \partial_\mu\theta
}
$$

**La demostración.** Con los campos transformados:

$$
D'_\mu\Psi' = \left(\partial_\mu + iq\,A'_\mu\right) e^{iq\theta}\,\Psi
= e^{iq\theta}\left(\partial_\mu\Psi + iq\,(\partial_\mu\theta)\,\Psi + iq\,A_\mu\Psi - iq\,(\partial_\mu\theta)\,\Psi\right)
= e^{iq\theta}\,D_\mu\Psi .
$$

$D_\mu\Psi$ se transforma **como $\Psi$**, con la misma fase: de ahí el nombre de *covariante*. Entonces

$$
(i\gamma^\mu D'_\mu - m)\,\Psi' = e^{iq\theta}\,(i\gamma^\mu D_\mu - m)\,\Psi = 0 ,
$$

y la ecuación $(i\gamma^\mu D_\mu - m)\,\Psi = 0$ es **invariante** bajo la fase local. Desarrollando $D_\mu$:

$$
(i\gamma^\mu\partial_\mu - m)\,\Psi = q\,\gamma^\mu A_\mu\,\Psi :
$$

el fermión libre a la izquierda y, a la derecha, el término **nuevo**: la interacción con $A_\mu$, con
intensidad $q$. La misma $A_\mu$ sirve para todos los fermiones, cada uno con su carga: por eso el
acoplo es **universal**.

### 5.3. El acoplamiento mínimo

Con $i\partial_\mu \leftrightarrow p_\mu$, la derivada covariante es $iD_\mu = i\partial_\mu - qA_\mu \leftrightarrow p_\mu - qA_\mu$. Con
$A^\mu = (\phi, {\bf A})$, es decir, $A_\mu = (\phi, -{\bf A})$:

$$
E \;\to\; E - q\phi, \qquad {\bf p} \;\to\; {\bf p} - q{\bf A} ,
$$

la **sustitución mínima** del electromagnetismo clásico. En forma hamiltoniana (multiplicando por $\gamma^0$,
como en la sección 1.4, con $\alpha_k = \gamma^0\gamma^k$):

$$
i\partial_t\Psi = \left[\vec\alpha\cdot(\hat{\bf p} - q{\bf A}) + \gamma^0 m + q\phi\right]\Psi ,
$$

el análogo relativista de $H = ({\bf p} - q{\bf A})^2/2m + q\phi$. La simetría local no ha añadido nada que no
supiéramos del electromagnetismo: lo ha **deducido**.

### 5.4. El campo del fotón

$A_\mu$ es un campo con su propia ecuación, que también debe ser invariante bajo
$A_\mu \to A_\mu - \partial_\mu\theta$.

**El tensor de campo es invariante.** Con $F_{\mu\nu} \equiv \partial_\mu A_\nu - \partial_\nu A_\mu$ (sus componentes son
${\bf E}$ y ${\bf B}$):

$$
F'_{\mu\nu} = F_{\mu\nu} - \partial_\mu\partial_\nu\theta + \partial_\nu\partial_\mu\theta = F_{\mu\nu} .
$$

Los campos eléctrico y magnético no cambian: lo que cambia, $A_\mu$, no es observable.

**Las ecuaciones de Maxwell**, con la corriente del fermión como fuente,

$$
\boxed{
\partial_\mu F^{\mu\nu} = q\,\bar\Psi\gamma^\nu\Psi
}
$$

son invariantes: $F^{\mu\nu}$ no cambia, y la corriente tampoco ($\bar\Psi'\gamma^\nu\Psi' = \bar\Psi\gamma^\nu\Psi$). Sin fuente, es
la ecuación del fotón libre de la tabla de perspectivas.

**La conservación es obligatoria.** Derivando otra vez, $\partial_\nu\partial_\mu F^{\mu\nu} = 0$, porque $F^{\mu\nu}$ es
antisimétrico y $\partial_\nu\partial_\mu$ simétrico. Luego $\partial_\nu\,(q\,\bar\Psi\gamma^\nu\Psi) = 0$: Maxwell solo es consistente
si la corriente se conserva, y la de Dirac se conserva (sección 4.5). La carga que se conserva es la
de la simetría.

**El fotón no tiene masa.** Una masa añadiría un término $m_\gamma^2 A^\nu$ a la ecuación,
$\partial_\mu F^{\mu\nu} + m_\gamma^2 A^\nu = q\,\bar\Psi\gamma^\nu\Psi$. Bajo la transformación cambia en
$-m_\gamma^2\,\partial^\nu\theta$, que no se compensa con nada: la simetría local exige $m_\gamma = 0$. [MT10.1]

1. Comprueba que con $A'_\mu = A_\mu + \partial_\mu\theta$ (el otro signo) $D_\mu\Psi$ **no** se transforma como $\Psi$.
2. Comprueba que $\bar\Psi'\gamma^\mu\Psi' = \bar\Psi\gamma^\mu\Psi$ con la fase local. ¿Y $\bar\Psi'\gamma^\mu\partial_\nu\Psi'$?
3. Deduce la forma hamiltoniana de la sección 5.3 a partir de $(i\gamma^\mu D_\mu - m)\,\Psi = 0$.
4. Un protón y un electrón en el mismo campo $A_\mu$: ¿cómo se transforman sus campos bajo la misma
   $\theta(x)$? ¿Impide la simetría que sus cargas fuesen $+e$ y $-1.0001\,e$? ¿Qué dice el experimento?
5. Escribe $F^{0k}$ y $F^{ij}$ en función de ${\bf E} = -\nabla\phi - \partial_t{\bf A}$ y ${\bf B} = \nabla\times{\bf A}$, y comprueba que
   $\partial_\mu F^{\mu 0} = q\,\bar\Psi\gamma^0\Psi$ es la ley de Gauss, $\nabla\cdot{\bf E} = \rho$.
