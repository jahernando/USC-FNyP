## Sobre la ecuación de Dirac

Dirac propuso en 1928 una ecuación que describe la dinámica de los fermiones libres.

Construyó su ecuación de tal forma que fuera invariente Lorentz, apareciesen linealmente los operadores de energía y momento de la mecánica cuántica,  $\hat{E} = i \partial_0, \; \hat{p} = -i \nabla$, y que contuviese implícitamente la ecuación de Einstein, $E^2 = m^2 + p^2$.

La solución de la ecuación de Dirac, $\Psi(x)$, es un nuevo tipo de objeto físico, un **espinor**, que describe los fermiones libres, partículas con un momento angular intrínseco o espín de $1/2$. $\Psi(x)$ no es sólo una función de ondas, si no cuatro, puestas en forma de columna. Este nuevo objeto se transformará con una reglas específicas bajo transformaciones Lorentz, que veremos, se derivan de la invariancia Lorentz de la propia ecuación de Dirac. En particular un espinor bajo rotaciones regresa a su forma original tras girar dos veces, $4\pi$, esto es, un giro le afecta "la mitad", de ahí que su *espín* sea 1/2. 

Dos de las soluciones de Dirac tenían energía negativa. Dirac mismo propuso una interpretación muy iluminadora de su significado, indicando que las soluciones de energía negativa se correspondían con anti-partículas. Anti-partícula sería la *ausencia* de partícula, y por lo tanto tendría sus cargas opuestas. El descubrimiento del positrón supuso el espaldarazo de la existencia de las anti-partículas y de la ecuación de Dirac. 

Las cuatro soluciones indenpendientes de Dirac corresponden a un fermión y sus dos componentes de espín, por ejemplo el electrón con carga $-e$, y al anti-fermión, por ejemplo, el positrón, que tiene la cargas opuesta, $+e$, y sus dos componentes de espín. La ecuación de Dirac introduce de forma natural el **espín**, el momento angular intrínsico de una partícula, y a las **antipartículas**. ¡Todo un logro!

En la ecuación aparecen ademas cuatro matrices, $\gamma^\mu$ con $\mu = 0, 1, 2, 3$, iguales en todos los sistemas inerciales, que aunque admiten diversas representaciones, cumplen unas reglas de conmutación que permiten como un engranaje, al asociarlas al momento y la energía de la partícula, recuperar de nuevo la relación $E^2 = m^2 + p^2$. Podemos decir que las matrices $\gamma$ asociadas con los operadores energía y momento están codificando, "encriptando", linealmente la famosa ecuación de Einstein.

La ecuación de Dirac es uno de los grandes logros de la Física Teórica, matemáticamente es de una simpleza sorprendente y al mismo tiempo de una riqueza física desbordante.

En lo siguiente desvelaremos algunos de sus secretos e intentaremos asociarlos a los elementos fundamentales de las interacciones.

### Notación

* Unidades naturales, $\hbar = c = 1$. Métrica $g_{\mu\nu} = \mathrm{diag}(1, -1, -1, -1)$.
* Índices griegos $\mu, \nu = 0, 1, 2, 3$; latinos $i, j, k = 1, 2, 3$. Un índice repetido se suma.
* $x^\mu = (t, {\bf x})$, $p^\mu = (E, {\bf p})$, $p_\mu = (E, -{\bf p})$ y
  $p_\mu x^\mu = Et - {\bf p} \cdot {\bf x}$. Las componentes físicas del momento son $p^k$.
* ${\bf p} = \mathrm{p} \, \hat{\bf v}$: $\mathrm{p} = |{\bf p}|$ es el módulo y $\hat{\bf v}$ la dirección.
* Operadores de la mecánica cuántica: $\hat{H} = i\partial_t$, $\hat{\bf p} = -i\nabla$, esto es,
  $\hat{p}^\mu = i\partial^\mu$.
* Con ${\bf p}$ en $z$: $p^\mu = (E, 0, 0, \mathrm{p})$, y los dos números que aparecerán en todas partes,

$$
\kappa \equiv \frac{\mathrm{p}}{E+m}, \qquad N \equiv \sqrt{E+m}.
$$

* $I$ es la identidad ($2\times2$ o $4\times4$, según el contexto); $\sigma_k$, las matrices de Pauli.
* Espinores: $u_s(p)$ (fermión) y $v_s(p)$ (antifermión), $s = 1, 2$; sus bloques $2\times1$
  de arriba y abajo llevan superíndice $a$, $b$. Bi-espinores base: $\chi_s$ y $\xi_s$. Con
  ${\bf p}$ en $z$ también $u_\pm$, $v_\pm$ (helicidad).
* Espín: ${\bf S} = \frac{1}{2}{\bf \Sigma}$. Transformación de Lorentz de un espinor: la matriz
  $S(\Lambda)$ (no confundir con el espín).
* Rapidez de un *boost*: $\eta$. Ángulo de una rotación en torno a $z$: $\phi$.

## 1. ¿Cómo se construye la ecuación de Dirac?

**Punto de partida.** En mecánica cuántica la energía y el momento son operadores:

$$
\hat{H} = i\partial_t, \qquad \hat{\bf p} = -i\nabla.
$$

Si se sustituyen en $E^2 = {\bf p}^2 + m^2$ se obtiene la ecuación de Klein-Gordon,
$-\partial^2_t \Psi = (-\nabla^2 + m^2)\Psi$. Es de **segundo orden** en el tiempo, y su densidad
de probabilidad puede ser negativa. [MT4.1]

**La idea de Dirac:**

* primer orden en $\partial_t$, como la ecuación de Schrödinger;
* la relatividad trata $t$ y ${\bf x}$ por igual: primer orden también en $\nabla$;
* la ecuación de Einstein, $E^2 = {\bf p}^2 + m^2$, debe seguir cumpliéndose.

La ecuación más general **lineal** en las derivadas es:

$$
(i\gamma^0 \partial_t + i\gamma^k \partial_k - m) \, \Psi = 0,
$$

con cuatro coeficientes $\gamma^\mu$ que no conocemos (el de $m$ se toma igual a la unidad sin
pérdida de generalidad).

**Paso 1: de derivadas a energía y momento.** Con los operadores, $i\partial_t \to E$ e
$i\partial_k \to -p^k$:

$$
(\gamma^0 E - \gamma^k p^k) \, \Psi = m \, \Psi,
$$

que en forma covariante es $(\gamma^\mu p_\mu - m)\Psi = 0$.

**Paso 2: elevar al cuadrado.** Aplicando dos veces el operador, $(\gamma^0 E - \gamma^k p^k)^2 \Psi = m^2 \Psi$.
Desarrollamos sin cambiar el orden de las $\gamma$, porque no sabemos si conmutan:

$$
(\gamma^0 E - \gamma^k p^k)^2 = (\gamma^0)^2 E^2 + \sum_k (\gamma^k)^2 (p^k)^2
+ \sum_{i<k} \{\gamma^i, \gamma^k\} \, p^i p^k - \sum_k \{\gamma^0, \gamma^k\} \, E \, p^k
$$

**Paso 3: exigir la ecuación de Einstein.** Para que esto sea $E^2 - {\bf p}^2$ para cualquier
$E$ y ${\bf p}$, término a término:

$$
(\gamma^0)^2 = I, \qquad (\gamma^k)^2 = -I, \qquad \{\gamma^0, \gamma^k\} = 0, \qquad
\{\gamma^i, \gamma^k\} = 0 \;\; (i \neq k),
$$

que se resume en:

$$
\boxed{
\{\gamma^\mu, \gamma^\nu\} \equiv \gamma^\mu \gamma^\nu + \gamma^\nu \gamma^\mu = 2 \, g^{\mu\nu} \, I
}
$$

Las $\gamma$ **codifican linealmente** la ecuación de Einstein.

**Paso 4: las $\gamma^\mu$ son matrices y $\Psi$ tiene cuatro componentes.** Dos números no nulos
no pueden anticonmutar: las $\gamma^\mu$ son matrices, y $\Psi$ es una columna sobre la que actúan.
¿De qué dimensión?

* Hacen falta **cuatro** matrices que anticonmuten dos a dos. Con $2\times2$ solo hay tres, las
  de Pauli.
* Las $\gamma^\mu$ tienen traza nula y autovalores $\pm 1$ (o $\pm i$), luego su dimensión es par.

La dimensión mínima es $4$: $\Psi$ tiene cuatro componentes y es un **espinor de Dirac**.

**Paso 5: hermiticidad.** Multiplicando el paso 1 por $\gamma^0$ por la izquierda se obtiene la
forma hamiltoniana, $E \, \Psi = H \, \Psi$, con

$$
H = \gamma^0 \gamma^k \, p^k + \gamma^0 \, m.
$$

$H$ debe ser hermítico (sus autovalores, las energías, son reales). Luego
$(\gamma^0)^\dagger = \gamma^0$ y $(\gamma^0\gamma^k)^\dagger = (\gamma^k)^\dagger \gamma^0 = \gamma^0\gamma^k = -\gamma^k\gamma^0$, es decir:

$$
\boxed{
(\gamma^0)^\dagger = \gamma^0, \qquad (\gamma^k)^\dagger = -\gamma^k
}
$$

**Paso 6: una representación.** Solo el álgebra está fijada; cualquier conjunto de matrices que
la cumpla sirve, y la física no depende de la elección. La habitual, y la que usaremos, es la
**representación de Pauli-Dirac**, en bloques $2\times2$:

$$
\boxed{
\gamma^0 = \begin{pmatrix} I & 0 \\ 0 & -I \end{pmatrix}, \qquad
\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ -\sigma_k & 0 \end{pmatrix}
}
$$

con las matrices de Pauli

$$
\sigma_1 = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \;\;
\sigma_2 = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \;\;
\sigma_3 = \begin{pmatrix} 1 & 0 \\ 0 & -1\end{pmatrix}, \qquad
\{\sigma_i, \sigma_j\} = 2 \, \delta_{ij} \, I, \;\; [\sigma_i, \sigma_j] = 2i \, \epsilon_{ijk} \sigma_k.
$$

**Resultado.** Volviendo a las derivadas:

$$
\boxed{
(i \gamma^\mu \partial_\mu - m) \, \Psi(x) = 0
}
$$

donde $\partial_\mu \equiv \partial/\partial x^\mu = (\partial_t, \nabla)$ es el **cuadrigradiente**.
*Nota:* no confundir con la *derivada covariante* $D_\mu = \partial_\mu + i q A_\mu$, que aparece al
imponer la simetría *gauge*.

El precio de la linealidad —matrices y cuatro componentes— es lo que nos va a dar el **espín** y
las **antipartículas**. [MT4.2]

*Cuestión:* comprueba que las $\gamma^\mu$ de Pauli-Dirac cumplen el álgebra y la hermiticidad.

### ¿Qué son las matrices $\gamma$?

**Como el número $i$.** Con números reales, $x^2 + y^2$ no se puede escribir como producto de
dos factores lineales. Hace falta un número nuevo, $i$, con $i^2 = -1$:

$$
x^2 + y^2 = (x + iy)(x - iy).
$$

Dirac necesita lo mismo con $E^2 - {\bf p}^2$, una suma de cuatro cuadrados con signos, y ningún
número sirve. Hacen falta cuatro objetos nuevos que:

* al elevarlos al cuadrado den el signo de la métrica: $(\gamma^0)^2 = I$, $(\gamma^k)^2 = -I$;
* **anticonmuten** entre sí, para que en el cuadrado se cancelen los términos cruzados
  ($E\,p^k$, $p^i p^k$).

Esos objetos son las $\gamma^\mu$, y resultan ser matrices. En tres dimensiones ya lo conocemos
por el espín en mecánica cuántica: $(\vec{\sigma} \cdot {\bf p})^2 = {\bf p}^2$, es decir,
$\vec{\sigma} \cdot {\bf p}$ es la «raíz cuadrada» de ${\bf p}^2$. Dirac hace lo mismo en el
espacio-tiempo, con cuatro matrices en lugar de tres:

$$
(\gamma^\mu p_\mu)^2 = p_\mu p^\mu = E^2 - {\bf p}^2 = m^2 .
$$

**La métrica escrita con matrices.** El álgebra

$$
\{\gamma^\mu, \gamma^\nu\} = 2 \, g^{\mu\nu} \, I
$$

dice que las $\gamma^\mu$ se comportan como cuatro **vectores unitarios del espacio-tiempo**:
el temporal de cuadrado $+1$, los espaciales de cuadrado $-1$, y «ortogonales» entre sí (el
anticonmutador de dos distintas es cero). Son la métrica de Minkowski, $g^{\mu\nu}$, escrita con
matrices. Por eso son las mismas en todos los sistemas inerciales, y por eso $\gamma^\mu p_\mu$,
el cuadrimomento contraído con ellas, tiene como cuadrado el invariante $m^2$.

## 2. ¿Cuáles son las soluciones? Fermión de masa $m$ que se mueve en $z$

Por simplicidad tomamos el momento lineal en la dirección $z$, ${\bf p} = \mathrm{p}\,\hat{z}$: de
las tres $\gamma^k$ solo sobrevive $\gamma^3$, y por eso en todo lo que sigue aparece $\sigma_3$. El
caso general se obtiene cambiando $\sigma_3$ por $(\hat{\bf v}\cdot\vec{\sigma})$ (sección 5).

**Ansatz: onda plana.** Probamos una columna de cuatro constantes, $w$, por una onda plana:

$$
\Psi(x) = w \, e^{-i p_\mu x^\mu} = w \, e^{-i(Et - \mathrm{p} z)}.
$$

Como $i\partial_\mu \Psi = p_\mu \Psi$, la ecuación de Dirac se vuelve **algebraica**:

$$
(\gamma^\mu p_\mu - m) \, w = 0, \qquad \gamma^\mu p_\mu = \gamma^0 E - \gamma^3 \mathrm{p}.
$$

En bloques $2\times2$ (representación de Pauli-Dirac), cada término es:

$$
\gamma^0 E = \begin{pmatrix} E \, I & 0 \\ 0 & -E \, I \end{pmatrix}, \qquad
-\gamma^3 \mathrm{p} = \begin{pmatrix} 0 & -\mathrm{p} \, \sigma_3 \\ \mathrm{p} \, \sigma_3 & 0 \end{pmatrix}, \qquad
-m = \begin{pmatrix} -m \, I & 0 \\ 0 & -m \, I \end{pmatrix},
$$

y sumando, con $w = (w^a, w^b)$ separado en sus dos bloques de dos componentes:

$$
\begin{pmatrix} (E-m) \, I & -\mathrm{p} \, \sigma_3 \\ \mathrm{p} \, \sigma_3 & -(E+m) \, I \end{pmatrix}
\begin{pmatrix} w^a \\ w^b \end{pmatrix} = 0 .
$$

Cada fila de bloques es una ecuación; quedan **dos ecuaciones acopladas** entre $w^a$ y $w^b$:

$$
(E - m) \, w^a - \mathrm{p} \, \sigma_3 \, w^b = 0, \qquad
\mathrm{p} \, \sigma_3 \, w^a - (E + m) \, w^b = 0 .
$$

**Los dos signos de la energía.** $(\gamma^\mu p_\mu - m) \, w = 0$ es un sistema lineal
**homogéneo** de cuatro ecuaciones. Si la matriz tiene determinante no nulo, es invertible y la
única solución es $w = 0$: no hay onda. Para que exista una solución $w \neq 0$ el determinante
tiene que **anularse**.

Como $\sigma_3 = \mathrm{diag}(1, -1)$, el sistema se separa en dos sistemas $2\times2$
independientes, uno para las componentes $(w_1, w_3)$ y otro para $(w_2, w_4)$:

$$
\begin{pmatrix} E-m & -\mathrm{p} \\ \mathrm{p} & -(E+m) \end{pmatrix}
\begin{pmatrix} w_1 \\ w_3 \end{pmatrix} = 0, \qquad
\begin{pmatrix} E-m & +\mathrm{p} \\ -\mathrm{p} & -(E+m) \end{pmatrix}
\begin{pmatrix} w_2 \\ w_4 \end{pmatrix} = 0 .
$$

Los dos tienen el mismo determinante,

$$
-(E-m)(E+m) + \mathrm{p}^2 = -(E^2 - \mathrm{p}^2 - m^2),
$$

y el de la matriz $4\times4$ es su producto:

$$
\det(\gamma^\mu p_\mu - m) = (E^2 - \mathrm{p}^2 - m^2)^2 = 0
\quad \Rightarrow \quad E^2 = \mathrm{p}^2 + m^2 .
$$

**¡Es la ecuación de Einstein!** La onda plana solo es solución si $E$ y $\mathrm{p}$ son los de una
partícula de masa $m$. Pero la ecuación fija $E^2$, no $E$:

$$
E = \pm \sqrt{\mathrm{p}^2 + m^2},
$$

y cada raíz aparece dos veces (una por cada sistema $2\times2$, que serán los dos estados de
espín): **cuatro soluciones**, dos de energía positiva y dos de energía negativa.

Se ve también directamente en las ecuaciones acopladas: despejando
$w^b = \frac{\mathrm{p}}{E+m} \sigma_3 w^a$ de la segunda y sustituyendo en la primera,

$$
\left(E - m - \frac{\mathrm{p}^2}{E+m}\right) w^a = \frac{E^2 - \mathrm{p}^2 - m^2}{E+m} \, w^a = 0,
$$

que con $w^a \neq 0$ exige de nuevo $E^2 = \mathrm{p}^2 + m^2$.

Las de energía negativa se escriben cambiando $E \to -E$ y $\mathrm{p} \to -\mathrm{p}$ en la fase,
de forma que en las fórmulas $E$ sea siempre positiva:

$$
\boxed{
\Psi_s(x) = u_s(p) \, e^{-i p_\mu x^\mu}, \qquad \Phi_s(x) = v_s(p) \, e^{+i p_\mu x^\mu}, \qquad
E = +\sqrt{\mathrm{p}^2 + m^2}, \;\; s = 1, 2
}
$$

Los espinores cumplen:

$$
\boxed{
(\gamma^\mu p_\mu - m) \, u_s(p) = 0, \qquad (\gamma^\mu p_\mu + m) \, v_s(p) = 0
}
$$

Con los operadores, $\Psi_s$ da energía $+E$ y momento $+\mathrm{p}$; $\Phi_s$ da $-E$ y $-\mathrm{p}$.
$\Phi_s$ es la **antipartícula** (ver *¿Qué son las soluciones $\Phi_s$?*, más abajo). [MT4.6]

**Espinores $u_s$.** Con $w = u$ y $E > 0$, las dos ecuaciones acopladas son:

$$
(E - m) \, u^a = \mathrm{p} \, \sigma_3 \, u^b, \qquad (E + m) \, u^b = \mathrm{p} \, \sigma_3 \, u^a.
$$

De la segunda, $u^b = \kappa \, \sigma_3 \, u^a$. La primera no añade nada, porque
$\kappa \, \mathrm{p} = \mathrm{p}^2/(E+m) = E - m$.

El bloque de arriba es libre: tomamos los dos independientes,

$$
\chi_1 = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \qquad \chi_2 = \begin{pmatrix} 0 \\ 1 \end{pmatrix},
$$

y obtenemos:

$$
u_s(p) = N \begin{pmatrix} \chi_s \\ \kappa \, \sigma_3 \, \chi_s \end{pmatrix}, \qquad
u_1(p) = N \begin{pmatrix} 1 \\ 0 \\ \kappa \\ 0 \end{pmatrix}, \;\;
u_2(p) = N \begin{pmatrix} 0 \\ 1 \\ 0 \\ -\kappa \end{pmatrix}.
$$

**Espinores $v_s$.** Con $+m$ la matriz es
$\begin{pmatrix} (E+m) \, I & -\mathrm{p} \, \sigma_3 \\ \mathrm{p} \, \sigma_3 & -(E-m) \, I \end{pmatrix}$, y
ahora conviene despejar el bloque de arriba de la primera fila: $v^a = \kappa \, \sigma_3 \, v^b$.

El bloque libre es el de abajo. Tomamos

$$
\xi_1 = \begin{pmatrix} 0 \\ 1 \end{pmatrix}, \qquad \xi_2 = \begin{pmatrix} 1 \\ 0 \end{pmatrix},
$$

en orden cruzado, para que $s = 1$ sea espín arriba también en la antipartícula (ver sección 3):

$$
v_s(p) = N \begin{pmatrix} \kappa \, \sigma_3 \, \xi_s \\ \xi_s \end{pmatrix}, \qquad
v_1(p) = N \begin{pmatrix} 0 \\ -\kappa \\ 0 \\ 1 \end{pmatrix}, \;\;
v_2(p) = N \begin{pmatrix} \kappa \\ 0 \\ 1 \\ 0 \end{pmatrix}.
$$

**Normalización.** Se toma $u^\dagger_s u_s = v^\dagger_s v_s = 2E$ ($2E$ partículas por unidad de
volumen, un convenio que se mantiene bajo *boosts*):

$$
u^\dagger_s u_s = N^2 (1 + \kappa^2) = N^2 \, \frac{2E}{E+m} = 2E \quad \Rightarrow \quad N = \sqrt{E+m}.
$$

En la sección 4 veremos que este $N$ es justo el que da el *boost* del espinor en reposo.

**En reposo** ($\mathrm{p} = 0$, $\kappa = 0$, $N = \sqrt{2m}$) solo sobreviven las componentes de
arriba en $u$ y las de abajo en $v$:

$$
u_s(m) = \sqrt{2m} \begin{pmatrix} \chi_s \\ 0 \end{pmatrix}, \qquad
v_s(m) = \sqrt{2m} \begin{pmatrix} 0 \\ \xi_s \end{pmatrix}.
$$

Todo el movimiento está en $\kappa$:

| caso | $\kappa$ | qué queda |
|:--|:--:|:--|
| en reposo, $\mathrm{p} = 0$ | $0$ | un bloque: el espinor de Pauli de siempre |
| en movimiento, $m > 0$ | $0 < \kappa < 1$ | el *boost* "enciende" el otro bloque |
| sin masa, o $E \gg m$ | $\to 1$ | los dos bloques pesan igual |

### ¿Qué son las soluciones $\Phi_s$?

Con los operadores, $\Phi_s$ tiene energía y momento **negativos**:

$$
i \partial_0 \, \Phi_s = -E \, \Phi_s, \qquad -i \partial_k \, \Phi_s = -p^k \, \Phi_s .
$$

Hay dos lecturas clásicas.

#### El mar de Dirac

Dirac propuso una ingeniosa y iluminadora idea para explicar las soluciones de energía negativa, basada en que en el vacío todos los estados negativos están ocupados. Esta idea también se conoce como el mar de Dirac.

De tal forma que la energía mínima para crear un electrón, para hacerlo saltar de la energía negativa, es de $2m_e$. Al hacer saltar un electrón del mar, en éste queda un hueco, que se comporta como una partícula de carga positiva: el positrón.

Esta idea es muy útil en otras ramas de Física, por ejemplo aparece en los pares electrón/hueco de los semi-conductores.

Pero ¿qué sentido tiene que el mar sea infinitamente negativo, que el nivel inferior de energía no esté acotado?

#### Feynman y Stückelberg

Stückelberg (1941) y Feynman (1949) propusieron que las antipartículas son "partículas que van
hacia atrás en el tiempo".

La idea: en la fase solo aparecen los productos $Et$ y ${\bf p}\cdot{\bf x}$, y cambiar el signo de
$E$ es indistinguible de cambiar el de $t$,

$$
e^{-i(-E)\,t} = e^{+iEt} = e^{-iE\,(-t)} .
$$

Es una igualdad, no una transformación: la fase por sí sola no distingue una energía negativa que
avanza en el tiempo de un reloj que va hacia atrás. La fase de la partícula gira en sentido
horario; la de $\Phi_s$, en sentido antihorario, como el reloj de la partícula marchando hacia atrás.

| |
|:--:|
|<img src="./imgs/dirac_reloj_fase.png" width = 420 align="center">|
| la fase en el plano complejo: partícula (horario) y antipartícula (antihorario) |

Con la fase completa, $\Phi_s$ es una partícula con lo que dan los operadores, energía $-E$ y momento
$-{\bf p}$:

$$
e^{+i(Et - {\bf p}\cdot{\bf x})} = e^{-i\left((-E)\,t \, - \, (-{\bf p})\cdot{\bf x}\right)} .
$$

La propuesta de Feynman y Stückelberg: esa partícula de cuadrimomento $(-E, -{\bf p})$ se propaga
**hacia atrás en el tiempo**. Invertir el sentido de propagación invierte el flujo de todo el
cuadrimomento, y lo que se ve hacia delante en el tiempo es una **antipartícula** de energía $+E$ y
momento $+{\bf p}$:

$$
(-E, -{\bf p}) \text{ hacia atrás en el tiempo} \quad \equiv \quad (+E, +{\bf p}) \text{ hacia delante}.
$$

Lo que convierte esta equivalencia formal en física es el **proceso**: emitir una partícula de
energía $-E$ y momento $-{\bf p}$ hacia delante en el tiempo equivale a absorber una antipartícula
de energía $+E$ y momento $+{\bf p}$, y viceversa. Por eso en los diagramas de Feynman la flecha de
la línea de un antifermión apunta hacia atrás en el tiempo. [MT4.7.1]

*Regla práctica:* en esta visión, las cantidades físicas de la antipartícula se leen en $\Phi_s$ invirtiendo el signo de lo que dan los operadores. Equivale a usar sobre $\Phi_s$ los operadores:

$$
\hat{H}^{(v)} = - i \partial_0, \;\; \hat{\bf p}^{(v)} = + i \nabla, \;\; \hat{\bf S}^{(v)} = - \hat{\bf S}
$$

Se invierten la energía y el momento (todo el cuadrimomento, $p^\mu \to -p^\mu$), el momento angular (espín, orbital y total) y las cargas. La helicidad, ${\bf S} \cdot \hat{\bf p}$, no cambia, porque se invierten sus dos factores; la quiralidad sí, y de ahí que para los antifermiones se defina $v_R = P_L \, v$.

*Ojo:* no es la regla de «lo que cambia al invertir el tiempo»: la energía y la carga no cambian bajo inversión temporal y, sin embargo, aquí se invierten. [MT4.7.3]

La interpretación de las anti-partículas de la ecuación de Dirac es *controvertida*, y su sentido aparece de forma natural cuando se cuantifica el campo de Dirac, esto es, en teoría cuántica de campos.

Tanto en TCQ como experimentalmente, la antipartícula es igual que su partícula en espín, masa, vida media pero con las cargas opuestas.


## 3. ¿Qué espín y qué helicidad tienen las soluciones? (${\bf p}$ en $z$)

El operador **espín** es

$$
\boxed{
{\bf S} \equiv \frac{1}{2} {\bf \Sigma}, \qquad \Sigma_k \equiv \begin{pmatrix} \sigma_k & 0 \\ 0 & \sigma_k \end{pmatrix}
}
$$

Con ${\bf p}$ en $z$, $S_3$ conmuta con $H$ y los cuatro espinores son autoestados de $S_3$.
Como $\Sigma_3 = \mathrm{diag}(1, -1, 1, -1)$:

$$
S_3 \, u_1 = +\tfrac{1}{2} \, u_1, \qquad S_3 \, u_2 = -\tfrac{1}{2} \, u_2, \qquad
S_3 \, v_1 = -\tfrac{1}{2} \, v_1, \qquad S_3 \, v_2 = +\tfrac{1}{2} \, v_2 .
$$

En la antipartícula el operador da el valor **cambiado de signo**, igual que con $E$ y ${\bf p}$
(regla práctica de Feynman-Stückelberg): el espín físico de $v_1$ es $+\frac{1}{2}$ y el de $v_2$,
$-\frac{1}{2}$. De ahí el orden cruzado de los $\xi_s$.

La **helicidad** es la proyección del espín en la dirección del momento:

$$
\boxed{
h \equiv {\bf S} \cdot \hat{\bf v} = \frac{1}{2}
\begin{pmatrix} \hat{\bf v} \cdot \vec{\sigma} & 0 \\ 0 & \hat{\bf v} \cdot \vec{\sigma} \end{pmatrix}
}
$$

Con ${\bf p}$ en $z$, $h = S_3$. En la antipartícula se invierten a la vez el espín y el momento, y la
helicidad no cambia de signo. Así, los cuatro espinores son también autoestados de helicidad:

$$
u_+ = u_1, \quad u_- = u_2, \quad v_+ = v_1, \quad v_- = v_2 .
$$

$u_\pm(p), v_\pm(p)$ son los espinores de las partículas y antipartículas respectivamente.

$u_{+}(p), v_{+}(p)$ son espinores de **helicidad positiva** (el espín tiene el mismo sentido que el momento).

$u_{-}(p), v_{-}(p)$ tienen **helicidad negativa** (el espín y el momento tienen sentido opuesto).



### Resumen de las secciones 2 y 3

Para un fermión de masa $m$ con momento $\mathrm{p}$ en $z$, las cuatro soluciones son:

$$
\boxed{
\Psi_s(x) = \sqrt{E+m} \begin{pmatrix} \chi_s \\ \kappa \, \sigma_3 \, \chi_s \end{pmatrix} e^{-i(Et - \mathrm{p} z)}, \qquad
\Phi_s(x) = \sqrt{E+m} \begin{pmatrix} \kappa \, \sigma_3 \, \xi_s \\ \xi_s \end{pmatrix} e^{+i(Et - \mathrm{p} z)}
}
$$

| espinor | componentes $/N$ | espín físico en $z$ | helicidad |
|:--:|:--:|:--:|:--:|
| $u_1 = u_+$ | $(1, 0, \kappa, 0)$ | $+\frac{1}{2}$ | $+$ |
| $u_2 = u_-$ | $(0, 1, 0, -\kappa)$ | $-\frac{1}{2}$ | $-$ |
| $v_1 = v_+$ | $(0, -\kappa, 0, 1)$ | $+\frac{1}{2}$ | $+$ |
| $v_2 = v_-$ | $(\kappa, 0, 1, 0)$ | $-\frac{1}{2}$ | $-$ |

Para una dirección cualquiera, ${\bf p} = \mathrm{p} \, \hat{\bf v}$, basta cambiar $\sigma_3$ por
$(\hat{\bf v} \cdot \vec{\sigma})$ (sección 5).

## 4. ¿Cómo se rota un espinor y cómo se le hace un *boost*? (en $z$)

**La regla.** Bajo una transformación de Lorentz $\Lambda$ (rotación o *boost*), un cuadrivector
se transforma con $\Lambda$, $p^\mu \to \Lambda^\mu{}_\nu \, p^\nu$. Un espinor lo hace con una
matriz compleja $4\times4$, $S(\Lambda)$:

$$
\Psi'(x) = S(\Lambda) \, \Psi(\Lambda^{-1} x), \qquad
u(p) \;\to\; S(\Lambda) \, u(p) \;\; \text{(un espinor de momento } \Lambda p \text{)}.
$$

$S(\Lambda)$ queda fijada al exigir que la ecuación de Dirac tenga la misma forma en todos los
sistemas inerciales, $S^{-1}\gamma^\mu S = \Lambda^\mu{}_\nu \, \gamma^\nu$ (la deducción está en
[ext-trans-Lorentz](./ext-trans-Lorentz.ipynb)). En $z$:

* **rotación** de ángulo $\phi$ en torno a $z$ (activa, en sentido antihorario). El generador es el
  espín:

$$
\boxed{
S(\phi \, \hat{z}) = e^{-i \phi \, S_3} = \cos\frac{\phi}{2} \, I - i \sin\frac{\phi}{2} \, \Sigma_3
}
$$

* ***boost*** de rapidez $\eta$ en $z$:

$$
\boxed{
S(\eta \, \hat{z}) = e^{\frac{\eta}{2} \gamma^0 \gamma^3} = \cosh\frac{\eta}{2} \, I + \sinh\frac{\eta}{2} \, \gamma^0 \gamma^3
}
$$

La serie de la exponencial se suma porque $(\Sigma_3)^2 = I$ y $(\gamma^0\gamma^3)^2 = I$. Lo
importante: **los ángulos aparecen divididos por dos**, $\phi/2$ y $\eta/2$.

En la representación de Pauli-Dirac ($\gamma^0\gamma^3 = \begin{pmatrix} 0 & \sigma_3 \\ \sigma_3 & 0 \end{pmatrix}$):

$$
S(\phi \, \hat{z}) = \begin{pmatrix} e^{-i\frac{\phi}{2}\sigma_3} & 0 \\ 0 & e^{-i\frac{\phi}{2}\sigma_3} \end{pmatrix}, \qquad
S(\eta \, \hat{z}) = \begin{pmatrix} \cosh\frac{\eta}{2} \, I & \sinh\frac{\eta}{2} \, \sigma_3 \\ \sinh\frac{\eta}{2} \, \sigma_3 & \cosh\frac{\eta}{2} \, I \end{pmatrix}
$$

La rotación no mezcla los bloques de arriba y de abajo; el *boost* sí: es el *boost* el que
"enciende" el segundo bloque.

### 4.1. Rotación en torno a $z$ de los espinores $u_s$ con ${\bf p}$ en $z$

Una rotación en torno a $z$ no cambia ${\bf p} = \mathrm{p}\,\hat{z}$: el espinor rotado debe ser
una solución con el mismo momento. Como $\sigma_3$ es diagonal,
$e^{-i\frac{\phi}{2}\sigma_3} = \mathrm{diag}(e^{-i\phi/2}, e^{+i\phi/2})$ y

$$
S(\phi \, \hat{z}) = \mathrm{diag}\left(e^{-i\phi/2}, \, e^{+i\phi/2}, \, e^{-i\phi/2}, \, e^{+i\phi/2}\right).
$$

Sobre los espinores:

$$
S(\phi \, \hat{z}) \, u_1(p) = N \begin{pmatrix} e^{-i\phi/2} \\ 0 \\ \kappa \, e^{-i\phi/2} \\ 0 \end{pmatrix} = e^{-i\phi/2} \, u_1(p), \qquad
S(\phi \, \hat{z}) \, u_2(p) = e^{+i\phi/2} \, u_2(p),
$$

esto es,

$$
\boxed{
S(\phi \, \hat{z}) \, u_s(p) = e^{-i m_s \phi} \, u_s(p), \qquad m_s = \pm\tfrac{1}{2}
}
$$

**Qué se lee:**

* Rotar en torno al eje del espín no cambia el estado: solo le añade una fase, como a cualquier
  autoestado de $J_z$ en mecánica cuántica, $e^{-i\phi J_z}|j, m\rangle = e^{-im\phi}|j, m\rangle$.
* La fase va con **medio ángulo**: una vuelta, $\phi = 2\pi$, da $S = -I$ y $u_s \to -u_s$. Hacen
  falta **dos vueltas**, $4\pi$, para volver al espinor de partida. Una fase global no se ve, pero
  el signo sí se ha medido, haciendo interferir dos haces de neutrones de los que solo uno se
  rota. [Rauch et al., Phys. Lett. A 54 (1975) 425; Werner et al., PRL 35 (1975) 1053]
* En la antipartícula la fase va al revés: $S(\phi\,\hat{z}) \, v_1 = e^{+i\phi/2} \, v_1$, coherente
  con su espín físico $+\frac{1}{2}$ (el operador da el signo cambiado).

### 4.2. *Boost* en $z$ del espinor en reposo

Partimos de $u_s(m) = \sqrt{2m} \, (\chi_s, 0)$ y le hacemos un *boost* en $z$ de rapidez $\eta$,
la que lleva la partícula del reposo al momento $\mathrm{p}$:

$$
\cosh\eta = \frac{E}{m}, \qquad \sinh\eta = \frac{\mathrm{p}}{m}.
$$

Con las fórmulas del ángulo mitad:

$$
\cosh\frac{\eta}{2} = \sqrt{\frac{\cosh\eta + 1}{2}} = \sqrt{\frac{E+m}{2m}}, \qquad
\sinh\frac{\eta}{2} = \sqrt{\frac{\cosh\eta - 1}{2}} = \sqrt{\frac{E-m}{2m}}, \qquad
\tanh\frac{\eta}{2} = \sqrt{\frac{E-m}{E+m}} = \frac{\mathrm{p}}{E+m} = \kappa .
$$

Aplicando $S(\eta\,\hat{z})$:

$$
S(\eta \, \hat{z}) \, u_s(m) = \sqrt{2m} \begin{pmatrix} \cosh\frac{\eta}{2} \, \chi_s \\ \sinh\frac{\eta}{2} \, \sigma_3 \, \chi_s \end{pmatrix}
= \sqrt{2m} \, \cosh\frac{\eta}{2} \begin{pmatrix} \chi_s \\ \tanh\frac{\eta}{2} \, \sigma_3 \, \chi_s \end{pmatrix}
= \sqrt{E+m} \begin{pmatrix} \chi_s \\ \kappa \, \sigma_3 \, \chi_s \end{pmatrix}
$$

$$
\boxed{
S(\eta \, \hat{z}) \, u_s(m) = u_s(p)
}
$$

Es el espinor de la sección 2, obtenido ahora **sin resolver la ecuación**.

**Qué se lee:**

* $\kappa = \tanh(\eta/2)$: es la tangente de la **mitad de la rapidez**. Va de $0$ (reposo) a $1$
  ($\eta \to \infty$), y por eso mide cuán relativista es el fermión.
* $N = \sqrt{E+m}$ no es arbitrario: es lo que da el *boost* partiendo de $\sqrt{2m}$ en reposo.
* El *boost* es a lo largo del eje del espín y no cambia $S_3$: $u_1(m)$, espín arriba, pasa a ser
  $u_1(p) = u_+(p)$, de helicidad positiva.
* El *boost* **no es unitario**: $u^\dagger u$ pasa de $2m$ a $2E$ (la densidad crece como $E/m$,
  porque el volumen se contrae). Lo que no cambia es $\bar{u}u = u^\dagger\gamma^0 u = N^2(1-\kappa^2) = 2m$.
  Por eso hace falta el espinor adjunto, $\bar{u} = u^\dagger \gamma^0$ (ver *El espinor adjunto*).
* Lo mismo para la antipartícula: $S(\eta\,\hat{z}) \, v_s(m) = \sqrt{2m} \, (\sinh\frac{\eta}{2}\,\sigma_3\,\xi_s, \; \cosh\frac{\eta}{2}\,\xi_s) = v_s(p)$.

2. Aplica $S(-\eta\,\hat{z})$ a $u_s(p)$. ¿Qué obtienes?
3. Comprueba que $S(\eta\,\hat{z})$ no es unitaria, $S^\dagger S \neq I$, y que la rotación sí lo es.

## 5. Rotaciones y *boosts* en una dirección cualquiera

Las fórmulas de la sección 4 se generalizan cambiando el eje $z$ por un eje unitario cualquiera, $\hat{\bf n}$:

* **rotación** de ángulo $\alpha$ en torno a $\hat{\bf n}$:

$$
\boxed{
S(\alpha \, \hat{\bf n}) = e^{-i \alpha \, \hat{\bf n} \cdot {\bf S}}
= \cos\frac{\alpha}{2} \, I - i \sin\frac{\alpha}{2} \, (\hat{\bf n} \cdot {\bf \Sigma})
}
$$

* ***boost*** de rapidez $\eta$ en la dirección $\hat{\bf n}$:

$$
\boxed{
S(\eta \, \hat{\bf n}) = e^{\frac{\eta}{2} \gamma^0 (\hat{\bf n} \cdot \vec{\gamma})}
= \cosh\frac{\eta}{2} \, I + \sinh\frac{\eta}{2} \, \gamma^0 (\hat{\bf n} \cdot \vec{\gamma})
}
$$

porque de nuevo $(\hat{\bf n} \cdot {\bf \Sigma})^2 = I$ y $[\gamma^0 (\hat{\bf n} \cdot \vec{\gamma})]^2 = I$.

### 5.1. Rotación en una dirección cualquiera

**Rotar el espín.** Llevamos el eje $z$ a una dirección cualquiera
$\hat{\bf v} = (\sin\theta\cos\phi, \, \sin\theta\sin\phi, \, \cos\theta)$ con dos rotaciones:
primero un ángulo $\theta$ en torno a $y$ y después un ángulo $\phi$ en torno a $z$,

$$
S(R) \equiv S(\phi\,\hat{z}) \, S(\theta\,\hat{y}), \qquad
S(\theta\,\hat{y}) = \begin{pmatrix} e^{-i\frac{\theta}{2}\sigma_2} & 0 \\ 0 & e^{-i\frac{\theta}{2}\sigma_2} \end{pmatrix}, \qquad
e^{-i\frac{\theta}{2}\sigma_2} = \begin{pmatrix} \cos\frac{\theta}{2} & -\sin\frac{\theta}{2} \\ \sin\frac{\theta}{2} & \cos\frac{\theta}{2} \end{pmatrix}.
$$

Como la rotación no mezcla bloques, basta ver qué les hace a $\chi_1$ y $\chi_2$:

$$
\chi_1 \xrightarrow{\;\theta\,\hat{y}\;} \begin{pmatrix} \cos\frac{\theta}{2} \\ \sin\frac{\theta}{2} \end{pmatrix}
\xrightarrow{\;\phi\,\hat{z}\;} e^{-i\phi/2} \begin{pmatrix} \cos\frac{\theta}{2} \\ e^{i\phi}\sin\frac{\theta}{2} \end{pmatrix} = e^{-i\phi/2} \, \chi_+(\hat{\bf v}),
$$

$$
\chi_2 \xrightarrow{\;\theta\,\hat{y}\;} \begin{pmatrix} -\sin\frac{\theta}{2} \\ \cos\frac{\theta}{2} \end{pmatrix}
\xrightarrow{\;\phi\,\hat{z}\;} e^{-i\phi/2} \begin{pmatrix} -\sin\frac{\theta}{2} \\ e^{i\phi}\cos\frac{\theta}{2} \end{pmatrix} = e^{-i\phi/2} \, \chi_-(\hat{\bf v}).
$$

Salen, **sin resolver ninguna ecuación de autovalores**, los bi-espinores de helicidad de la
sección *Helicidad en una dirección cualquiera*, con los mismos ángulos. La fase común
$e^{-i\phi/2}$ es global y se omite (allí se eligió real la componente de arriba). En reposo,
$S(R) \, u_1(m) \propto \sqrt{2m} \, (\chi_+(\hat{\bf v}), 0)$: el espín apunta ahora en $\hat{\bf v}$,
$(\hat{\bf v}\cdot\vec{\sigma}) \, \chi_\pm = \pm \chi_\pm$.

**Rotar el fermión en movimiento.** Aplicamos la misma rotación a $u_\pm(p)$ con ${\bf p}$ en $z$.
Ahora gira también el momento, $\mathrm{p}\,\hat{z} \to \mathrm{p}\,\hat{\bf v}$, y $\sigma_3$ se convierte en
la componente de $\vec\sigma$ a lo largo del eje rotado, $R \, \sigma_3 \, R^{-1} = \hat{\bf v}\cdot\vec{\sigma}$
(con $R = e^{-i\frac{\phi}{2}\sigma_3} e^{-i\frac{\theta}{2}\sigma_2}$). Resultado:

$$
S(R) \, u_\pm(\mathrm{p}\,\hat{z}) \propto N \begin{pmatrix} \chi_\pm(\hat{\bf v}) \\ \pm\kappa \, \chi_\pm(\hat{\bf v}) \end{pmatrix} = u_\pm(\mathrm{p}\,\hat{\bf v}),
$$

que son los $u_r(p)$ de helicidad $r = \pm$. **La helicidad no cambia bajo rotaciones**: giran a la
vez el espín y el momento.

Para la antipartícula, $\xi_1 = \chi_2$ y $\xi_2 = \chi_1$, luego $\xi_1 \to \chi_-(\hat{\bf v}) = \xi_+(\hat{\bf v})$ y
$\xi_2 \to \xi_-(\hat{\bf v})$: se obtienen igualmente los $v_\pm(\mathrm{p}\,\hat{\bf v})$.

### 5.2. *Boost* en una dirección cualquiera

Como $\gamma^0\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ \sigma_k & 0 \end{pmatrix}$,
el *boost* de rapidez $\eta$ en $\hat{\bf v}$ es el de $z$ cambiando $\sigma_3$ por $\hat{\bf v}\cdot\vec{\sigma}$:

$$
S(\eta\,\hat{\bf v}) = \begin{pmatrix} \cosh\frac{\eta}{2} \, I & \sinh\frac{\eta}{2} \, (\hat{\bf v}\cdot\vec{\sigma}) \\ \sinh\frac{\eta}{2} \, (\hat{\bf v}\cdot\vec{\sigma}) & \cosh\frac{\eta}{2} \, I \end{pmatrix}
= \sqrt{\frac{E+m}{2m}} \begin{pmatrix} I & \kappa \, (\hat{\bf v}\cdot\vec{\sigma}) \\ \kappa \, (\hat{\bf v}\cdot\vec{\sigma}) & I \end{pmatrix}.
$$

Sobre el espinor en reposo, con cualquier $\chi$:

$$
\boxed{
S(\eta\,\hat{\bf v}) \, \sqrt{2m} \begin{pmatrix} \chi \\ 0 \end{pmatrix} = N \begin{pmatrix} \chi \\ \kappa \, (\hat{\bf v}\cdot\vec{\sigma}) \, \chi \end{pmatrix}
}
$$

que es la solución general $u_s(p)$ con ${\bf p} = \mathrm{p}\,\hat{\bf v}$. El *boost* no toca el bloque de
arriba: el espín de la solución es el que tenía en reposo.

* Si en reposo el espín está en $z$ ($\chi_1$) y el *boost* va en $\hat{\bf v} \neq \hat{z}$, el resultado
  **no** es autoestado de helicidad: espín en $z$, momento en $\hat{\bf v}$.
* Si en reposo el espín ya apunta en $\hat{\bf v}$ ($\chi_\pm(\hat{\bf v})$), el resultado es $u_\pm(\mathrm{p}\,\hat{\bf v})$.

### 5.3. Dos caminos, el mismo espinor

Rotar y luego acelerar, o acelerar y luego rotar:

$$
u_1(m) \xrightarrow{\;S(\eta\,\hat{z})\;} u_+(\mathrm{p}\,\hat{z}) \xrightarrow{\;S(R)\;} u_+(\mathrm{p}\,\hat{\bf v}),
\qquad
u_1(m) \xrightarrow{\;S(R)\;} \sqrt{2m}\begin{pmatrix}\chi_+(\hat{\bf v}) \\ 0\end{pmatrix} \xrightarrow{\;S(\eta\,\hat{\bf v})\;} u_+(\mathrm{p}\,\hat{\bf v}),
$$

porque $S(\eta\,\hat{\bf v}) \, S(R) = S(R) \, S(\eta\,\hat{z})$: un *boost* en $\hat{\bf v}$ es el *boost* en
$z$ visto desde el sistema rotado. (Salvo la fase global $e^{-i\phi/2}$.)

## El espinor adjunto, la norma y la corriente

En mecánica cuántica nuestra función de ondas esta normalizada, toda transformación física preserva la norma. Además exite una corriente de probabilidad que cumple la ecuación de continuidad.

Vamos a identificar ahora la norma y la corriente para los soluciones de la ec. de Dirac, pero para ello necesitamos introducir antes el espinor adjunto.


### El espinor adjunto

Definimos el espinor adjunto como:

$$
\bar{\Psi}(x) \equiv \Psi^\dagger(x) \gamma^0
$$

En la representación de Pauli-Dirac:

$$
\bar{\Psi}(x) = (\psi^*_1(x), \psi^*_2(x), -\psi^*_3(x), -\psi^*_4(x)),
$$

que tiene forma de vector fila.

### La ecuación de Dirac para el espinor adjunto


La ec. de Dirac para el espinor adjunto se obtiene tras aplicar el transpuesto conjugado a la ec. de Dirac:

$$
- i \partial_\mu \Psi^\dagger \, (\gamma^\mu)^\dagger - m \Psi^\dagger = 0 \\
$$

si primero usamos $(\gamma^0)^\dagger = \gamma^0, \, (\gamma^k)^\dagger = - \gamma^k$, luego multiplicamos por $\gamma^0$ por la derecha, y aplicando $\gamma^k\gamma^0 = - \gamma^0\gamma^k$, obtenemos:

$$
-i\partial_0 \Psi^\dagger \gamma^0 + i \partial_k \Psi^\dagger \gamma^k - m \Psi^\dagger = 0 \\
-i\partial_0 \Psi^\dagger \gamma^0 \gamma^0 + i \partial_k \Psi^\dagger \gamma^k  \gamma^0 - m \Psi^\dagger \gamma^0 = 0 \\
-i\partial_0 \Psi^\dagger \gamma^0 \gamma^0 - i \partial_k \Psi^\dagger \gamma^0  \gamma^k - m \Psi^\dagger \gamma^0 = 0 \\
-i\partial_\mu \bar{\Psi} \gamma^\mu - m \bar{\Psi}= 0 \\
$$

Finalmente:

$$
i\partial_\mu \bar{\Psi} \, \gamma^\mu + m \bar{\Psi}= 0
$$

que es la ecuación de Dirac para el espinor adjunto.


### Norma

La norma asociada a las soluciones de la ecuación de dirac es:

$$
\bar{\Psi}(x)\Psi(x)
$$

que se trata de un escalar Lorentz, preserva unitariedad (ver [ext-trans-Lorentz](./ext-trans-Lorentz.ipynb)), y cuyo valor (por convención) se toma como $2m$ para los fermiones, $\bar{u}u = 2m$, y $-2m$ para los antifermiones, $\bar{v}v = -2m$.


### Corriente fermiónica de probabilidad

La corriente de probabilidad fermiónica es:

$$
\bar{\Psi} \gamma^\mu \Psi = (\rho, {\bf j})
$$

donde $\rho$, es la densidad, y ${\bf j}$ será el corriente de probabilidad del spinor de Dirac.

Notar que los cuatro elementos de la corriente corresponden al valor de los índices de $\mu$

En la extensión [ext-trans-Lorentz] se demuestra que efecticamente la corriente se transforma como un vector bajo transformaciones de Lorentz. Veamos aquí que cumple la ecuación de conservación:

$$
\frac{\partial \rho}{\partial t} + \nabla \cdot {\bf j} = 0.
$$

O en forma covariante:

$$
\partial_\mu j^\mu = 0
$$


La densidad es:

$$
\rho = \bar{\Psi} \gamma^0 \Psi = \Psi^\dagger \Psi = \sum_{i = 1}^4 |\psi_i|^2
$$

Podemos comprobar que es cierto:

$$
\partial_\mu \left(\bar{\Psi} \gamma^\mu \Psi\right) = (\partial_\mu \bar{\Psi}) \gamma^\mu \Psi + \bar{\Psi} \gamma^\mu (\partial_\mu\Psi)
$$

De las ecuaciones de Dirac del espinor y el espinor adjunto, tenemos que:

$$
\partial_\mu \bar{\Psi} \gamma^\mu = -i m \bar{\Psi}, \;\;\; \gamma^\mu \partial_\mu \Psi = +i m \Psi
$$

Si sustituimos:

$$
(\partial_\mu \bar{\Psi} \gamma^\mu) \Psi + \bar{\Psi} (\gamma^\mu \partial_\mu\Psi) = (-i m \bar{\Psi}) \Psi + \bar{\Psi} (i m \Psi) = 0.
$$

Luego:

$$
\partial_\mu \left(\bar{\Psi} \gamma^\mu \Psi\right) = 0, \;\; \partial_\mu j^\mu = 0.
$$

Concluimos que $j^\mu = \bar{\Psi} \gamma^\mu \Psi$ es una corriente conservada, que llamaremos corriente fermiónica.


## Helicidad en una dirección cualquiera

Para un momento en una dirección cualquiera, ${\bf p} = \mathrm{p} \, \hat{\bf v}$, todo lo de la
sección 2 vale cambiando $\sigma_3$ por $(\hat{\bf v} \cdot \vec{\sigma})$:

$$
u_s(p) = N \begin{pmatrix} \chi_s \\ \kappa \, (\hat{\bf v} \cdot \vec{\sigma}) \, \chi_s \end{pmatrix}, \qquad
v_s(p) = N \begin{pmatrix} \kappa \, (\hat{\bf v} \cdot \vec{\sigma}) \, \xi_s \\ \xi_s \end{pmatrix},
$$

con

$$
(\hat{\bf v} \cdot \vec{\sigma}) = \begin{pmatrix} v_z & v_x - i v_y \\ v_x+iv_y & -v_z \end{pmatrix} =
\begin{pmatrix} \cos\theta & \sin\theta \, e^{-i\phi} \\ \sin\theta \, e^{i\phi} & -\cos\theta \end{pmatrix},
\qquad (\hat{\bf v} \cdot \vec{\sigma})^2 = I,
$$

donde $\theta, \phi$ son los ángulos polar y azimutal de $\hat{\bf v}$.

Pero ahora $\chi_1, \chi_2$ ya no son autoestados de helicidad: tienen el espín en $z$ y el momento
en $\hat{\bf v}$.

### Soluciones autoestados de helicidad

Nos podemos preguntar ahora cuáles son las soluciones de la ec. de Dirac que son a su vez autoestados de helicidad, esto es que tienen su espín alineado con el momento.

Los autovalores de helicidad, como los de tercera componente de espín, son $\pm 1/2$. 

$$
h \, u_s(p) = \pm \frac{1}{2} u_s(p)
$$

Lo que se traduce:

$$
\frac{1}{2} \begin{pmatrix} (\vec{\sigma} \cdot \hat{\bf v}) & 0 \\ 0  & (\vec{\sigma} \cdot \hat{\bf v}) \end{pmatrix} \, 
\begin{pmatrix} \chi_r \\ \kappa (\vec{\sigma} \cdot \hat{\bf v}) \chi_r  \end{pmatrix} = \pm \frac{1}{2} \, \begin{pmatrix} \chi_r \\  \kappa (\vec{\sigma} \cdot \hat{\bf v}) \chi_r\end{pmatrix}
$$

Que se traduce en dos ecuaciones, pero la segunda es una identidad si se cumple la primera:

$$
(\sigma \cdot \hat{\bf v}) \, \chi_r(\hat{\bf v}) =  \pm \, \chi_r(\hat{\bf v}) \\
\kappa \, (\sigma \cdot \hat{\bf v}) \left[ (\sigma \cdot \hat{\bf v}) \, \chi_r(\hat{\bf v}) \right]=  \pm \kappa \, (\sigma \cdot \hat{\bf v}) \, \chi_r(\hat{\bf v})
$$



Vamos a encontrar pues el bi-espinor, $\chi_r(\hat{\bf v})$, que cumple

$$
(\sigma \cdot \hat{\bf v}) \, \chi_r(\hat{\bf v}) =  \pm \, \chi_r(\hat{\bf v})
$$

que vemos que depende ahora de $\hat{\bf v}$. 

Si escribimos explícitamente:

$$
(\vec{\sigma} \cdot \hat{\bf v}) = \begin{pmatrix} v_z & v_x - i v_y \\ v_x+iv_y & -v_z \end{pmatrix} = 
\begin{pmatrix} \cos\theta & \sin\theta e^{-i\phi} \\ \sin\theta e^{i\phi} & -\cos\theta \end{pmatrix}
$$

La condición de las dos componentes $a, b$ de $\chi_r(\hat{\bf v})$ es:

$$
\begin{pmatrix} \cos\theta & \sin\theta e^{-i\phi} \\ \sin\theta e^{i\phi} & -\cos\theta \end{pmatrix}
\begin{pmatrix} a \\ b \end{pmatrix} = 
\pm \begin{pmatrix} a \\ b \end{pmatrix}
$$



Esto es, con $\lambda = \pm 1$:

$$
a \cos \theta + b \sin \theta e^{-i\phi}  = \lambda \, a, \\
b \sin \theta e^{-i\phi}  = (\lambda -  \cos\theta) \, a \\
$$

Por lo tanto, las componentes $a, b$ cumplen:

$$
\frac{b}{a} = \frac{\lambda -\cos \theta}{\sin\theta} e^{i\phi}
$$


Para el caso de helicidad positiva $\lambda = +1$, obtenemos:

$$
\frac{b}{a} = \frac{\lambda -\cos \theta}{\sin\theta} e^{i\phi} = e^{i\phi} \tan \frac{\theta}{2}
$$

Resolviendo también para $\lambda = -1$, imponiendo que tienen norma unidad, y haciendo uso que sin pérdida de generalidad podemos tomar la componente de arriba, $a$, como real, los bi-espinores quedan:

$$
\chi_+(\hat{\bf v}) = \begin{pmatrix} \cos \frac{\theta}{2} \\ e^{i\phi}\sin \frac{\theta}{2}\end{pmatrix}, \;\;\;
\chi_-(\hat{\bf v}) = \begin{pmatrix} -\sin \frac{\theta}{2} \\ e^{i\phi}\cos \frac{\theta}{2}\end{pmatrix}, \;\;\;
$$



Notar que el bi-espinor definido antes, $\chi_s$,  es simplemente $\chi_r(\hat{k})$, que corresponde a seleccionar como dirección $\hat{\bf v} = \hat{k}$.

De forma similar, los bi-espinores de los anti-fermiones que son autoestados de helicidad quedan:

$$
\xi_+(\hat{\bf v}) = \begin{pmatrix} -\sin \frac{\theta}{2} \\ e^{i\phi}\cos \frac{\theta}{2}\end{pmatrix}, \;\;\;
\xi_-(\hat{\bf v}) = \begin{pmatrix} \cos \frac{\theta}{2} \\ e^{i\phi}\sin \frac{\theta}{2}\end{pmatrix}, \;\;\;
$$

Así pues, los espinores autoestados de helicidad que son soluciones de la ec. de Dirac son entonces:

$$
u_r(p) = N \begin{pmatrix} \chi_r(\hat{\bf v}) \\ \lambda_r \kappa \, \chi_r(\hat{\bf v}) \end{pmatrix}, \;\;\;
v_r(p) = N \begin{pmatrix} -\lambda_r \kappa \, \xi_r(\hat{\bf v}) \\ \xi_r(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

donde $r = \pm, \, \lambda_\pm = \pm$.


Si reescribimos explícitamente los espinores de helicidad en la dirección $z$, ($\theta = \phi = 0$), volvemos a obtener:

$$
\boxed{
u_{+}(p) = N \begin{pmatrix} 1 \\ 0 \\ \kappa  \\ 0 \end{pmatrix}, \;
u_{-}(p) = N \begin{pmatrix} 0 \\ 1 \\ 0 \\ -\kappa \end{pmatrix}, \;
v_{+}(p) = N \begin{pmatrix} 0  \\ -\kappa \\ 0 \\ 1 \end{pmatrix}, \;
v_{-}(p) = N \begin{pmatrix} \kappa  \\ 0 \\ 1 \\ 0 \end{pmatrix}. 
}
$$

con el factor de normalización $N = \sqrt{E+m}$

*Nota*: Podemos reinterpretar el espinor $u_+(p)$ como aquel que se obtiene de rotar un espinor inicialmente en $u_1(p)$, con su espín alineado en la dirección hacia arriba en $z$, hasta situar su espín alineado con $\hat{\bf v}$, con la dirección y sentido del momento, ${\bf p}$.

Es lo que se hace en la sección 5.1, sin resolver la ecuación de autovalores: rotar en torno a $y$ un ángulo $\theta$ y después en torno a $z$ un ángulo $\phi$.

## Quiralidad

La violación de paridad de las interacciones débiles está relacionada con una propiedad del espinor de Dirac, llamada quiralidad.

Decimos que un espinor tiene quiralidad a derechas o izquierdas si es autoestado de $\gamma^5$ con autovalores +1 y -1 respectivamente. Esto es:

$$
\boxed{
\gamma^5 \, u_R \equiv + u_R, \;\;\; \gamma^5 \, u_L \equiv - u_L
}
$$

La quiralidad es una propiedad *matemática* que no tiene un equivalente simple físico. Contrariamente a la helicidad, que si tiene un equivalente físico: la proyección del espín en la dirección el momento.

Veremos que un espinor de Dirac puede descomponerse en un espinor de quiralidad a izquierdas y otro a derechas, tales que la inversión de paridad intercambia la quiralidad, la derecha por la izquierda y viceversa.

Esto hace que una interacción que sea invariante por paridad, no hará distinción entre los dos quiralidades, pero una que sí lo haga, favorecerá o incluso suprimirá una quiralidad.

#### La matriz $\gamma^5$

Vamos a introducir una quinta matriz, $\gamma^5$, que va a estar muy relacionada con una propiedad que llamaremos *quiralidad* y que juega un papel fundamental en la interacciones débiles.

La matriz $\gamma^5$ se define:

$$
\gamma^5 \equiv i \gamma^0 \gamma^1 \gamma^2 \gamma^3,
$$

Y tiene las siguientes propiedades:

$$
(\gamma^5)^2 = I, \;\; \gamma^{5\dagger} = \gamma^5, \;\; \gamma^5\gamma^\mu = - \gamma^\mu \gamma^5
$$

que se derivan de las propiedades de las matrices $\gamma^\mu$.

*Cuestión:* Verifica las propiedades de $\gamma^5$

En la representación de Pauli-Dirac:

$$
\gamma^5 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix}
$$

Los proyectores de quiralidad son:

$$
\boxed{
P_L = \frac{1}{2} (1-\gamma^5), \;\;\; P_R = \frac{1}{2} (1+\gamma^5)
}
$$

De tal forma que $\Psi(x)$ se descompone en dos partes, a izquierdas y derechas.

$$
\Psi_L(x) = P_L \, \Psi(x), \;\;\, \Psi_R(x) = P_R \, \Psi(x)
$$

Los proyectores son en realidad dos matrices que están actuado sobre la parte espinorial de $\Psi(x)$.


Vemos que efectivamene un espinor de quiralidad a derechas o izquierdas cumple:

$$
\gamma^5 u_R = + u_R \Rightarrow P_R \,u_R = \frac{1}{2}(1+\gamma^5) \, u_R = u_R, 
$$
$$
\gamma^5 u_L = - u_L \Rightarrow P_L \,u_L = \frac{1}{2}(1-\gamma^5) \, u_L = u_L. 
$$


Esto es para los espinores de los fermiones, actuan sobre $u(p)$:

$$
u_L(p) = P_L \, u(p), \;\;\, u_R(p) = P_R \, u(p)
$$

donde hemos eliminado el subíndice $s$ de los espinores $u_s(p)$ por simplicidad.

Vamos a comprobar que se trata de proyectores.

Estos deben cumplir:

$$
P_R + P_L = I, \;\; P^2_R = P_R, \;\;, P^2_L = P_L, \;\; P_L P_R = P_R P_L = 0
$$

La primera es inmediata.

Comprobemos que $P^2_R = P_R$

$$
\left[\frac{1}{2} (1+\gamma^5) \right]^2 =
\frac{1}{4}(I + 2\gamma^5 + (\gamma^5)^2) = \frac{1}{4}(I + 2\gamma^5 + I) = \frac{1}{2}(1+\gamma^5)
$$

dado que $(\gamma^5)^2 = I$.

De igual manera obraríamos con $P^2_L = P_L$

Y finalmente, $P_L \, P_R = P_R \, P_L = 0$:

$$
\frac{1}{2}(1-\gamma^5) \frac{1}{2}(1+\gamma^5) = \frac{1}{4} (I-(\gamma^5)^2) = 0
$$


Vamos a aplicar paridad sobre $u_L(p), \; u_R(p)$, recordemos que la representación espinorial de la inversión por paridad es $S(P) = \gamma^0$

$$
\frac{1}{2}(1\pm\gamma^5) \, u(p) \to S(P) \, \frac{1}{2}(1 \pm \gamma^5) \, u(E, -{\bf p}) = \\
\gamma^0 \, \frac{1}{2}(1 \pm \gamma^5) \, u(E, -{\bf p}) = \\ \frac{1}{2} (1 \mp \gamma^5) (\gamma^0 \, u(E, -{\bf p})) = \frac{1}{2}(1 \mp \gamma^5) u(p)
$$

Luego la inversión por paridad nos intercambia los espinores de quiralidad a derechas por izquierdas y viceversa:

$$
u_L(p) \leftrightarrow u_R(p)
$$


Para los espinores de los antifermiones definimos la quiralidad de forma complementaria a la de los fermiones, esto es:

$$
v_R(p) =  P_L\, v(p), \;\;\; v_L(p) = P_R \, v(p), \;\;\; 
$$

Igualmente la inversión por paridad intercambia sus quiralidades

$$
v_R(p) \leftrightarrow v_L(p) 
$$

### Las espinores en su expresión quiral

Si los aplicamos sobre los espinores de Dirac que son autoestados de helicidad. Recordémos que son:

$$
u_r(p) = N \begin{pmatrix} \chi_r(\hat{\bf v}) \\ \lambda_r \kappa \, \chi_r(\hat{\bf v}) \end{pmatrix}, \;\;\;
v_r(p) = N \begin{pmatrix} -\lambda_r \kappa \, \xi_r(\hat{\bf v}) \\ \xi_r(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

donde $r = \pm, \, \lambda_\pm = \pm 1$.



Obtenemos:

$$
P_R u_\pm(p) = \frac{1}{2} (1 \pm \kappa) N \begin{pmatrix} \chi_\pm(\hat{\bf v}) \\ \chi_\pm(\hat{\bf v}) \end{pmatrix}, \;\;\;
P_L u_\pm(p) = \frac{1}{2} (1 \mp \kappa) N \begin{pmatrix} \chi_\pm(\hat{\bf v}) \\ -\chi_\pm(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

Que podemos explicitar en la siguiente tabla:

| helicidad $\downarrow$ \ quiralidad  $\rightarrow$| --- R --- | --- L --- |
| :--: | :--: | :--: | 
| + | $\frac{1}{2}(1+\kappa)$ | $\frac{1}{2}(1-\kappa)$ |
| - | $\frac{1}{2}(1-\kappa)$ | $\frac{1}{2}(1+\kappa)$ |



En el caso en el que la partícula es ultra-relativista $E \gg m$ o sin masa, $m=0$, esto es, cuando $\kappa = 1$

En ese caso el espinor de helicidad negativa solo tiene componene de quiralidad a izquierdas. Y el espinor de helicidad positiva solo quiralidad a derechas.


$$
u_R(p) = u_+(p) = N \begin{pmatrix} \chi_+(\hat{\bf v}) \\ \chi_+(\hat{\bf v}) \end{pmatrix}, \;\;\;
u_L(p) = u_-(p) = N \begin{pmatrix} \chi_-(\hat{\bf v}) \\ -\chi_-(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

solo cuando $\kappa = 1$.

Que si explicitamos para el caso $\hat{\bf v} = \hat{k}$, esto es, elegimos el eje $z$ como la dirección de movimiento de la partícula.

y en el caso ultra-relativista, 

$$
u_R(p) =  u_+(p) = \sqrt{E} \begin{pmatrix} 1 \\ 0 \\ 1  \\ 0 \end{pmatrix}, \;\;\;
u_L(p) =  u_-(p) = \sqrt{E} \begin{pmatrix} 0 \\ 1 \\ 0  \\  -1 \end{pmatrix}, \;\;\;
$$

Donde, recordemos $N = \sqrt{E + m } = \sqrt{E}$


Para los espinores de los antifermiones:

$$
v_R(p) = P_L v_\pm(p) = \frac{1}{2} (1 \pm \kappa)  N \begin{pmatrix} - \xi_\pm(\hat{\bf v}) \\ \xi_\pm(\hat{\bf v}) \end{pmatrix}, \;\;\;
v_L(p) = P_R v_\pm(p) = \frac{1}{2} (1 \mp \kappa)  N \begin{pmatrix}  \xi_\pm(\hat{\bf v}) \\ \xi_\pm(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

que si explicitamos para el caso $\hat{\bf v} = \hat{k}$ y en el caso ultra-relativista, 

$$
v_R(p) =  P_L v_+(p) = \sqrt{E} \begin{pmatrix} 0 \\ -1 \\ 0  \\ 1 \end{pmatrix}, \;\;\;
v_L(p) =  P_R v_-(p) = \sqrt{E} \begin{pmatrix} 1 \\ 0 \\ 1  \\  0 \end{pmatrix}, \;\;\;
$$

Podemos especificar los proyectores de quiralidad en la representación Pauli-Dirac:

$$
P_R = \frac{1}{2}\begin{pmatrix} I & I \\ I & I \end{pmatrix}, \;\;\;
P_L = \frac{1}{2}\begin{pmatrix} I & -I \\ -I & I \end{pmatrix},
$$


### La representación Weyl o quiral

Vamos a considerar la representación de Weyl (o quiral) de las matrices $\gamma^\mu$:

$$
\gamma^0 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix}, \;\;
\gamma^k = \begin{pmatrix} 0 & \sigma^k \\ -\sigma^k & 0 \end{pmatrix}, \;\;
\gamma^5 = \begin{pmatrix} -I & 0 \\ 0 & I \end{pmatrix}
$$

Con lo que que:

$$
P_L = \begin{pmatrix} I & 0 \\ 0 & 0 \end{pmatrix}, \;
P_R = \begin{pmatrix} 0 & 0 \\ 0 & I \end{pmatrix} \;
$$

Esto es el espinor de Dirac está compuesto de dos bi-espinores, $u_L, u_R$, llamados de Weyl, que corresponden con la parte de quiralidad a izquierdas y a derechas del espinor de Dirac.

$$
u(p) = \begin{pmatrix} u_L(p) \\ u_R(p)\end{pmatrix}
$$

Si ahora consideramos el caso revelador de un fermión sin masa $m=0$, la ec. de Dirac queda:

$$
\gamma^\mu p_\mu \, u(p) = 0
$$

que se descompone en dos ecuaciones independientes:

$$
(E - {\bf p} \cdot \vec{\sigma}) \, u_R(p) = 0, \\
(E + {\bf p} \cdot \vec{\sigma}) \, u_L(p) = 0.
$$

Esto es, el espinor de Dirac se compone de dos espinores de Weyl independientes. 

O en otras palabras, es el término de masa en la ec. de Dirac, la que hace que los dos espinores de Weyl estén conectados.


dado que $E = \text{p}$, donde ${\bf p} = \text{p} \hat{\bf v}$, estas ecuaciones son equivalentes a:

$$
(\hat{\bf v} \cdot \vec{\sigma}) \; u_R(p) = + u_R(p), \\ (\hat{\bf v} \cdot \vec{\sigma}) \; u_L(p) = - u_L(p).
$$

Esto es, hay una conexión uno a uno entre helicidad y quiralidad. El espinor de Weyl a izquierdas tiene helicidad negativa y el de derechas positiva.

En este caso de las partículas sin masa, quiralidad y la helicidad son ambas propiedades estacionarias e invariantes Lorentz.

En el Modelo Estándard este caso correspondía al neutrino.


### la norma y la corriente bajo quiralidad

Como $(\gamma^5)^\dagger = \gamma^5$, los projectores no se ven afectados por la operación $\dagger$:

$$
\frac{1}{2}(1\pm \gamma^5)^\dagger = \frac{1}{2}(1\pm \gamma^5), \;\; P^\dagger_{L/R} = P_{L/R}
$$

Mientras que el intercambio con $\gamma^0$ intercambia los proyectores:

$$
\gamma^0 \, P_{L/R} = P_{R/L} \, \gamma^0
$$

Esto separa en dos las formas invariantes Lorentz

Para la escalar $\bar{\Psi}(x) \Psi(x)$, si la escribimos solo en función de sus espinores $\bar{u}(p) u(p)$:

$$
\bar{u}(p) \, u_L(p) = \bar{u}(p) \, P_L u_L(p) = (u^\dagger(p) \gamma^0) P_L \, u_L(p) = \\ 
(u^\dagger(p)  P_R \gamma^0) \, u_L(p) = 
(P_R u(p))^\dagger \gamma^0 u_L(p) = u^\dagger_R(p) \gamma^0 u_L(p) = \bar{u}_R(p) u_L(p)
$$

Y de forma recíproca:

$$
\bar{u}(p) \, u_R(p) = \bar{u}_L \, u_R(p)
$$

Luego la forma escalar:

$$
\bar{u}(p) \, u(p) = \bar{u}_R \, u_L(p) + \bar{u}_L \, u_R(p)
$$

Se compone de dos partes donde las dos quiralidades se acoplan una a la otra. 

Esto implica que un término escalar en el Lagrangiano (el que estará asociado a la masa) exige la existencia de las dos quiralidades.

Recordando que $\gamma^5\gamma^\mu = - \gamma^\mu\gamma^5$, vemos que:

$$
\gamma^\mu \, P_{L/R} = P_{R/L} \gamma^\mu
$$

el intercambio de lado de $\gamma^\mu$, intercambia los proyectores.

Veamos ahora cómo queda la forma vectorial, $\bar{\Psi}(x)\gamma^\mu\Psi(x)$ escrita en sus componentes de quiralidad. Desarrollamos nuevamente la expresión con os espinores, $\bar{u}(p) \gamma^\mu u(p)$.

$$
\bar{u}(p) \gamma^\mu u_L(p) = \bar{u}(p) \gamma^\mu P_L u_L(p) =  (u^\dagger(p) \gamma^0) \gamma^\mu P_L \, u_L(p) = \\ 
u^\dagger(p) \gamma^0 P_R \, \gamma^\mu \, u_L(p) = (u^\dagger(p) P_L) \gamma^0 \gamma^\mu \, u_L(p) = \\
(P_L u(p))^\dagger \gamma^0 \gamma^\mu u_L(p) = (u^\dagger_L(p) \gamma^0) \gamma^\mu u_L(p) = \bar{u}_L(p) \gamma^\mu u_L(p)
$$

De forma similar:

$$
\bar{u}(p) \gamma^\mu u_R(p) = \bar{u}_R(p) \gamma^\mu u_R(p)
$$

Así la forma vectorial queda:

$$
\bar{u}(p) \gamma^\mu u(p) = \bar{u}_R(p) \gamma^\mu u_R(p) + \bar{u}_L(p) \gamma^\mu u_L(p)
$$

Esto es la corriente se divide en dos, una corriente de quiralidad a izquierdas y otra a derechas.

En resumen:

$$
\bar{\Psi}\Psi =  \bar{\Psi}_R \Psi_L + \bar{\Psi}_L \Psi_R, \\
\bar{\Psi} \gamma^\mu \Psi = \bar{\Psi}_R \gamma^\mu \Psi_R + \bar{\Psi}_L \gamma^\mu \Psi_L
$$
