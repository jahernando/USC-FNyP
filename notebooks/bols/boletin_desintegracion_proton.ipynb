## El problema

En el Modelo Estándar el número bariónico $B$ se conserva, esto implica que el protón no se desintegra, es el barión más ligero.

Las teorías de Gran Unificación (GUT) sitúan quarks y leptones en los mismos multipletes, y predicen que el protón se desintegra, esto es violan la conservación del número bariónico, con vidas medias del orden de $10^{34}$–$10^{36}$ años.

El canal favorito de muchas GUT es:

$$
p \to e^+ + \pi^0
$$

Nota: el $\pi^0$ es un mesón, con masa $m=135$ MeV. 

</div>

<details>
<summary><b>Solución</b></summary>

1. Carga: $+1 \to +1 + 0$. Energía: $m_p = 938$ MeV $> m_e + m_{\pi^0} \approx 0.5 + 135$ MeV. Cinemáticamente permitida, con $Q = m_p - m_e - m_{\pi^0} \simeq 938.3 - 0.5 - 135.0 = 802.8$ MeV.

Nota: Esta reacción no conserva el número leptónico y el bariónico: $B$: $1 \to 0$. $L$: $0 \to -1$. Viola $B$ y $L$, pero **conserva** $B-L$ ($1 \to 0-(-1) = 1$), como predicen muchas GUT.

Nota: Desintegración a dos cuerpos: $e^+$ y $\pi^0$ **back-to-back**, con $p \simeq 459$ MeV cada uno.
   El $\pi^0 \to \gamma + \gamma$ se desintegra inmediatamente. En un detector Cherenkov: tres anillos tipo electrón (*showering*), momento total $\approx 0$ y masa invariante total $\approx m_p$.

</details>

## ¿Qué es $\tau/\mathcal{B}$?

El protón podría desintegrarse por varios canales $i$, cada uno con su anchura parcial $\Gamma_i$:

$$
\Gamma = \sum_i \Gamma_i, \qquad \tau = \frac{1}{\Gamma}, \qquad \mathcal{B}_i = \frac{\Gamma_i}{\Gamma}
$$

La *anchura parcial* del canal $i$:

$$
\Gamma_i = {\mathcal{B}_i} \, \Gamma =  \frac{\mathcal{B}_i}{\tau}
$$

Un experimento que busca **un** canal solo es sensible a $\Gamma_i$, SK da la **vida media parcial** $\tau_i \equiv 1/\Gamma_i = \tau/\mathcal{B}_i$.


## ¿Cuántos protones tiene SK?

1. ¿Cuántas moléculas de agua hay en 1 kt?
2. ¿Cuántos protones **libres** (núcleos de H)? ¿Y **totales** (H + O)?

* De cada 10 protones, 2 son **libres** (H) y 8 están **ligados** en el $^{16}$O. SK usa los **10**.
* El artículo **no** da el número de protones: da la exposición en kt·año y hay que calcularlo.

*Nota* — **Protones libres y ligados**

* SK cuenta **todos** los protones del agua, 2 libres y 8 ligados por molécula, y **supone que todos se desintegran con la misma probabilidad**:

> *"When generating proton decay events, the decay is assumed to happen with equal probability for each proton in the water molecule."* [SK2020, Sec. IV]

* Es una hipótesis razonable: la ligadura nuclear ($\lesssim 40$ MeV) es pequeña frente a $Q \simeq 800$ MeV.

* La diferencia entre libres y ligados **solo** se tiene en cuenta al estimar, con Monte Carlo, la **eficiencia de selección** $\varepsilon$:
  * **libre**: en reposo, dos cuerpos, $e^+$ y $\pi^0$ *back-to-back* con $p \simeq 459$ MeV. Masa total $\simeq m_p$ y momento total $\simeq 0$.
  * **ligado**: momento de Fermi ($\sim 200$ MeV), energía de ligadura, y el $\pi^0$ puede dispersarse, absorberse o intercambiar carga en el núcleo antes de salir. Masa y momento reconstruidos se deforman y $\varepsilon$ es menor.

* Por eso SK usa dos regiones de señal: $P_{tot} < 100$ MeV/c (casi solo protones libres, sin fondo) y $100 \leq P_{tot} < 250$ MeV/c (sobre todo ligados).

* La $\varepsilon$ del artículo es un **promedio sobre los 10 protones**.

## ¿Cuántas desintegraciones esperamos?

Con $N_0$ protones y vida media parcial $\tau$ (omitimos el $\mathcal{B}$), tras un tiempo $t \ll \tau$:

$$
N(t) = N_0 \, e^{-t/\tau} \simeq N_0 \left(1 - \frac{t}{\tau}\right)
$$

El número **medio** de desintegraciones:

$$
\nu = N_0 \, \frac{t}{\tau}
$$

Y el número medio que **detectamos**, con eficiencia de selección $\varepsilon$:

$$
\nu = \varepsilon \, N_0 \, \frac{t}{\tau}
$$

$N_0 \, t$ es la exposición en protones·año.

## ¿Y si se desintegraran 10 en un año?

<div class="admonition tip">
<p class="admonition-title"><b>[>] Taller — Diez desintegraciones</b></p>

Supón que en el volumen fiducial convencional (22.5 kt) se observan $n = 10$ desintegraciones
$p \to e^+ \pi^0$ -supón que no hay fondo de contaminación- y con $\varepsilon = 1$.

1. ¿Qué vida media parcial $\tau/\mathcal{B}$ estimas si las 10 se observan en **1 año**? ¿Y en **2 años**?
2. ¿Con qué precisión? Pista: el error de un número de cuentas $n$ es $\sqrt{n}$.
3. ¿Es compatible con el límite de SK? ¿Cuántas desintegraciones habría esperado SK en sus 22 años?
4. ¿Cambia algo si la eficiencia es $\varepsilon \simeq 0.4$?

</div>

```python
n = 10                              # desintegraciones observadas
for T in (1, 2):                    # años
    tau_est = N0 * T / n
    err_rel = 1 / np.sqrt(n)
    print('T = {} año(s): tau/B = ({:.1f} +- {:.1f}) x 1e32 años'.format(
          T, tau_est / 1e32, err_rel * tau_est / 1e32))

tau_10 = N0 * 1 / 10                # la hipótesis de 10 al año
nu_SK  = 450 * p_total / tau_10     # exposición del artículo: 450 kt año
print()
print('Con tau/B = {:.1e} años SK esperaría nu = {:.0f} desintegraciones'.format(tau_10, nu_SK))
print('Probabilidad de ver 0 si nu = {:.0f} : P(0) = {:.1e}'.format(nu_SK, poisson.pmf(0, nu_SK)))
```

<details>
<summary><b>Discusión</b></summary>

1. $\tau/\mathcal{B} = N_0 T / n \approx 7.5 \times 10^{32}$ años (1 año) y $1.5 \times 10^{33}$ años (2 años).
   Mismo número de sucesos en más tiempo: vida media más larga.
2. $\sigma_n/n = 1/\sqrt{n} \approx 30\%$: el error relativo es el mismo en los dos casos, **solo depende de $n$**.
3. No: $\tau/\mathcal{B}$ sería $\sim 30$ veces menor que el límite. SK habría esperado $\sim 200$ sucesos,
   y la probabilidad de ver 0 es $e^{-200}$. Esta hipótesis está **excluida**.
4. Con $\varepsilon$, las desintegraciones reales son $n/\varepsilon = 25$, y $\tau/\mathcal{B} = \varepsilon N_0 T/n$
   es aún menor: todavía más excluida.

</details>

## Un año de SK, repetido muchas veces

Simulamos el conteo anual de desintegraciones en SK (22.5 kt, $\varepsilon = 1$) para las dos hipótesis:

* $\tau/\mathcal{B} = 7.5 \times 10^{32}$ años ($\nu = 10$ al año)
* $\tau/\mathcal{B} = 2.4 \times 10^{34}$ años ($\nu = 0.31$ al año), el límite de SK

```python
rng = np.random.default_rng()
n_años = 10000

fig, axs = plt.subplots(1, 2, figsize=(10, 4))
for ax, tau_i in zip(axs, (tau_10, 2.4e34)):
    nu_i = N0 * 1 / tau_i
    cuentas = rng.poisson(nu_i, size=n_años)       # un número por cada 'año' simulado
    ns = np.arange(0, 25)
    ax.hist(cuentas, bins=ns - 0.5, density=True, alpha=0.5, label='simulación')
    ax.plot(ns, poisson.pmf(ns, nu_i), 'o', label=r'Poisson $\nu$ = {:.2f}'.format(nu_i))
    ax.set_xlabel('desintegraciones en un año'); ax.set_ylabel('probabilidad')
    ax.set_title(r'$\tau/\mathcal{{B}}$ = {:.1e} años'.format(tau_i))
    ax.grid(alpha=0.3); ax.legend()
fig.tight_layout()
```

*Cuestión* Con $\nu = 10$, ¿es raro ver 5 o 15? Con $\nu = 0.31$, ¿cuál es el resultado más probable?

### Desintegraciones a lo largo del tiempo

Cada desintegración ocurre en un instante aleatorio: los tiempos entre sucesos siguen una exponencial
de media $1/R$, con $R = N_0/\tau$ la razón de desintegraciones. Simulamos 22 años de SK:

```python
T_max = 22   # años

fig, ax = plt.subplots(figsize=(8, 4))
for tau_i in (7.5e32, 7.5e33, 2.4e34):
    R = N0 / tau_i                                      # desintegraciones / año
    dt = rng.exponential(1 / R, size=int(3 * R * T_max) + 20)
    t = np.cumsum(dt); t = t[t < T_max]
    ax.step(np.concatenate(([0], t, [T_max])), np.concatenate(([0], np.arange(1, len(t) + 1), [len(t)])),
            where='post', label=r'$\tau/\mathcal{{B}}$ = {:.1e} años, $\nu$ = {:.1f}'.format(tau_i, R * T_max))
ax.set_xlabel('tiempo (años)'); ax.set_ylabel('desintegraciones acumuladas')
ax.set_yscale('symlog'); ax.grid(alpha=0.3); ax.legend();
```

*Cuestión* Ejecuta la celda varias veces. ¿Qué ocurre con la curva de $\tau/\mathcal{B} = 2.4 \times 10^{34}$ años?

## La distribución de Poisson

Cada protón es un "intento" con una probabilidad minúscula de desintegrarse en el tiempo $t$:

$$
p = \frac{t}{\tau} \sim 10^{-34}
$$

El número de desintegraciones entre $N_0 \sim 10^{34}$ protones sigue una **binomial** $B(n \,|\, N_0, p)$.

Cuando $N_0 \to \infty$, $p \to 0$ con $\nu = N_0 p$ fijo, la binomial tiende a la **Poisson**:

$$
P(n \,|\, \nu) = \frac{\nu^n}{n!} \, e^{-\nu}, \qquad \langle n \rangle = \nu, \qquad \sigma_n = \sqrt{\nu}
$$

Solo depende de $\nu$: da igual cuántos protones y cuánto tiempo, solo importa el número medio.

<div class="admonition tip">
<p class="admonition-title"><b>[>] Taller — De la binomial a la Poisson</b></p>

No podemos simular $10^{34}$ protones, pero sí comprobar el límite con números más modestos.

1. Fija $\nu = 3$ y compara la binomial con $N_0 = 10, 100, 10^4$ (y $p = \nu/N_0$) con la Poisson. ¿A partir de qué $N_0$ son indistinguibles?
2. Dibuja la Poisson para $\nu = 0.3, \, 2.3, \, 10, \, 20$. ¿Cómo cambia la forma? ¿Y la anchura relativa $\sigma_n/\nu$?

</div>

```python
nu = 3
ns = np.arange(0, 12)

fig, axs = plt.subplots(1, 2, figsize=(11, 4))
for N in (10, 100, 10000):
    axs[0].plot(ns, binom.pmf(ns, N, nu / N), 'o--', alpha=0.6, label='binomial $N_0$ = {}'.format(N))
axs[0].bar(ns, poisson.pmf(ns, nu), alpha=0.3, color='k', label=r'Poisson $\nu$ = {}'.format(nu))
axs[0].set_xlabel('$n$'); axs[0].set_ylabel('probabilidad'); axs[0].legend(); axs[0].grid(alpha=0.3)

ns = np.arange(0, 35)
for nu_i in (0.3, 2.3, 10, 20):
    axs[1].plot(ns, poisson.pmf(ns, nu_i), 'o-', ms=4, label=r'$\nu$ = {}'.format(nu_i))
axs[1].set_xlabel('$n$'); axs[1].set_ylabel(r'$P(n | \nu)$'); axs[1].legend(); axs[1].grid(alpha=0.3)
fig.tight_layout()
```

<details>
<summary><b>Discusión</b></summary>

1. Con $N_0 = 100$ ya casi coinciden; con $N_0 = 10^4$ son indistinguibles. Con $10^{34}$ protones, la Poisson es exacta a efectos prácticos.
2. Para $\nu \lesssim 1$ el máximo está en $n = 0$ y la distribución es muy asimétrica.
   Para $\nu$ grande se parece a una gaussiana. La anchura relativa $1/\sqrt{\nu}$ disminuye:
   **10 sucesos → 30%**, **100 sucesos → 10%**.

</details>

## No vemos nada: el límite al 90% CL

SK observa $n = 0$. ¿Qué valores de $\tau$ podemos **excluir**?

Idea: un valor de $\nu$ (y por tanto de $\tau$) queda **excluido al 90% CL** si, siendo cierto, la probabilidad de obtener
un resultado como el observado **o más extremo** (aquí, menos sucesos) es menor del 10%.

Con $n = 0$ no hay nada "más extremo" que 0:

$$
P(0 \,|\, \nu) = e^{-\nu} < 0.10 \quad \Rightarrow \quad \nu > \ln 10 = 2.30
$$

Es decir, si ese $\nu$ fuera el verdadero, **más del 90% de los experimentos observarían más sucesos** ($n \geq 1$) que los que hemos visto.

*Ojo*: no significa que haya un 90% de probabilidad de que ese $\nu$ sea falso.

```python
nus = np.linspace(0, 5, 200)
nu90 = -np.log(0.10)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(nus, poisson.pmf(0, nus), label=r'$P(0 | \nu) = e^{-\nu}$')
ax.axhline(0.10, color='gray', ls='--')
ax.axvspan(nu90, 5, color='C3', alpha=0.2, label=r'excluido al 90% CL, $\nu > {:.2f}$'.format(nu90))
ax.set_xlabel(r'$\nu$'); ax.set_ylabel('probabilidad de observar 0')
ax.grid(alpha=0.3); ax.legend();
print('nu_90 = {:.3f}'.format(nu90))
```

Como $\nu = \varepsilon N_0 t/\tau$, un límite superior en $\nu$ es un límite **inferior** en $\tau$:

$$
\frac{\tau}{\mathcal{B}} > \frac{\varepsilon \, N_0 \, t}{2.30} \qquad (90\%\ \mathrm{CL})
$$

*Nota* Que $\tau$ esté en el límite no significa que no pueda ser el verdadero: con ese valor, en el 10% de los experimentos se vería 0.

*Nota* El valor del límite **depende del método** estadístico. Para $n = 0$ y sin fondo, al 90% CL:

| método | límite superior en $\nu$ |
| :-- | :--: |
| clásico, $P(0 \,|\, \nu) < 0.10$ | 2.30 |
| bayesiano, prior plano en $\nu$ | 2.30 |
| Feldman-Cousins | 2.44 |

Feldman y Cousins construyen el intervalo ordenando los resultados posibles por el cociente de verosimilitudes
$P(n \,|\, \nu + b)/P(n \,|\, \nu_{best} + b)$. Así el intervalo nunca es no físico y pasa de forma natural
de límite superior a intervalo cuando hay señal. El precio: para $n = 0$ el límite es algo menos restrictivo.

SK usa el método **bayesiano con prior plano** en la tasa $\Gamma$, que incluye los sistemáticos (exposición, $\varepsilon$, fondo) como priors y combina 8 regiones de señal (SK-I a IV, cada uno con dos regiones de $P_{tot}$).

G. J. Feldman, R. D. Cousins, *A Unified Approach to the Classical Statistical Analysis of Small Signals*,
Phys. Rev. D 57, 3873 (1998), [arXiv:physics/9711021](https://arxiv.org/abs/physics/9711021)

```python
exposicion = 450 * p_total            # protones x año (10 protones por molécula)
tau_eps1   = exposicion / nu90
eps_impl   = tau_lim / tau_eps1

print('exposición             = {:.2e} protones año'.format(exposicion))
print('límite con eps = 1     = {:.1e} años'.format(tau_eps1))
print('eficiencia implícita   = {:.3f}'.format(eps_impl))

# Tabla IV de [SK2020], FV ampliado (27.2 kt), p -> e+ pi0: regiones Ptot < 100 y 100-250 MeV/c
expo_SK = np.array([92.1 + 19.3, 49.1 + 10.3, 31.9 + 6.7, 199.5 + 41.8])     # kt año: SK-I, II, III, IV
eps_SK  = np.array([18.3 + 20.0, 16.6 + 19.4, 18.7 + 20.3, 18.2 + 19.2]) / 100
eps_art = np.sum(eps_SK * expo_SK) / np.sum(expo_SK)

print('exposición tabla IV    = {:.1f} kt año'.format(np.sum(expo_SK)))
print('eficiencia artículo    = {:.3f}'.format(eps_art))
print('límite con eps art.    = {:.1e} años'.format(eps_art * exposicion / nu90))
print('solo H, misma eps      = {:.1e} años'.format(eps_art * 450 * p_libre / nu90))
```

<details>
<summary><b>Discusión</b></summary>

1. $\tau/\mathcal{B} > 6.5 \times 10^{34}$ años con $\varepsilon = 1$.
2. $\varepsilon \approx 0.37$: se pierde señal por los cortes de selección que rechazan el fondo atmosférico, por la
   absorción o dispersión del $\pi^0$ en el oxígeno...
3. El artículo da $\varepsilon \approx 0.38$ (promedio sobre los 10 protones), y con ella $\tau/\mathcal{B} > 2.5 \times 10^{34}$ años,
   casi el resultado de SK. La diferencia se debe a los sistemáticos, que SK incluye en el cálculo.
4. Sería 5 veces menor. No es correcto: la $\varepsilon$ de SK es un promedio sobre libres y ligados. Si solo se contaran los libres
   habría que usar su eficiencia, más alta, pero nunca compensaría perder 8 de cada 10 protones.

La estimación ignora el fondo, pero reproduce el límite del artículo.

</details>

## ¿Qué significa el 90%? Experimentos simulados

Repetimos el experimento de SK 10000 veces para varios valores de $\tau/\mathcal{B}$ y contamos
en qué fracción se observa $n = 0$.

```python
eps    = eps_impl
n_exp  = 10000
taus   = np.logspace(33.5, 35.5, 30)

frac0 = []
for tau_i in taus:
    nu_i = eps * exposicion / tau_i
    frac0.append(np.mean(rng.poisson(nu_i, size=n_exp) == 0))

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(taus, frac0, 'o', label='fracción de experimentos con $n = 0$')
ax.plot(taus, np.exp(-eps * exposicion / taus), label=r'$e^{-\nu(\tau)}$')
ax.axhline(0.10, color='gray', ls='--')
ax.axvline(tau_lim, color='C3', ls=':', label='límite SK')
ax.axvspan(taus[0], tau_lim, color='C3', alpha=0.15)
ax.set_xscale('log'); ax.set_xlabel(r'$\tau/\mathcal{B}$ (años)'); ax.set_ylabel('probabilidad')
ax.grid(alpha=0.3); ax.legend();
```

*Cuestión* ¿Por qué la región excluida está a la **izquierda** del límite? ¿Qué ocurre con valores de $\tau$ muy grandes?

## El límite crece con la exposición

Sin fondo y sin candidatos, el límite crece **linealmente** con la exposición, $\tau_{lim} \propto \varepsilon N_0 t$.

Con fondo $b$ el crecimiento se frena: en el mejor caso, $\propto \sqrt{N_0 t}$.

```python
años = np.linspace(0.5, 25, 100)

fig, ax = plt.subplots(figsize=(7, 4))
for M, nombre in [(22.5, 'SK, FV 22.5 kt'), (27.2, 'SK, FV 27.2 kt'), (187, 'HK, FV 187 kt')]:
    ax.plot(años, eps * M * p_total * años / nu90, label=nombre)
ax.axhline(tau_lim, color='gray', ls='--', label=r'SK 2020: $2.4 \times 10^{34}$ años')
ax.set_yscale('log'); ax.set_xlabel('años de toma de datos')
ax.set_ylabel(r'$\tau/\mathcal{B}$ límite 90% CL (años)')
ax.grid(alpha=0.3); ax.legend();
```

*Cuestión* ¿Por qué las curvas de SK a 22 años quedan **por encima** del límite real? Pista: años de calendario frente a tiempo vivo de toma de datos.

## Extensión: ¿y si hay fondo?

En $p \to \mu^+ \pi^0$ SK observa **1** candidato, compatible con el fondo esperado $b$.

Generalizando el argumento: $\nu_s$ (señal) está excluido al 90% CL si

$$
P(n \leq n_{obs} \,|\, \nu_s + b) = \sum_{k=0}^{n_{obs}} P(k \,|\, \nu_s + b) < 0.10
$$

<div class="admonition tip">
<p class="admonition-title"><b>[>] Taller — El límite con fondo</b></p>

1. Calcula el límite superior $\nu_{s}^{90}$ para $n_{obs} = 0, 1, 2$ en función del fondo $b$.
2. ¿Por qué el límite con 1 candidato es peor que con 0? Relaciónalo con $1.6$ frente a $2.4 \times 10^{34}$ años.
3. Fíjate en $n_{obs} = 0$ con $b$ grande. ¿Qué pasa? ¿Tiene sentido que un experimento con **más** fondo dé un límite **mejor**?

</div>

```python
from scipy.optimize import brentq

def nu_s_90(n_obs, b, cl=0.90):
    """Límite superior clásico en la señal: P(n <= n_obs | s + b) = 1 - cl."""
    f = lambda s: poisson.cdf(n_obs, s + b) - (1 - cl)
    return brentq(f, -b, 50)       # f(-b) = 0.9 > 0 y f(50) < 0

bs = np.linspace(0, 4, 81)
fig, ax = plt.subplots(figsize=(7, 4))
for n_obs in (0, 1, 2):
    ax.plot(bs, [nu_s_90(n_obs, b) for b in bs], label='$n_{{obs}}$ = {}'.format(n_obs))
ax.axhline(0, color='k', lw=0.8)
ax.set_xlabel('fondo esperado $b$'); ax.set_ylabel(r'límite superior $\nu_s$ (90% CL)')
ax.grid(alpha=0.3); ax.legend();
print('b = 0 : n = 0 -> {:.2f},  n = 1 -> {:.2f},  n = 2 -> {:.2f}'.format(*[nu_s_90(n, 0) for n in (0, 1, 2)]))
```

<details>
<summary><b>Discusión</b></summary>

1. Sin fondo: $\nu_s^{90} = 2.30, \, 3.89, \, 5.32$ para $n_{obs} = 0, 1, 2$. Con fondo, las rectas son simplemente $\nu_s^{90}(n_{obs}, b) = \nu^{90}(n_{obs}) - b$.
2. Un candidato más obliga a admitir más señal: el límite en $\nu$ sube y el de $\tau/\mathcal{B}$ baja. (La eficiencia
   de $\mu^+\pi^0$ también es algo distinta, así que el cociente no es exactamente $3.89/2.30$.)
3. Con $n_{obs} = 0$ y $b > 2.3$ el límite es **negativo**: el método clásico excluye toda señal, ¡incluso $\nu_s = 0$!
   Es un resultado sin sentido físico, que premia la "mala suerte" de observar menos que el fondo.

La solución moderna es la construcción de [Feldman y Cousins](https://arxiv.org/abs/physics/9711021) (1998),
dos físicos de neutrinos, que siempre da intervalos físicos y pasa de forma natural de límite a intervalo.
Ver también la revisión de estadística del [PDG](https://pdg.lbl.gov/2024/reviews/rpp2024-rev-statistics.pdf).

</details>

## Resumen

* SK tiene $\sim 10^{34}$ protones en su volumen fiducial. Aun así, con $\tau/\mathcal{B} \sim 10^{34}$ años se espera $\lesssim 1$ desintegración al año.

* SK cuenta los **10 protones** de cada molécula, suponiendo que todos se desintegran por igual; libres y ligados solo se distinguen en la eficiencia.

* El número de desintegraciones observadas es una variable de **Poisson** de media $\nu = \varepsilon N_0 t / \tau$.

* 10 desintegraciones en uno o dos años darían $\tau/\mathcal{B} \sim 10^{33}$ años con un 30% de error, excluido por SK.

* Si no se observa nada, $\nu < 2.30$ al 90% CL, y $\tau/\mathcal{B} > \varepsilon N_0 t / 2.30$.

* El límite crece con la **exposición** (kt·año): de ahí Hyper-Kamiokande.

* Con fondo, el método clásico da resultados patológicos: Feldman-Cousins.
