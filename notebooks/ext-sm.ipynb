## 1. De $U(1)$ a $SU(2) \times U(1)$

### 1.1. Lo que hay que reproducir

* El $W$ solo se acopla a la parte **izquierda** de los fermiones, y siempre a **parejas**: $\nu_e \leftrightarrow e$, $u \leftrightarrow d$.
* El fotón se acopla igual a las dos partes, con la carga.
* Existe un bosón neutro más, el $Z$, que ve las dos partes con intensidades distintas.

### 1.2. Dobletes y singletes

Las partes izquierdas forman **dobletes** de isoespín débil, $T = \frac{1}{2}$; las derechas, **singletes**, $T = 0$:

$$
L = \begin{pmatrix} \nu_e \\ e \end{pmatrix}_L, \quad e_R; \qquad
Q_L = \begin{pmatrix} u \\ d \end{pmatrix}_L, \quad u_R, \quad d_R .
$$

La simetría $SU(2)_L$ gira los dos miembros de cada doblete entre sí, $L \to e^{i\alpha_a(x) T_a}L$, y deja quietos los
singletes. Una partícula y su compañera de doblete son, para esta simetría, lo mismo.

**Sin masa, por ahora**: con esta asignación, ningún fermión puede tener masa (sección 1.4).

### 1.3. La hipercarga

El fotón no puede ser el bosón de la $SU(2)$: se acopla igual a $e_L$ y a $e_R$, que están en sitios distintos. Hace falta
otra $U(1)$, la de **hipercarga** $Y$, con su propio bosón, $B_\mu$. La carga eléctrica es

$$
\boxed{Q = T_3 + Y}
$$

| | $\nu_L$ | $e_L$ | $e_R$ | $u_L$ | $d_L$ | $u_R$ | $d_R$ | $\Phi$ (sección 4) |
| :-- | :--: | :--: | :--: | :--: | :--: | :--: | :--: | :--: |
| $T_3$ | $+\frac{1}{2}$ | $-\frac{1}{2}$ | $0$ | $+\frac{1}{2}$ | $-\frac{1}{2}$ | $0$ | $0$ | $\pm\frac{1}{2}$ |
| $Y$ | $-\frac{1}{2}$ | $-\frac{1}{2}$ | $-1$ | $+\frac{1}{6}$ | $+\frac{1}{6}$ | $+\frac{2}{3}$ | $-\frac{1}{3}$ | $+\frac{1}{2}$ |
| $Q$ | $0$ | $-1$ | $-1$ | $+\frac{2}{3}$ | $-\frac{1}{3}$ | $+\frac{2}{3}$ | $-\frac{1}{3}$ | $+1$, $0$ |

Los dos miembros de un doblete tienen la misma $Y$ (la $U(1)$ no distingue arriba de abajo); su carga difiere en una
unidad, la que lleva el $W$.

### 1.4. El término cinético y sus simetrías globales

Sin masa, el lagrangiano libre de la primera familia de leptones, escrito con el doblete y el singlete, es

$$
\mathcal{L}_0 = \bar L\, i\gamma^\mu\partial_\mu L + \bar e_R\, i\gamma^\mu\partial_\mu e_R .
$$

El primer término es un producto escalar en el espacio del doblete, con $\bar L = (\bar\nu_L,\ \bar e_L)$:

$$
\bar L\, i\gamma^\mu\partial_\mu L
= (\bar\nu_L,\ \bar e_L)\; i\gamma^\mu\partial_\mu \begin{pmatrix} \nu_L \\ e_L \end{pmatrix}
= \bar\nu_L\, i\gamma^\mu\partial_\mu\nu_L + \bar e_L\, i\gamma^\mu\partial_\mu e_L .
$$

Y como el término vectorial no mezcla las quiralidades ([ext-quiral](./ext-quiral.ipynb), sección 9.3),
$\bar e\, i\gamma^\mu\partial_\mu e = \bar e_L\, i\gamma^\mu\partial_\mu e_L + \bar e_R\, i\gamma^\mu\partial_\mu e_R$. Sumando:

$$
\boxed{
\mathcal{L}_0 = \bar\nu_L\, i\gamma^\mu\partial_\mu\nu_L + \bar e\, i\gamma^\mu\partial_\mu e
}
$$

un neutrino a izquierdas y un electrón de Dirac, los dos sin masa: la misma física de siempre, agrupada de otra
manera. La ventaja es que ahora se ven las simetrías.

#### Las dos simetrías globales

**$SU(2)$.** Un giro en el espacio del doblete, $L \to U L$, con $U = e^{i\alpha_a T_a}$ unitaria y constante, y $e_R \to e_R$. $U$ actúa
sobre el índice del doblete, no sobre el del espinor: conmuta con las $\gamma^\mu$, y $\bar L \to \bar L\, U^\dagger$. Entonces

$$
\bar L\, i\gamma^\mu\partial_\mu L \;\to\; \bar L\, U^\dagger\, i\gamma^\mu\partial_\mu (U L) = \bar L\, U^\dagger U\, i\gamma^\mu\partial_\mu L = \bar L\, i\gamma^\mu\partial_\mu L .
$$

Es invariante, aunque $U$ mezcle $\nu_L$ con $e_L$: **para esta simetría, el neutrino y la parte izquierda del electrón son el
mismo objeto**. Cada término por separado, $\bar\nu_L\,i\gamma\partial\nu_L$ o $\bar e_L\,i\gamma\partial e_L$, no lo es.

**$U(1)_Y$.** Un cambio de fase, con la hipercarga de cada campo: $L \to e^{iY_L\beta}L$, $e_R \to e^{iY_R\beta}e_R$, con $Y_L = -\frac{1}{2}$ e
$Y_R = -1$. Cada término es invariante por separado: la fase de $\psi$ se cancela con la de $\bar\psi$.

**El término de masa no es invariante.** $m\,\bar e e = m(\bar e_L e_R + \bar e_R e_L)$:

* bajo $SU(2)$, $e_L$ se mezcla con $\nu_L$ y $e_R$ no: $\bar e_L e_R$ no se transforma en sí mismo;
* bajo $U(1)_Y$, $\bar e_L e_R \to e^{i(Y_R - Y_L)\beta}\,\bar e_L e_R = e^{-i\beta/2}\,\bar e_L e_R$: las dos hipercargas no se cancelan.

Por eso, en $SU(2) \times U(1)$, **todos los fermiones empiezan sin masa**.

### 1.5. De global a local: cuatro bosones

Si el giro depende del punto, $U(x)$, el término cinético deja de ser invariante: $\partial_\mu(UL) = U\,\partial_\mu L + (\partial_\mu U)\,L$, y sobra
el segundo sumando. Como en perspectivas, se arregla con una **derivada covariante**, ahora con un bosón por cada
generador: tres $W^a_\mu$ para $SU(2)$ y uno, $B_\mu$, para $U(1)$:

$$
\boxed{
D_\mu = \partial_\mu + i g\,T_a W^a_\mu + i g'\,Y B_\mu
}
$$

Para un singlete, $T_a = 0$ y solo queda $B$. Los términos cinéticos $\bar\psi\,i\gamma^\mu D_\mu\psi$ de todos los fermiones dan sus
interacciones:

$$
\mathcal{L}_{\text{int}} = -\bar\psi\,\gamma^\mu\left(g\,T_a W^a_\mu + g'\,Y B_\mu\right)\psi .
$$

Dos acoplos, $g$ y $g'$, para todas las interacciones electrodébiles de todos los fermiones.

## 2. Los bosones físicos y sus acoplos

### 2.1. Los $W^\pm$

Con $T_\pm = T_1 \pm iT_2$, los dos primeros bosones se combinan en

$$
T_1 W^1 + T_2 W^2 = \frac{1}{\sqrt{2}}\left(T_+ W^+ + T_- W^-\right), \qquad
W^\pm \equiv \frac{1}{\sqrt{2}}\left(W^1 \mp iW^2\right) .
$$

$T_+$ sube del miembro de abajo al de arriba. Para la primera familia de leptones:

$$
-\frac{g}{\sqrt{2}}\,\bar L\gamma^\mu T_+ L\,W^+_\mu = -\frac{g}{\sqrt{2}}\,\bar\nu_L\gamma^\mu e_L\,W^+_\mu ,
$$

el vértice $e \to \nu_e$ con un $W$, con acoplo $g/\sqrt{2}$ y **solo a izquierdas**: la violación máxima de la paridad sale por
construcción. Es el vértice de [leptones](./leptones.ipynb).

### 2.2. La mezcla: el $Z$ y el fotón

Quedan dos bosones neutros, $W^3$ y $B$, que se acoplan a $T_3$ y a $Y$. Ninguno de los dos es el fotón: los físicos son
dos combinaciones ortogonales, giradas un ángulo $\theta_W$,

$$
\boxed{
\begin{pmatrix} Z \\ A \end{pmatrix} =
\begin{pmatrix} c_W & -s_W \\ s_W & c_W \end{pmatrix}
\begin{pmatrix} W^3 \\ B \end{pmatrix}
}
\qquad \Longleftrightarrow \qquad
W^3 = c_W Z + s_W A, \quad B = -s_W Z + c_W A .
$$

Sustituyendo en la parte neutra de $D_\mu$:

$$
g\,T_3 W^3 + g' Y B = A\left(g s_W T_3 + g' c_W Y\right) + Z\left(g c_W T_3 - g' s_W Y\right) .
$$

**El fotón tiene que acoplarse a la carga**, $eQ = e(T_3 + Y)$. Eso exige

$$
\boxed{
g\,s_W = g'\,c_W = e
}
\qquad \Longrightarrow \qquad \tan\theta_W = \frac{g'}{g}, \quad g = \frac{e}{s_W} .
$$

Es la relación $g_W = e/\sin\theta_W$ del tema: $\theta_W$ mide cuánto pesa la hipercarga frente al isoespín.

### 2.3. Los acoplos del $Z$

En el coeficiente del $Z$, con $Y = Q - T_3$ y $g' = g\,s_W/c_W$:

$$
g c_W T_3 - g' s_W (Q - T_3) = \frac{g}{c_W}\left[T_3\,(c_W^2 + s_W^2) - s_W^2\,Q\right] = g_Z\left(T_3 - Q\,s_W^2\right),
\qquad g_Z \equiv \frac{g}{c_W} .
$$

Para la parte izquierda, $T_3 = \pm\frac{1}{2}$; para la derecha, $T_3 = 0$:

$$
\boxed{
c_L = T_3 - Q\,s_W^2, \qquad c_R = -Q\,s_W^2
}
$$

la tabla de la sección 1.1 del tema, ahora deducida. Dos consecuencias:

* El $Z$ se acopla a los neutrinos ($T_3 = \frac{1}{2}$, $Q = 0$), aunque no tengan carga.
* Solo hay $Z$ entre un fermión y él mismo: $T_3$ y $Q$ son diagonales. **No hay corrientes neutras que cambien el sabor** a
  orden árbol (sección 5.3).

<details>
<summary><b>[+] Mirador — Por qué hay tres colores: las anomalías</b></summary>

**La cuenta.** Suma las cargas eléctricas de todas las partículas de la primera familia, contando los colores:

$$
\underbrace{0}_{\nu_e} + \underbrace{(-1)}_{e} + 3\times\underbrace{\tfrac{2}{3}}_{u} + 3\times\underbrace{\left(-\tfrac{1}{3}\right)}_{d} = 0 .
$$

**La familia es neutra.** Con $N_c$ colores la suma sería $-1 + N_c/3$: solo se anula con **tres**.

**Por qué tiene que ser cero.** No es una casualidad, sino una condición de consistencia. Una **anomalía** es una
simetría del lagrangiano que se pierde al cuantizar. Ocurre en un diagrama con un lazo de fermiones en forma de
**triángulo**, con un bosón *gauge* en cada vértice; cada fermión que circula por el lazo contribuye con el producto de
sus tres cargas. En una teoría *gauge* la simetría no se puede perder: es la que garantiza que las probabilidades
sumen uno y que los infinitos se puedan absorber (la renormalización). Así que **la suma sobre todos los fermiones
tiene que anularse**.

**Por qué es la carga.** En el triángulo con dos $W$ y un $B$, cada doblete izquierdo contribuye con su hipercarga, y la
condición es $\sum Y = 0$ sobre los dobletes: $-\frac{1}{2}$ (leptones) $+ N_c \times \frac{1}{6}$ (quarks) $= 0$, otra vez $N_c = 3$. Como $Q = T_3 + Y$ y $T_3$
se cancela dentro de cada doblete, pedir esto es lo mismo que pedir que la familia sea neutra.

**Lo que explica.**

* **Por qué las familias vienen completas:** los leptones solos, o los quarks solos, no forman una teoría consistente.
  Cada familia de leptones necesita su familia de quarks (la pregunta abierta de [leptones](./leptones.ipynb)).
* **La carga del protón.** Si el electrón tiene carga $-1$ y dentro de un doblete la carga cambia en una unidad, la
  condición fija $Q_d = -\frac{1}{3}$ y $Q_u = +\frac{2}{3}$. El protón, $uud$, tiene carga $+1$: **exactamente** la opuesta a la del electrón. Por eso
  la materia es neutra.

Hay más triángulos (con tres $B$, con gluones, con gravitones), y todos se anulan familia a familia. El taller comprueba
dos.

## 3. El modelo abeliano de Higgs

### 3.1. Un bosón con masa rompe la simetría

Antes del $W$ y del $Z$, el caso más sencillo: un fotón y un escalar complejo $\phi$ de carga $q$. Un término de masa para el
fotón, $\frac{1}{2}m^2 A_\mu A^\mu$, no es invariante bajo $A_\mu \to A_\mu - \partial_\mu\theta$. La simetría prohíbe la masa (perspectivas, 2.4).

El escalar, con $D_\mu = \partial_\mu + iqA_\mu$ y el potencial del tema,

$$
\mathcal{L} = (D_\mu\phi)^*(D^\mu\phi) - V(\phi), \qquad V = \mu^2\,\phi^*\phi + \lambda\,(\phi^*\phi)^2, \quad \lambda > 0 ,
$$

es invariante bajo $\phi \to e^{iq\theta(x)}\phi$, $A_\mu \to A_\mu - \partial_\mu\theta$, la misma transformación que en perspectivas.

### 3.2. La rotura

Con $\mu^2 < 0$, el mínimo está en el círculo $|\phi|^2 = -\mu^2/2\lambda \equiv v^2/2$, con $v^2 = -\mu^2/\lambda$ (el sombrero del tema, sección
3.3). Desarrollando alrededor de un punto del círculo, el real,

$$
\phi = \frac{1}{\sqrt{2}}\left(v + h + i\chi\right),
$$

con $h$ en la dirección radial y $\chi$ a lo largo del fondo, el potencial queda

$$
V = -\frac{\lambda v^4}{4} + \lambda v^2\,h^2 + \lambda v\,h\,(h^2 + \chi^2) + \frac{\lambda}{4}(h^2 + \chi^2)^2 .
$$

* $h$ tiene masa: $\frac{1}{2}m_h^2 = \lambda v^2$, **$m_h^2 = 2\lambda v^2$**.
* $\chi$ no aparece en ningún término cuadrático: **no tiene masa**. Es el **bosón de Goldstone**: moverse por el fondo del
  sombrero no cuesta energía.

Sin el fotón, este sería el final: una partícula con masa y otra sin ella (teorema de Goldstone).

### 3.3. El fotón se come el Goldstone

Con el fotón, la fase de $\phi$ se puede quitar con una transformación *gauge*: escribiendo $\phi = e^{i\chi/v}(v + h)/\sqrt{2}$ y
eligiendo $\theta = -\chi/qv$, el campo queda real en todas partes (la **gauge unitaria**). $\chi$ desaparece. El término
cinético da

$$
(D_\mu\phi)^*(D^\mu\phi) = \frac{1}{2}\partial_\mu h\,\partial^\mu h + \frac{q^2}{2}\,(v + h)^2 A_\mu A^\mu ,
$$

y en él:

* $\frac{1}{2}(qv)^2 A_\mu A^\mu$: **un término de masa para el fotón**, $m_A = qv$;
* $q^2 v\,h\,A_\mu A^\mu$ y $\frac{q^2}{2}h^2 A_\mu A^\mu$: vértices $hAA$ y $hhAA$, con acoplo proporcional a $m_A^2$.

**Grados de libertad.** Antes: $2$ (el escalar complejo) $+ 2$ (las polarizaciones de un fotón sin masa) $= 4$. Después:
$1$ ($h$) $+ 3$ (un vector con masa tiene tres polarizaciones) $= 4$. El Goldstone no se ha perdido: es la
**polarización longitudinal** del fotón. Es el mecanismo de un superconductor (el mirador del tema).

## 4. El doblete de Higgs

### 4.1. Por qué un doblete

Para dar masa a **tres** bosones ($W^+$, $W^-$, $Z$) hacen falta tres Goldstone, y para dar masa a los fermiones el campo
tiene que poder unir un doblete con un singlete (sección 5). Las dos cosas las hace un **doblete de escalares
complejos**, con $Y = +\frac{1}{2}$:

$$
\Phi = \begin{pmatrix} \phi^+ \\ \phi^0 \end{pmatrix}, \qquad
\mathcal{L}_\Phi = (D_\mu\Phi)^\dagger(D^\mu\Phi) - \mu^2\,\Phi^\dagger\Phi - \lambda\,(\Phi^\dagger\Phi)^2 .
$$

Cuatro campos reales. Con $\mu^2 < 0$, el mínimo está en $\Phi^\dagger\Phi = v^2/2$. En la *gauge* unitaria, los tres Goldstone se
absorben y queda

$$
\boxed{
\Phi = \frac{1}{\sqrt{2}}\begin{pmatrix} 0 \\ v + h \end{pmatrix}
}
$$

El valor en el vacío está en la componente de abajo, la de **carga cero** ($T_3 = -\frac{1}{2}$, $Y = +\frac{1}{2}$): el vacío no tiene
carga eléctrica.

### 4.2. Las masas del $W$ y del $Z$

Con $T_a W^a = \frac{1}{2}\begin{pmatrix} W^3 & \sqrt{2}\,W^+ \\ \sqrt{2}\,W^- & -W^3 \end{pmatrix}$ y $Y = \frac{1}{2}$, sobre $\Phi = (0, \varphi)$, $\varphi = (v + h)/\sqrt{2}$:

$$
\left(g\,T_a W^a + g' Y B\right)\Phi = \frac{1}{2}\begin{pmatrix} \sqrt{2}\,g\,W^+ \\ -g W^3 + g' B \end{pmatrix}\varphi
= \frac{1}{2}\begin{pmatrix} \sqrt{2}\,g\,W^+ \\ -g_Z\,Z \end{pmatrix}\varphi ,
$$

porque $-gW^3 + g'B = -(g c_W + g' s_W)\,Z + (g' c_W - g s_W)\,A = -g_Z\,Z$: **el fotón no aparece**, gracias a $g s_W = g' c_W$. El
término cinético da

$$
(D_\mu\Phi)^\dagger(D^\mu\Phi) = \frac{1}{2}\partial_\mu h\,\partial^\mu h
+ \frac{g^2}{4}\,(v + h)^2\,W^+_\mu W^{-\mu} + \frac{g_Z^2}{8}\,(v + h)^2\,Z_\mu Z^\mu ,
$$

y comparando con los términos de masa, $m_W^2 W^+W^-$ y $\frac{1}{2}m_Z^2 ZZ$:

$$
\boxed{
m_W = \frac{g v}{2}, \qquad m_Z = \frac{g_Z v}{2} = \frac{m_W}{c_W}, \qquad m_A = 0
}
$$

### 4.3. Lo que sale además

* **$m_W/m_Z = \cos\theta_W$**, una predicción: no hay parámetros nuevos para las masas de los bosones. Se suele escribir
  $\rho \equiv m_W^2/(m_Z^2 c_W^2) = 1$; se cumple al $10^{-3}$, la diferencia son lazos (en ellos, el top y el Higgs: el mirador
  de la sección 2.5 del tema).
* **$v$ a partir de $G_F$.** A bajas energías el propagador del $W$ es $1/m_W^2$, y $\frac{G_F}{\sqrt{2}} = \frac{g^2}{8 m_W^2} = \frac{1}{2v^2}$:
  $v = (\sqrt{2}\,G_F)^{-1/2} = 246$ GeV.
* **Los vértices del Higgs con los bosones**, desarrollando $(v + h)^2 = v^2 + 2vh + h^2$:

$$
\frac{2 m_W^2}{v}\,h\,W^+_\mu W^{-\mu} + \frac{m_Z^2}{v}\,h\,Z_\mu Z^\mu
+ \frac{m_W^2}{v^2}\,h^2\,W^+_\mu W^{-\mu} + \frac{m_Z^2}{2v^2}\,h^2\,Z_\mu Z^\mu :
$$

  el Higgs se acopla a los bosones **en proporción al cuadrado de su masa** (la figura de ATLAS de la sección 4.4 del
  tema, con $\sqrt{\kappa_V}\,m_V/v$).
* **Cuenta de grados de libertad**, en el taller.

<div class="admonition tip">
<p class="admonition-title"><b>[>] Taller — La masa del fotón es cero</b></p>

La parte neutra del término cinético, $\frac{g_Z^2}{8}v^2 Z^2$, escrita con $W^3$ y $B$, es $\frac{v^2}{8}(gW^3 - g'B)^2 = \frac{1}{2}V^T M^2 V$, con $V = (W^3, B)$ y

$$
M^2 = \frac{v^2}{4}\begin{pmatrix} g^2 & -g g' \\ -g g' & g'^2 \end{pmatrix} .
$$

1. Ejecuta la celda: con $g$ y $g'$ a partir de $e$ y $\sin^2\theta_W$, diagonaliza $M^2$. ¿Cuáles son las masas? ¿Qué ángulo forma
   el autovector sin masa con el eje de $B$?
2. ¿Por qué el determinante de $M^2$ tiene que ser cero, sean cuales sean $g$ y $g'$?
3. Cuenta los grados de libertad antes de la rotura (el doblete y cuatro bosones sin masa) y después ($h$, tres bosones
   con masa y el fotón).

</div>

```python
# matriz de masas de (W3, B) y sus autovectores
alfa, s2w, v = 1 / 128.9, 0.2315, 246.22
e  = np.sqrt(4 * np.pi * alfa)
g, gp = e / np.sqrt(s2w), e / np.sqrt(1 - s2w)
M2 = v**2 / 4 * np.array([[g**2, -g * gp], [-g * gp, gp**2]])
m2, vecs = np.linalg.eigh(M2)
print(f' g = {g:.3f}, g\' = {gp:.3f}')
for m2i, vec in zip(m2, vecs.T):
    print(f' masa = {np.sqrt(abs(m2i)):6.2f} GeV,  autovector (W3, B) = ({vec[0]:+.3f}, {vec[1]:+.3f})')
foton = vecs[:, 0]
print(f' angulo del foton con B: {np.degrees(np.arccos(abs(foton[1]))):.2f} grados;  theta_W = {np.degrees(np.arcsin(np.sqrt(s2w))):.2f} grados')
print(f' m_W = g v / 2 = {g * v / 2:.2f} GeV')
```

## 5. Las masas de los fermiones

### 5.1. El electrón

$\bar L\,\Phi$ es invariante bajo $SU(2)$ (un doblete por otro) y tiene $Y = +\frac{1}{2} + \frac{1}{2} = 1$; multiplicado por $e_R$ ($Y = -1$),
el producto es invariante del todo. Es el **acoplo de Yukawa**:

$$
\mathcal{L}_Y = -y_e\left(\bar L\,\Phi\,e_R + \bar e_R\,\Phi^\dagger L\right) .
$$

En la *gauge* unitaria, $\bar L\Phi = \bar e_L\,(v + h)/\sqrt{2}$, y

$$
\mathcal{L}_Y = -\frac{y_e}{\sqrt{2}}(v + h)\left(\bar e_L e_R + \bar e_R e_L\right) = -\underbrace{\frac{y_e v}{\sqrt{2}}}_{m_e}\,\bar e e - \frac{m_e}{v}\,h\,\bar e e .
$$

**La masa y el acoplo al Higgs salen del mismo término**: $m_e = y_e v/\sqrt{2}$, y el vértice $h\bar e e$ vale $m_e/v$. Es la
ecuación de la sección 3.2 del tema.

**El neutrino no coge masa**: haría falta un $\nu_R$, que en el Modelo Estándar no existe.

### 5.2. Los quarks

El $d$ es como el electrón: $-y_d\,\bar Q_L\,\Phi\,d_R + \text{h.c.}$, con $Y = -\frac{1}{6} + \frac{1}{2} - \frac{1}{3} = 0$. Para el $u$, que está arriba del
doblete, hace falta un doblete con el valor en el vacío **arriba**: el conjugado,

$$
\tilde\Phi \equiv i\sigma_2\,\Phi^* = \frac{1}{\sqrt{2}}\begin{pmatrix} v + h \\ 0 \end{pmatrix}, \qquad Y = -\frac{1}{2} ,
$$

y el término $-y_u\,\bar Q_L\,\tilde\Phi\,u_R + \text{h.c.}$ ($Y = -\frac{1}{6} - \frac{1}{2} + \frac{2}{3} = 0$). Los dos dan

$$
m_u = \frac{y_u v}{\sqrt{2}}, \qquad m_d = \frac{y_d v}{\sqrt{2}} ,
$$

y los vértices $h\bar u u$ y $h\bar d d$ con $m_q/v$. **Un acoplo de Yukawa libre por fermión cargado**: nueve con las tres
familias. El Higgs explica el mecanismo, no los valores.

### 5.3. Tres familias: de dónde sale la CKM

Con tres familias, nada impide que el Higgs una un $\bar Q_{L,i}$ de una familia con el $d_{R,j}$ de otra: los Yukawas son
**matrices** $3\times 3$, y las de masa, $M^d = y^d v/\sqrt{2}$ y $M^u = y^u v/\sqrt{2}$, no son diagonales. Se diagonalizan con matrices
unitarias distintas para la parte izquierda y la derecha:

$$
d_L = U^d_L\,d'_L, \quad u_L = U^u_L\,u'_L, \qquad (U^d_L)^\dagger M^d\, U^d_R = \mathrm{diag}(m_d, m_s, m_b) ,
$$

donde $d'$, $u'$ son los quarks con masa definida. En la corriente del $W$:

$$
\bar u_L\gamma^\mu d_L = \bar u'_L\gamma^\mu \underbrace{(U^u_L)^\dagger U^d_L}_{V_{\text{CKM}}} d'_L .
$$

**La CKM es el desajuste entre cómo diagonalizan las masas los quarks de arriba y los de abajo** (el tema de
[hadrones](./quarks_hadrons.ipynb), parte 4). En las corrientes del $Z$ y del fotón, $\bar d_L\gamma^\mu d_L = \bar d'_L\gamma^\mu (U^d_L)^\dagger U^d_L\,d'_L$,
y el producto es la identidad: **no cambian el sabor**. Con neutrinos sin masa, en los leptones no hay mezcla; la PMNS
de las oscilaciones pide ir más allá del Modelo Estándar.

## 6. El potencial desarrollado

Con $\Phi^\dagger\Phi = (v + h)^2/2$ y $\mu^2 = -\lambda v^2$, el potencial es

$$
V = \mu^2\,\frac{(v + h)^2}{2} + \lambda\,\frac{(v + h)^4}{4}
= -\frac{\lambda v^4}{4} + \lambda v^2\,h^2 + \lambda v\,h^3 + \frac{\lambda}{4}\,h^4 .
$$

Es exactamente el $V(\phi) = \mu^2\phi^2/2 + \lambda\phi^4/4$ del tema con $\phi = v + h$: la simplificación del tema no pierde nada.

* El término lineal en $h$ se anula: $v$ es el mínimo.
* $\lambda v^2 h^2 = \frac{1}{2}m_H^2 h^2$: **$m_H^2 = 2\lambda v^2$**. Con $m_H = 125.2$ GeV, $\lambda = 0.13$.
* $\lambda v\,h^3 = \dfrac{m_H^2}{2v}\,h^3$ y $\dfrac{\lambda}{4}h^4 = \dfrac{m_H^2}{8v^2}\,h^4$: los **autoacoplos**, fijados por $m_H$ y $v$. Medir el de tres
  Higgs, en $p + p \to H + H$, es comprobar la forma del potencial (el mirador de la sección 3.5 del tema).

## 7. Los parámetros

| sector | parámetros del lagrangiano | lo que se mide mejor |
| :-- | :-- | :-- |
| *gauge* | $g$, $g'$, $g_s$ | $\alpha$, $\sin^2\theta_W$ (o $m_Z$), $\alpha_S$ |
| Higgs | $\mu^2$, $\lambda$ | $G_F$ (es decir, $v$), $m_H$ |
| Yukawa | $9$ acoplos | las masas de los $9$ fermiones cargados |
| mezcla | $3$ ángulos y $1$ fase | la CKM |
| QCD | $\theta_{QCD}$ | $< 10^{-10}$ (momento dipolar del neutrón) |

**Diecinueve.** Todo lo demás —$m_W$, $m_Z/m_W$, los acoplos del $Z$, las anchuras, los acoplos y las desintegraciones del
Higgs— es predicción. Con neutrinos con masa hacen falta al menos siete más.
