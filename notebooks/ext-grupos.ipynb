## Introducción

Esta extensión es el **desarrollo matemático de las simetrías internas** del Modelo Estándar, $U(1)$, $SU(2)$ y
$SU(3)$, que los temas usan sin demostrar: la carga y el fotón ([perspectivas](./perspectivas.ipynb),
[ext-dirac](./ext-dirac.ipynb)), el isoespín débil ([leptones](./leptones.ipynb), [ext-sm](./ext-sm.ipynb)), el
color y los multipletes de hadrones ([quarks y hadrones](./quarks_hadrons.ipynb)).

**Simetrías internas.** Son transformaciones que mezclan las etiquetas «de dentro» de un campo, su fase, su sabor o su
color, en el mismo punto del espacio-tiempo, sin tocar $t$ ni ${\bf x}$. Fijan las **cargas** y las **interacciones**. Son
distintas de las simetrías del **espacio-tiempo**, las traslaciones, las rotaciones y los *boosts*, que mueven los
puntos y mezclan las componentes de los vectores y del espín, y fijan la energía, el momento y el espín. Las dos se
describen con la misma matemática, los grupos de Lie, pero **esta extensión trata de las internas**. Tiene siete
secciones sobre ellas y una octava, aparte, que mira el grupo de Lorentz con las mismas herramientas; cada una
responde a una pregunta:

1. **Grupos, generadores y representaciones.** *¿Qué es una simetría continua?* Unas matrices que se obtienen
   exponenciando unos pocos generadores. Sus autovalores son los números cuánticos.
2. **$U(1)$: la fase y la carga.** *¿Qué simetría da la carga eléctrica?* La más sencilla: una fase.
3. **$SU(2)$: el espín y el isoespín.** *¿Por qué dobletes y tripletes?* Las representaciones de $SU(2)$ y cómo se
   combinan.
4. **$SU(3)$: ocho generadores.** *¿Cómo se dibujan sus representaciones?* En el plano $(T_3, Y)$: triángulos,
   hexágonos y triángulos grandes.
5. **El color.** *¿Por qué solo $qqq$ y $q\bar q$?* Son las combinaciones que pueden ser singletes de color.
6. **El sabor: el camino óctuple.** *¿Por qué los hadrones ligeros vienen en octetes y decupletes?* Por una simetría
   aproximada entre $u$, $d$ y $s$.
7. **De global a local: un bosón por generador.** *¿Por qué un fotón, tres $W$ y ocho gluones?* Y por qué los
   gluones interaccionan entre sí y el fotón no.
8. **Aparte: el grupo de Lorentz.** *¿Qué son, para un grupo, un escalar, un cuadrivector y un espinor?*
   Representaciones de una simetría del espacio-tiempo, no interna. Y por qué hace falta $\bar\Psi$.

Convenios: $\hbar = c = 1$; generadores hermíticos, $U = e^{i\theta_a T_a}$, con suma sobre los índices repetidos.

## 1. Grupos, generadores y representaciones

### 1.1. Simetrías y grupos

Un **grupo** es un conjunto de transformaciones que se pueden componer: el producto de dos es otra del grupo, hay
una identidad, cada una tiene su inversa, y el producto es asociativo. Si el orden no importa, $g_1 g_2 = g_2 g_1$, el
grupo es **abeliano**.

Las simetrías de la física forman grupos: si dos transformaciones dejan la física igual, aplicar una tras otra
también. Las hay **discretas**, como la paridad y la conjugación de carga ([ext-quiral](./ext-quiral.ipynb),
sección 2), y **continuas**, que dependen de parámetros reales, como un ángulo. Las continuas y derivables son los
**grupos de Lie**: las rotaciones, las transformaciones de Lorentz y los tres de esta extensión.

**Simetría y conservación** (E. Noether, 1918). Cada simetría continua del lagrangiano da una cantidad
conservada: la traslación en el tiempo, la energía; las rotaciones, el momento angular; una fase, una carga
(sección 2).

### 1.2. Generadores

Cerca de la identidad, $U \simeq I + i\theta_a T_a$: los $T_a$ son los **generadores**. Una transformación finita es
muchas pequeñas seguidas:

$$
U(\theta) = \lim_{N \to \infty}\left(I + \frac{i\theta_a T_a}{N}\right)^N = e^{i\theta_a T_a} .
$$

* **Si $U$ es unitaria, los generadores son hermíticos.** Son observables, y sus autovalores, **números
  cuánticos**.
* **El álgebra.** Que el grupo sea o no abeliano lo decide el conmutador de los generadores,
  $[T_a, T_b] = i f_{abc}\,T_c$; las $f_{abc}$ son las **constantes de estructura**. Si son todas nulas, el grupo es
  abeliano.
* **El rango.** Los generadores que conmutan entre sí se diagonalizan a la vez y etiquetan los estados con números
  cuánticos **aditivos**. Cuántos hay es el rango del grupo: $1$ en $SU(2)$ ($T_3$) y $2$ en $SU(3)$ ($T_3$ y $T_8$).

**Cuántos generadores.** $SU(n)$ son las matrices $n \times n$ unitarias (U) con determinante $1$ (S, *special*).
Una matriz compleja $n \times n$ tiene $2n^2$ parámetros reales; $U^\dagger U = I$ impone $n^2$ condiciones, y
$\det U = 1$ una más. Quedan $n^2 - 1$: **$3$ en $SU(2)$, $8$ en $SU(3)$**, y $1$ en $U(1)$. Como
$\det e^{iT} = e^{i\,\mathrm{tr}\,T}$, los generadores de $SU(n)$ tienen traza nula.

### 1.3. Representaciones

El mismo grupo actúa sobre objetos de distinta dimensión: las rotaciones giran un espinor de Pauli (espín $\frac{1}{2}$,
dos componentes) con matrices $2 \times 2$ y un vector con matrices $3 \times 3$. El espinor de Dirac, con cuatro
componentes, se gira con matrices $4 \times 4$, $S(\Lambda)$; pero en la base quiral son dos bloques $2 \times 2$ que no se
mezclan, uno por quiralidad ([ext-quiral](./ext-quiral.ipynb), sección 6): es **reducible**, y solo la paridad, que
intercambia los bloques, lo une. Cada conjunto de matrices que se multiplican como los elementos del grupo es una
**representación**; los vectores sobre los que actúan forman su base, y son los
**estados** de las partículas.

* **Irreducible:** no se puede partir en bloques que no se mezclen. Las partículas son las bases de las
  representaciones irreducibles, que se nombran por su dimensión: $\mathbf{1}$ (singlete), $\mathbf{2}$ (doblete),
  $\mathbf{3}$ (triplete), $\mathbf{8}$ (octete)...
* **Fundamental:** la más pequeña que no es trivial, $\mathbf{2}$ en $SU(2)$ y $\mathbf{3}$ en $SU(3)$. Los quarks
  están en ella. Su **conjugada**, con $U^*$, es la de los antiquarks: $\bar{\mathbf{3}}$.
* **Adjunta:** su dimensión es el número de generadores, $\mathbf{3}$ en $SU(2)$ y $\mathbf{8}$ en $SU(3)$. Los
  bosones *gauge* están en ella (sección 7).
* **Singlete:** no cambia. Es lo que no lleva la carga de ese grupo: un hadrón es un singlete de color.
* **Combinar** dos sistemas es multiplicar sus representaciones, $\otimes$, y descomponer el producto en
  irreducibles, $\oplus$: $\mathbf{2} \otimes \mathbf{2} = \mathbf{3} \oplus \mathbf{1}$ (sección 3). Las dimensiones
  cuadran, $2 \times 2 = 3 + 1$.
* **Casimir:** un operador que conmuta con todos los generadores, como ${\bf J}^2$. Vale lo mismo en toda una
  representación irreducible, y la etiqueta.

## 2. $U(1)$: la fase y la carga

$U(1)$ son las fases, $U(\theta) = e^{iq\theta}$. Tiene un solo generador, cuyo autovalor es la carga $q$, y es
abeliano: $e^{iq\theta_1}e^{iq\theta_2} = e^{iq(\theta_1 + \theta_2)}$, en cualquier orden. Sus representaciones
irreducibles son de dimensión $1$ y se etiquetan con $q$: **cada partícula tiene su carga**, y cualquier valor está
permitido. Es el mismo grupo que las rotaciones del plano, $SO(2)$: $e^{i\theta}$ y la matriz de giro
$\begin{pmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{pmatrix}$ se componen igual.

**En la física:**

* **Fase global**, $\Psi \to e^{iq\theta}\Psi$. La ecuación de Dirac no cambia. Por Noether se conserva la corriente,
  $\partial_\mu(\bar\Psi\gamma^\mu\Psi) = 0$, y con ella la carga ([ext-dirac](./ext-dirac.ipynb), sección 4). Con una fase
  para los quarks y otra para cada familia de leptones salen la conservación del número bariónico y de los
  leptónicos.
* **Fase local**, $\theta(x)$. Hace falta un campo que compense la derivada de la fase: el fotón,
  $D_\mu = \partial_\mu + iqA_\mu$ ([ext-dirac](./ext-dirac.ipynb), sección 5). Un generador, un bosón.
* **Las cargas se suman.** Un estado de dos partículas de cargas $q_1$ y $q_2$ se transforma con
  $e^{i(q_1 + q_2)\theta}$. Es la versión abeliana, trivial, de la suma de momentos angulares de $SU(2)$.

## 3. $SU(2)$: el espín y el isoespín

### 3.1. Generadores, álgebra y estados

Los generadores son las matrices de Pauli entre dos, $T_a = \sigma_a/2$:

$$
\sigma_1 = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \quad
\sigma_2 = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \quad
\sigma_3 = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}, \qquad
[T_a, T_b] = i\,\varepsilon_{abc}\,T_c .
$$

Es el álgebra del momento angular: $SU(2)$ **no es abeliano**. Su rango es $1$; se diagonalizan $T_3$ y el Casimir
${\bf T}^2 = T_1^2 + T_2^2 + T_3^2$, y los estados son $|t, m\rangle$:

$$
{\bf T}^2\,|t, m\rangle = t(t+1)\,|t, m\rangle, \qquad T_3\,|t, m\rangle = m\,|t, m\rangle, \qquad m = -t, \dots, t .
$$

Cada $t = 0, \frac{1}{2}, 1, \frac{3}{2}, \dots$ es una representación irreducible de dimensión $2t + 1$: $\mathbf{1}$,
$\mathbf{2}$, $\mathbf{3}$, $\mathbf{4}$... Los **operadores escalera**, $T_\pm = T_1 \pm iT_2$, recorren cada una:

$$
T_\pm\,|t, m\rangle = \sqrt{t(t+1) - m(m \pm 1)}\;|t, m \pm 1\rangle ,
\qquad \text{en } \mathbf{2}: \;
T_+ = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}, \;
T_- = \begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix} .
$$

### 3.2. Combinar dos dobletes: $\mathbf{2} \otimes \mathbf{2} = \mathbf{3} \oplus \mathbf{1}$

Dos espines $1/2$ tienen cuatro estados: $\uparrow\uparrow$, $\uparrow\downarrow$, $\downarrow\uparrow$, $\downarrow\downarrow$. Se
parte del de $m$ máximo, $\uparrow\uparrow$, y se baja con $T_- = T_-^{(1)} + T_-^{(2)}$:

$$
|1, 1\rangle = \uparrow\uparrow, \qquad
|1, 0\rangle = \frac{1}{\sqrt{2}}\left(\uparrow\downarrow + \downarrow\uparrow\right), \qquad
|1, -1\rangle = \downarrow\downarrow ;
\qquad
|0, 0\rangle = \frac{1}{\sqrt{2}}\left(\uparrow\downarrow - \downarrow\uparrow\right) .
$$

El singlete es el estado con $m = 0$ ortogonal a $|1, 0\rangle$. **El triplete es simétrico** al intercambiar las dos
partículas, y **el singlete, antisimétrico**.

En general, $t_1 \otimes t_2 = |t_1 - t_2| \oplus \dots \oplus (t_1 + t_2)$, con los coeficientes de Clebsch-Gordan del
[PDG](https://pdg.lbl.gov/2024/reviews/rpp2024-rev-clebsch-gordan-coefs.pdf). Con tres dobletes,
$\mathbf{2} \otimes \mathbf{2} \otimes \mathbf{2} = \mathbf{4} \oplus \mathbf{2} \oplus \mathbf{2}$: espín $3/2$, totalmente
simétrico, y dos de espín $1/2$ de simetría mixta. Son los espines de los bariones (sección 6). [MT9.2, MT9.3]

### 3.3. $SU(2)$ y las rotaciones: las dos vueltas

Las rotaciones del espacio, $SO(3)$, tienen la misma álgebra; en la representación $\mathbf{3}$, la de los vectores,
$(T_a)_{bc} = -i\,\varepsilon_{abc}$. Pero no son el mismo grupo. Un giro de $2\pi$ alrededor de $z$ en el doblete es

$$
e^{i 2\pi T_3} = \begin{pmatrix} e^{i\pi} & 0 \\ 0 & e^{-i\pi} \end{pmatrix} = -I :
$$

**el espinor vuelve cambiado de signo**, y hacen falta $4\pi$ para recuperarlo, como en
[perspectivas](./perspectivas.ipynb) y en [ext-dirac](./ext-dirac.ipynb) (sección 3.3). $U$ y $-U$ son el mismo
giro del espacio: $SU(2)$ recubre dos veces $SO(3)$. Las representaciones de $t$ entero lo son de los dos grupos;
las de $t$ semientero, los espinores, solo de $SU(2)$.

### 3.4. Tres usos de $SU(2)$

* **El espín.** El doblete $(\uparrow, \downarrow)$ y la suma de momentos angulares de la mecánica cuántica.
* **El isoespín de sabor** (W. Heisenberg, 1932). El protón y el neutrón pesan casi lo mismo, $938.3$ y $939.6$ MeV, y
  la interacción fuerte los trata igual: son un doblete de una simetría $SU(2)$ **global y aproximada**. En quarks,
  el doblete es $(u, d)$, con $m_u \simeq m_d$, despreciables frente a la escala de los hadrones, $\sim 1$ GeV. Los
  piones, $(\pi^+, \pi^0, \pi^-)$, son un triplete, $t = 1$. Los antiquarks forman el doblete $(-\bar d, \bar u)$, que se
  transforma como $(u, d)$ (el signo menos lo hace posible). La simetría la rompen $m_d - m_u$ y la carga eléctrica:
  por eso $m_n - m_p = 1.3$ MeV y $m_{\pi^\pm} - m_{\pi^0} = 4.6$ MeV. [MT9.2, MT9.5]
* **El isoespín débil.** Las partes a izquierdas forman dobletes, $(\nu_e, e)_L$ y $(u, d)_L$, y las partes a
  derechas son singletes ([leptones](./leptones.ipynb)). Es una simetría **local**, que da los $W$
  ([ext-sm](./ext-sm.ipynb)), y exacta salvo por la rotura del Higgs.

*Ojo con la hipercarga.* En sabor, $Q = T_3 + Y/2$ con $Y = B + S$ (sección 6); en la teoría electrodébil, la
hipercarga débil se define con $Q = T_3 + Y$ ([ext-sm](./ext-sm.ipynb)). Son dos números distintos con el mismo nombre.

## 4. $SU(3)$: ocho generadores

### 4.1. Las matrices de Gell-Mann

Los ocho generadores son $T_a = \lambda_a/2$, con las matrices de Gell-Mann:

$$
\lambda_1 = \begin{pmatrix} 0 & 1 & 0 \\ 1 & 0 & 0 \\ 0 & 0 & 0 \end{pmatrix}, \;
\lambda_2 = \begin{pmatrix} 0 & -i & 0 \\ i & 0 & 0 \\ 0 & 0 & 0 \end{pmatrix}, \;
\lambda_3 = \begin{pmatrix} 1 & 0 & 0 \\ 0 & -1 & 0 \\ 0 & 0 & 0 \end{pmatrix}, \;
\lambda_4 = \begin{pmatrix} 0 & 0 & 1 \\ 0 & 0 & 0 \\ 1 & 0 & 0 \end{pmatrix},
$$

$$
\lambda_5 = \begin{pmatrix} 0 & 0 & -i \\ 0 & 0 & 0 \\ i & 0 & 0 \end{pmatrix}, \;
\lambda_6 = \begin{pmatrix} 0 & 0 & 0 \\ 0 & 0 & 1 \\ 0 & 1 & 0 \end{pmatrix}, \;
\lambda_7 = \begin{pmatrix} 0 & 0 & 0 \\ 0 & 0 & -i \\ 0 & i & 0 \end{pmatrix}, \;
\lambda_8 = \frac{1}{\sqrt{3}}\begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & -2 \end{pmatrix} .
$$

* Son hermíticas y de traza nula. $\lambda_{1,2,3}$ son las de Pauli en el bloque de arriba: **$SU(2)$ está dentro
  de $SU(3)$**, y hay otras dos copias, en los bloques $(1, 3)$ y $(2, 3)$.
* Cumplen $[T_a, T_b] = i f_{abc}\,T_c$, con $f_{123} = 1$, $f_{458} = f_{678} = \sqrt{3}/2$ y otras, totalmente antisimétricas.
* **Rango 2:** solo $\lambda_3$ y $\lambda_8$ son diagonales. Se usan $T_3 = \lambda_3/2$ y la hipercarga
  $Y = \lambda_8/\sqrt{3}$; en la representación fundamental,

$$
T_3 = \mathrm{diag}\left(\tfrac{1}{2}, -\tfrac{1}{2}, 0\right), \qquad
Y = \mathrm{diag}\left(\tfrac{1}{3}, \tfrac{1}{3}, -\tfrac{2}{3}\right), \qquad
\sum_a T_a^2 = \tfrac{4}{3}\,I .
$$

### 4.2. Los diagramas de pesos

Cada estado tiene dos números cuánticos aditivos, $(T_3, Y)$, y se dibuja como un punto del plano. El triplete
$\mathbf{3}$ es un triángulo con el vértice abajo; el antitriplete $\bar{\mathbf{3}}$, con todos los números
cambiados de signo, uno con el vértice arriba:

Hay tres pares de **operadores escalera**, uno por cada $SU(2)$ de dentro:

$$
T_\pm = \tfrac{1}{2}(\lambda_1 \pm i\lambda_2), \qquad
V_\pm = \tfrac{1}{2}(\lambda_4 \pm i\lambda_5), \qquad
U_\pm = \tfrac{1}{2}(\lambda_6 \pm i\lambda_7),
$$

con $T_+\,d = u$, $V_+\,s = u$ y $U_+\,s = d$. Cada uno mueve el punto en una dirección fija: $T_+$, $(+1, 0)$; $V_+$,
$(+\frac{1}{2}, +1)$; $U_+$, $(-\frac{1}{2}, +1)$. Por eso los diagramas de $SU(3)$ tienen simetría hexagonal.

### 4.3. Combinar tripletes

Para combinar dos representaciones se coloca el diagrama de la segunda sobre cada punto de la primera: los números
cuánticos aditivos se suman. Con $\bar{\mathbf{3}}$ sobre $\mathbf{3}$ salen nueve puntos: un hexágono con tres en
el centro. Es un octete y un singlete:

$$
\mathbf{3} \otimes \bar{\mathbf{3}} = \mathbf{8} \oplus \mathbf{1}, \qquad
\mathbf{3} \otimes \mathbf{3} = \mathbf{6} \oplus \bar{\mathbf{3}}, \qquad
\mathbf{3} \otimes \mathbf{3} \otimes \mathbf{3} = \mathbf{10} \oplus \mathbf{8} \oplus \mathbf{8} \oplus \mathbf{1} .
$$

Las dimensiones cuadran: $9 = 8 + 1$, $9 = 6 + 3$, $27 = 10 + 8 + 8 + 1$. En el producto de tres, el $\mathbf{10}$ es
totalmente simétrico, el $\mathbf{1}$ totalmente antisimétrico y los dos $\mathbf{8}$ tienen simetría mixta, igual que
$\mathbf{4}$, $\mathbf{2}$ y $\mathbf{2}$ en $SU(2)$. [MT9.6]

## 5. El color

Cada quark es un triplete de color, $(r, g, b)$, y cada antiquark, un $\bar{\mathbf{3}}$. La simetría de color es
**exacta y local**: da la cromodinámica cuántica, QCD (sección 7). Y solo los **singletes de color** pueden estar
libres: es el confinamiento ([quarks y hadrones](./quarks_hadrons.ipynb), sección 2.1).

* **Mesones**, $q\bar q$: $\mathbf{3} \otimes \bar{\mathbf{3}} = \mathbf{8} \oplus \mathbf{1}$. El singlete es
  $\frac{1}{\sqrt{3}}(r\bar r + g\bar g + b\bar b)$.
* **Dos quarks**, $qq$: $\mathbf{3} \otimes \mathbf{3} = \mathbf{6} \oplus \bar{\mathbf{3}}$. **No hay singlete**: no
  existen hadrones de dos quarks.
* **Bariones**, $qqq$: un singlete, el totalmente antisimétrico,

$$
\frac{1}{\sqrt{6}}\,\varepsilon_{ijk}\,q_i q_j q_k = \frac{1}{\sqrt{6}}\left(rgb - rbg + gbr - grb + brg - bgr\right) .
$$

  Es el que resuelve el problema del $\Delta^{++}$ ([quarks y hadrones](./quarks_hadrons.ipynb), sección 1.2).

**Por qué son invariantes.** $\delta_{ij}\,q_i\bar q_j$ lo es porque $U$ es unitaria, $U^\dagger U = I$; y
$\varepsilon_{ijk}\,q_i q_j q_k$, porque $\det U = 1$: $\varepsilon_{ijk}\,U_{ia}U_{jb}U_{kc} = \det U\;\varepsilon_{abc}$. Las
dos condiciones de $SU(3)$, unitaria y especial, son justo las que hacen falta.

**Los gluones** llevan color y anticolor: están en la adjunta, el $\mathbf{8}$ de $\mathbf{3} \otimes \bar{\mathbf{3}}$
($r\bar g$, $r\bar b$, $\frac{1}{\sqrt{2}}(r\bar r - g\bar g)$...). El singlete no es un gluón. Si lo fuera, no tendría
color, se acoplaría a los hadrones, que tampoco lo tienen, y habría una fuerza fuerte de largo alcance entre ellos,
como la eléctrica entre cargas. No se ve: el grupo es $SU(3)$, con ocho, y no $U(3)$, con nueve. [MT10.2, MT10.3]

**Otras combinaciones blancas.** $qq\bar q\bar q$ y $qqqq\bar q$ también contienen singletes: son los tetraquarks y
pentaquarks ([quarks y hadrones](./quarks_hadrons.ipynb), sección 1.2). La regla: hace falta que el número de quarks
menos el de antiquarks sea múltiplo de $3$ (cuestión 6).

## 6. El sabor: el camino óctuple

### 6.1. Una simetría aproximada

La interacción fuerte no distingue el sabor. Si $u$, $d$ y $s$ pesaran lo mismo, los hadrones se agruparían en
representaciones de un $SU(3)$ **de sabor**, con los quarks en el triplete $(u, d, s)$ de la figura de la sección 4.2.
No pesan lo mismo, $m_s \simeq 93$ MeV frente a unos pocos MeV, pero la diferencia es pequeña frente a la escala de los
hadrones, $\sim 1$ GeV: la simetría es **aproximada**, y las masas de un mismo multiplete difieren en un $10$–$30\,\%$.
Con el $c$, el $b$ o el $t$, mucho más pesados, ya no sirve.

En sabor, $Y = B + S$ (número bariónico más extrañeza; el $s$ tiene $S = -1$), y la carga es
$Q = T_3 + Y/2$ (Gell-Mann y Nishijima): $u$, $(\frac{1}{2}, \frac{1}{3})$, $Q = \frac{2}{3}$; $d$, $(-\frac{1}{2}, \frac{1}{3})$,
$Q = -\frac{1}{3}$; $s$, $(0, -\frac{2}{3})$, $Q = -\frac{1}{3}$.

* **Mesones**, $\mathbf{3} \otimes \bar{\mathbf{3}} = \mathbf{8} \oplus \mathbf{1}$: el octete de los mesones
  pseudoescalares, $K$, $\pi$ y $\eta$, y un singlete, el $\eta'$ (los $\eta$ físicos son mezclas de los dos).
* **Bariones**, $\mathbf{3} \otimes \mathbf{3} \otimes \mathbf{3} = \mathbf{10} \oplus \mathbf{8} \oplus \mathbf{8} \oplus \mathbf{1}$,
  combinado con el espín, $\mathbf{2} \otimes \mathbf{2} \otimes \mathbf{2} = \mathbf{4} \oplus \mathbf{2} \oplus \mathbf{2}$.
  El color es antisimétrico (sección 5) y, con $L = 0$, el espacio es simétrico: Pauli pide que **sabor por espín sea
  simétrico**. Solo hay dos formas: el $\mathbf{10}$ con espín $\frac{3}{2}$, simétricos los dos, y el $\mathbf{8}$ con espín
  $\frac{1}{2}$, combinando las dos simetrías mixtas. **Un decuplete $\frac{3}{2}^+$ y un octete $\frac{1}{2}^+$**, y ningún
  singlete: exactamente lo que se observa. [MT9.6]

**¿Qué ves?**

* Cada fila horizontal es un multiplete de isoespín, con la misma extrañeza: $(p, n)$ es un doblete, los $\Sigma$ y
  los $\pi$, tripletes; los $\Delta$, un cuadruplete.
* La carga crece hacia la derecha, $Q = T_3 + Y/2$: las diagonales de arriba a la izquierda a abajo a la derecha tienen
  la misma carga.
* Cada fila hacia abajo tiene un $s$ más, y pesa más: en el decuplete, unos $150$ MeV por cada $s$.
* El vértice de abajo del decuplete, $sss$, es el $\Omega^-$.

<details>
<summary><b>[+] Mirador — La predicción del $\Omega^-$</b></summary>

En 1962 se conocían nueve bariones de espín $\frac{3}{2}$: los cuatro $\Delta$, los tres $\Sigma^*$ y los dos $\Xi^*$. Gell-Mann vio
que encajaban en un decuplete al que le faltaba el vértice: una partícula $sss$, de carga $-1$, extrañeza $-3$ y masa,
siguiendo el espaciado, $1533 + 148 \simeq 1680$ MeV. Con esa masa no podía desintegrarse por la interacción fuerte: no
hay un estado más ligero con extrañeza $-3$. Tendría que hacerlo por la débil, cambiando la extrañeza de una en una,
y vivir lo bastante como para dejar una traza.

En 1964, en la cámara de burbujas de Brookhaven, apareció el $\Omega^-$ con una masa de $1672$ MeV, y vive $0.8 \cdot 10^{-10}$ s.
Fue la confirmación del camino óctuple, y la primera pista de que detrás había tres quarks. [OM]

</details>

## 7. De global a local: un bosón por generador

Con una transformación local, $\psi \to e^{i\theta_a(x)T_a}\,\psi$, la derivada actúa sobre los $\theta_a(x)$ y la
ecuación deja de ser invariante, como en $U(1)$ ([ext-dirac](./ext-dirac.ipynb), sección 5). Para compensarlo hace
falta **un campo por generador** en la derivada covariante:

$$
D_\mu = \partial_\mu + i g\,T_a A^a_\mu ,
\qquad
A'^a_\mu = A^a_\mu - \frac{1}{g}\,\partial_\mu\theta_a - f_{abc}\,\theta_b A^c_\mu \quad (\theta \text{ pequeño}) .
$$

* **Cuántos bosones:** $U(1)$, uno, el fotón; $SU(2)$, tres, $W^1$, $W^2$ y $W^3$ (los $W^\pm$, y el $W^3$ se mezcla para dar
  el $Z$ y el fotón: [ext-sm](./ext-sm.ipynb)); $SU(3)$, ocho gluones.
* **Los bosones llevan la carga.** El último término de $A'$ dice que los campos se transforman entre sí: están en la
  adjunta. En $U(1)$, $f_{abc} = 0$ y el fotón es neutro; los gluones tienen color y los $W$, isoespín débil.
* **Interaccionan entre sí.** El tensor de campo es
  $F^a_{\mu\nu} = \partial_\mu A^a_\nu - \partial_\nu A^a_\mu - g\,f_{abc}\,A^b_\mu A^c_\nu$, y el término cinético,
  $-\frac{1}{4}F^a_{\mu\nu}F^{a\,\mu\nu}$, contiene productos de tres y de cuatro campos: **vértices de tres y de cuatro
  gluones**, y $WW\gamma$, $WWZ$ ([el Modelo Estándar](./sm.ipynb), sección 1.1). En QCD, los gluones que se
  acoplan a sí mismos dan la libertad asintótica y el confinamiento ([quarks y hadrones](./quarks_hadrons.ipynb),
  sección 2).
* **Un solo acoplo.** En $U(1)$ cada partícula puede tener su carga $q$; en un grupo no abeliano el álgebra fija la
  normalización y hay **un solo $g$**: todos los quarks tienen la misma carga de color, y todos los dobletes, el
  mismo acoplo con los $W$. Es la universalidad.
* **Sin masa.** Un término $m^2 A^a_\mu A^{a\,\mu}$ no es invariante: los bosones *gauge* no tienen masa, salvo que la
  simetría se rompa, que es lo que hace el Higgs con los $W$ y el $Z$ ([ext-sm](./ext-sm.ipynb), sección 3).
  [MT10.1, MT15]

## 8. Aparte: el grupo de Lorentz

**No es una simetría interna.** Las secciones anteriores transforman las etiquetas internas de un campo en un punto
fijo. El grupo de Lorentz transforma el propio espacio-tiempo: gira los ejes o cambia de sistema de referencia, y con
ellos se mezclan las componentes de los vectores y de los espinores. Pero es un grupo de Lie, con generadores y
representaciones, y las herramientas de esta extensión ordenan lo que [ext-dirac](./ext-dirac.ipynb) (sección 3) y
[ext-quiral](./ext-quiral.ipynb) (sección 6) calculan.

### 8.1. Rotaciones y *boosts*

Tiene seis generadores: tres de rotaciones, ${\bf J}$, y tres de *boosts*, ${\bf K}$, con el álgebra

$$
[J_i, J_j] = i\,\varepsilon_{ijk}\,J_k, \qquad [J_i, K_j] = i\,\varepsilon_{ijk}\,K_k, \qquad [K_i, K_j] = -i\,\varepsilon_{ijk}\,J_k .
$$

* **Las rotaciones solas** forman un subgrupo: el $SU(2)$ del espín (sección 3).
* **Los *boosts* no.** Dos *boosts* en direcciones distintas dan un *boost* y una rotación: es la última relación, y
  su efecto físico es la precesión de Thomas.
* **En el espinor de Dirac**, ${\bf J} = \frac{1}{2}{\bf \Sigma}$ y $K_k = \frac{i}{2}\gamma^0\gamma^k$: son los generadores de
  [ext-dirac](./ext-dirac.ipynb), sección 3.2 (cuestión 8).

### 8.2. Dos $SU(2)$ escondidas

Las combinaciones ${\bf A} = \frac{1}{2}({\bf J} + i{\bf K})$ y ${\bf B} = \frac{1}{2}({\bf J} - i{\bf K})$ cumplen cada una el álgebra de
$SU(2)$ y conmutan entre sí, $[A_i, B_j] = 0$. Cada representación irreducible se etiqueta con **dos espines**,
$(j_A, j_B)$, tiene dimensión $(2j_A + 1)(2j_B + 1)$ y, como ${\bf J} = {\bf A} + {\bf B}$, contiene los espines de
$j_A \otimes j_B$ (sección 3.2):

| $(j_A, j_B)$ | dimensión | espín | qué es | dónde aparece |
| :--: | :--: | :--: | :-- | :-- |
| $(0, 0)$ | $1$ | $0$ | escalar | el Higgs, $\bar\Psi\Psi$ |
| $(\frac{1}{2}, 0)$ | $2$ | $\frac{1}{2}$ | espinor de Weyl a izquierdas | [ext-quiral](./ext-quiral.ipynb), sección 4 |
| $(0, \frac{1}{2})$ | $2$ | $\frac{1}{2}$ | espinor de Weyl a derechas | [ext-quiral](./ext-quiral.ipynb), sección 4 |
| $(\frac{1}{2}, \frac{1}{2})$ | $4$ | $0 \oplus 1$ | cuadrivector | $x^\mu$, $p^\mu$, $A^\mu$, $\bar\Psi\gamma^\mu\Psi$ |
| $(\frac{1}{2}, 0) \oplus (0, \frac{1}{2})$ | $4$ | $\frac{1}{2} \oplus \frac{1}{2}$ | espinor de Dirac | [ext-dirac](./ext-dirac.ipynb) |

* **Cuatro componentes no son lo mismo.** Bajo rotaciones, el cuadrivector es un escalar, $t$, y un vector, ${\bf x}$;
  el espinor de Dirac, dos espines $\frac{1}{2}$ (cuestión 9).
* **El espinor de Dirac es reducible.** En la base quiral, $S(\Lambda)$ son dos bloques $2 \times 2$ que no se mezclan, uno
  por quiralidad. La **paridad** intercambia $(\frac{1}{2}, 0)$ y $(0, \frac{1}{2})$ ([ext-quiral](./ext-quiral.ipynb),
  sección 2.3): si se incluye, ya no se separan, y el de Dirac es el espinor más pequeño en una teoría que conserve la
  paridad, como QED. La débil, que no la conserva, usa solo la mitad.
* **Los bilineales.** El conjugado de un espinor a izquierdas se transforma como uno a derechas:
  $\psi_L^\dagger$ es un $(0, \frac{1}{2})$. Es lo que hace el $\gamma^0$ del adjunto: en la base quiral cruza los bloques,
  $\bar\Psi = (\psi_R^\dagger, \psi_L^\dagger)$, y $\bar\Psi\Psi = \psi_L^\dagger\psi_R + \psi_R^\dagger\psi_L = \bar\Psi_L\Psi_R + \bar\Psi_R\Psi_L$
  ([ext-quiral](./ext-quiral.ipynb), secciones 3 y 9). Así, $\psi_L^\dagger\psi_R$, $(0, \frac{1}{2}) \otimes (0, \frac{1}{2}) = (0, 0) \oplus (0, 1)$,
  contiene un escalar: la masa, que une L con R. Y $\bar\Psi_L\gamma^\mu\Psi_L$, que en componentes de Weyl es $\psi_L^\dagger$
  por $\psi_L$ con una matriz $2 \times 2$ para cada $\mu$, es $(0, \frac{1}{2}) \otimes (\frac{1}{2}, 0) = (\frac{1}{2}, \frac{1}{2})$: un
  cuadrivector, la corriente, que no mezcla las quiralidades.

### 8.3. No es compacto: por qué hace falta $\bar\Psi$

Los ángulos de una rotación tienen período, $2\pi$ (o $4\pi$ en los espinores); la rapidez de un *boost* va de $-\infty$
a $+\infty$. El grupo de Lorentz **no es compacto**, y eso tiene una consecuencia: sus representaciones de dimensión
finita **no son unitarias**.

* Los generadores de los *boosts* no son hermíticos: $K_k = \frac{i}{2}\gamma^0\gamma^k$ es antihermítico.
* $S(\Lambda)$ no conserva la norma: $\Psi^\dagger\Psi$ **no es invariante**. Es la componente temporal de un cuadrivector,
  y crece como $E$ (la normalización $u^\dagger u = 2E$).
* Por eso los invariantes se construyen con $\bar\Psi = \Psi^\dagger\gamma^0$, que se transforma con $S^{-1}$
  ([ext-dirac](./ext-dirac.ipynb), sección 4.1).

En los grupos internos, compactos, no pasa: $U^\dagger U = I$, y $\psi^\dagger\psi$ ya es invariante.

<details>
<summary><b>[+] Mirador — Una partícula es una representación</b></summary>

Las representaciones unitarias del grupo de Lorentz existen, pero son de dimensión infinita: actúan sobre los estados
con todos los momentos posibles. E. Wigner (1939) las clasificó para el grupo de Poincaré, Lorentz más traslaciones, y
encontró que cada una se etiqueta con una **masa** y un **espín** (o una helicidad, si la masa es nula). Desde
entonces, la definición más limpia de partícula es esa: una representación irreducible del grupo de Poincaré. Las
simetrías internas le añaden sus cargas. [WI]

¿Se pueden mezclar las dos clases de simetrías, por ejemplo con un grupo que transforme a la vez el espín y el color?
S. Coleman y J. Mandula (1967) demostraron que no: en una teoría razonable, el grupo de simetría es el producto de
Poincaré por las internas, sin mezcla. Es la razón de fondo para tratarlas por separado. La única salida es la
**supersimetría**, que relaciona fermiones y bosones. [CM]

</details>

<summary><b>Soluciones</b></summary>

1. $e^{i2\pi T_3}$ es diagonal, con $e^{i2\pi m}$: en el doblete, $m = \pm\frac{1}{2}$ y $e^{\pm i\pi} = -1$; en el triplete,
   $m = 1, 0, -1$ y $e^{i2\pi m} = 1$. Los espinores cambian de signo, los vectores no.
2. $n^2$: sin la condición del determinante. El de más es la identidad, una fase común a los tres colores: un $U(1)$.
   En color sería un noveno gluón, el singlete, sin color y de largo alcance entre hadrones (sección 5). No existe.
3. Tres en el centro, $u\bar u$, $d\bar d$ y $s\bar s$: dos combinaciones son del octete ($\pi^0$ y $\eta_8$) y una es el
   singlete. En los vértices del hexágono, seis estados, uno en cada uno.
4. $p$ ($uud$): $Y = 1$, $T_3 = \frac{1}{2}$, $Q = 1$. $\Sigma^+$ ($uus$): $Y = 0$, $T_3 = 1$, $Q = 1$. $\Xi^-$ ($dss$): $Y = -1$,
   $T_3 = -\frac{1}{2}$, $Q = -1$. $\Omega^-$ ($sss$): $Y = -2$, $T_3 = 0$, $Q = -1$. $K^+$ ($u\bar s$): $Y = 1$, $T_3 = \frac{1}{2}$, $Q = 1$.
5. Los saltos son $153$ y $148$ MeV: $m_\Omega \simeq 1533 + 148 \simeq 1681$ MeV (medida, $1672$). $\Xi^0 + K^-$ pesan
   $1315 + 494 = 1809$ MeV, más que el $\Omega^-$: ningún estado con $S = -3$ es más ligero, y solo puede desintegrarse
   por la débil, a $\Lambda + K^-$ o $\Xi + \pi$, cambiando la extrañeza en una unidad.
6. Las sumas posibles son $\frac{2}{3}$, $-\frac{1}{3}$ y $-\frac{4}{3}$: nunca $0$. Un singlete tiene $T_3 = Y = 0$, así que en
   $qq$ no hay ninguno. En $qqq$, $\frac{1}{3} + \frac{1}{3} - \frac{2}{3} = 0$ sí es posible ($rgb$), y en $q\bar q$ también
   ($r\bar r$). En general, el número de quarks menos el de antiquarks tiene que ser múltiplo de $3$.
7. $\varepsilon_{ijk}\,U_{i1}U_{j2}U_{k3}$ es la fórmula del determinante (suma con signo de productos de una entrada por
   columna); cambiar el orden de las columnas solo cambia el signo, como $\varepsilon_{abc}$. Así,
   $\varepsilon_{ijk}\,q'_i q'_j q'_k = \varepsilon_{ijk}\,U_{ia}U_{jb}U_{kc}\,q_a q_b q_c = \det U\,\varepsilon_{abc}\,q_a q_b q_c$: es invariante
   solo si $\det U = 1$. En $U(3)$ cambiaría de fase.
8. $[K_1, K_2] = -\frac{1}{4}[\gamma^0\gamma^1, \gamma^0\gamma^2]$. Como $\gamma^0$ anticonmuta con $\gamma^k$ y $(\gamma^0)^2 = I$,
   $\gamma^0\gamma^1\gamma^0\gamma^2 = -\gamma^1\gamma^2$ y $\gamma^0\gamma^2\gamma^0\gamma^1 = -\gamma^2\gamma^1 = \gamma^1\gamma^2$: el conmutador es
   $-2\gamma^1\gamma^2$, y $[K_1, K_2] = \frac{1}{2}\gamma^1\gamma^2 = -\frac{i}{2}\Sigma_3 = -iJ_3$. $(\gamma^0\gamma^3)^\dagger = \gamma^0\gamma^3$, así que
   $K_3^\dagger = -\frac{i}{2}\gamma^0\gamma^3 = -K_3$: **antihermítico**, y $e^{-i\eta K_3}$ no es unitaria.
9. $\frac{1}{2} \otimes \frac{1}{2} = 0 \oplus 1$: en $x^\mu$, $t$ es el espín $0$ (no cambia al girar) y $(x, y, z)$ el espín $1$, $1 + 3 = 4$.
   El espinor de Dirac, $\frac{1}{2} \oplus \frac{1}{2}$: dos dobletes, $2 + 2 = 4$, los dos bloques de ${\bf \Sigma} = \mathrm{diag}(\vec\sigma, \vec\sigma)$.
