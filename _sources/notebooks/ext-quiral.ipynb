## Introducción

Esta extensión es el **desarrollo matemático de la quiralidad**, que el tema de [leptones](./leptones.ipynb) presenta
en clase: la interacción débil solo ve la mano izquierda de los fermiones. Parte de lo que se construye en
[ext-dirac](./ext-dirac.ipynb): la ecuación de Dirac, los espinores, sus transformaciones de Lorentz y las formas
bilineales. Tiene nueve secciones; cada una responde a una pregunta:

1. **La quiralidad.** *¿Qué es?* Los autovalores de $\gamma^5$.
2. **Los proyectores y la paridad.** *¿Cómo se separa un espinor en sus dos quiralidades?* Con $P_L$ y $P_R$. La
   paridad las intercambia.
3. **La base quiral.** *¿En qué base se ven las dos partes por separado?* La representación de Weyl.
4. **Fermiones sin masa: los espinores de Weyl.** *¿Qué pasa si $m = 0$?* Las dos partes se desacoplan.
5. **Fermiones con masa.** *¿Qué hace la masa?* Acopla las dos partes.
6. **Rotaciones y *boosts* en la base quiral.** *¿Cómo se transforma cada parte?* Por separado: la quiralidad es
   invariante Lorentz.
7. **Helicidad y quiralidad.** *¿Cómo se relacionan?* Coinciden sin masa; con masa, un estado de helicidad es una
   mezcla de las dos quiralidades.
8. **Helicidad y quiralidad bajo transformaciones.** *¿Qué cambia con el observador y qué al propagarse?*
9. **Las formas bilineales y la quiralidad.** *¿Qué partes ve cada forma?* El escalar (la masa) une L con R; la
   corriente vectorial no las mezcla, y la $V - A$ solo ve la izquierda, en cualquier sistema de referencia.

## 1. La quiralidad

### 1.1. Los autovalores de $\gamma^5$

$\gamma^5$ es hermítica y su cuadrado es la identidad. Sus autovalores son reales y su cuadrado vale $1$: solo pueden
ser $\pm 1$. Como su traza es cero, hay **dos de cada**.

$$
\boxed{
\gamma^5\,\Psi_L = -\Psi_L, \qquad \gamma^5\,\Psi_R = +\Psi_R
}
$$

La **quiralidad** de un espinor es su autovalor de $\gamma^5$: $-1$, **a izquierdas** (L); $+1$, **a derechas** (R). El
nombre viene del griego *kheir*, mano: la paridad convierte una en otra, como un espejo convierte la mano
izquierda en la derecha (sección 2.3).

### 1.2. En Pauli-Dirac

$\gamma^5 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix}$ intercambia los dos bloques del espinor, así que sus autovectores son

$$
\Psi_L = \begin{pmatrix} \chi \\ -\chi \end{pmatrix}, \qquad \Psi_R = \begin{pmatrix} \chi \\ \chi \end{pmatrix},
$$

con cualquier $\chi$ de dos componentes. Un espinor cualquiera **no es autoestado** de $\gamma^5$: es una mezcla de
las dos quiralidades. Por ejemplo, el espinor en reposo, $u(m) = \sqrt{2m}\,(\chi, 0)$, es mitad y mitad:

$$
\sqrt{2m}\begin{pmatrix} \chi \\ 0 \end{pmatrix} = \sqrt{\frac{m}{2}}\begin{pmatrix} \chi \\ \chi \end{pmatrix} + \sqrt{\frac{m}{2}}\begin{pmatrix} \chi \\ -\chi \end{pmatrix} .
$$

## 2. Los proyectores y la paridad

### 2.1. Los proyectores de quiralidad

$$
\boxed{
P_L \equiv \frac{1}{2}(I - \gamma^5), \qquad P_R \equiv \frac{1}{2}(I + \gamma^5)
}
$$

Con $(\gamma^5)^2 = I$:

* son **proyectores complementarios**: $P_L + P_R = I$, $P_{L,R}^2 = P_{L,R}$, $P_L P_R = 0$;
* **dan los autoestados**: $\gamma^5 P_L = \frac{1}{2}(\gamma^5 - I) = -P_L$ y $\gamma^5 P_R = +P_R$. Luego $P_L\Psi$ es a izquierdas y $P_R\Psi$ a
  derechas, sea cual sea $\Psi$.

Cualquier espinor se separa, de una sola manera, en sus dos partes quirales:

$$
\Psi = \Psi_L + \Psi_R, \qquad \Psi_{L,R} = P_{L,R}\,\Psi .
$$

En Pauli-Dirac,

$$
P_L = \frac{1}{2}\begin{pmatrix} I & -I \\ -I & I \end{pmatrix}, \qquad
P_R = \frac{1}{2}\begin{pmatrix} I & I \\ I & I \end{pmatrix} .
$$

### 2.2. La paridad en los espinores

La paridad es la transformación de Lorentz impropia $\Lambda = \mathrm{diag}(1, -1, -1, -1)$. La condición de
[ext-dirac](./ext-dirac.ipynb) (sección 3.1), $S^{-1}\gamma^\mu S = \Lambda^\mu{}_\nu\gamma^\nu$, pide una $S$ que deje $\gamma^0$ igual y cambie el signo
de las $\gamma^k$. $S = \gamma^0$ lo cumple: $\gamma^0\gamma^0\gamma^0 = \gamma^0$ y $\gamma^0\gamma^k\gamma^0 = -\gamma^k$. Así,

$$
\boxed{
\Psi(t, {\bf x}) \;\to\; \gamma^0\,\Psi(t, -{\bf x}), \qquad \bar\Psi \;\to\; \bar\Psi\,\gamma^0
}
$$

**Las formas bilineales bajo paridad.** Con $\gamma^0\gamma^5\gamma^0 = -\gamma^5$ (porque $\gamma^5$ anticonmuta con $\gamma^0$):

| forma | $\to$ | bajo paridad |
|:--|:--|:--|
| $\bar\Psi\Psi$ | $\bar\Psi\gamma^0\gamma^0\Psi = \bar\Psi\Psi$ | escalar |
| $\bar\Psi\gamma^5\Psi$ | $\bar\Psi\gamma^0\gamma^5\gamma^0\Psi = -\bar\Psi\gamma^5\Psi$ | pseudoescalar |
| $\bar\Psi\gamma^\mu\Psi$ | $(j^0, {\bf j}) \to (j^0, -{\bf j})$ | vector |
| $\bar\Psi\gamma^\mu\gamma^5\Psi$ | $(j^0_A, {\bf j}_A) \to (-j^0_A, +{\bf j}_A)$ | axial |

(para el axial, $\gamma^0\gamma^\mu\gamma^5\gamma^0 = (\gamma^0\gamma^\mu\gamma^0)(\gamma^0\gamma^5\gamma^0) = -\gamma^0\gamma^\mu\gamma^0\,\gamma^5$). Son las columnas de «bajo
paridad» de [perspectivas](./perspectivas.ipynb) y de [leptones](./leptones.ipynb).

### 2.3. La paridad intercambia las quiralidades

Como $\gamma^0$ anticonmuta con $\gamma^5$, $\gamma^0 P_L = P_R\,\gamma^0$, y

$$
\gamma^0\,\Psi_L = \gamma^0 P_L\Psi = P_R\,(\gamma^0\Psi) :
$$

la parte izquierda de $\Psi$ se convierte en la parte derecha del espinor transformado, y al revés:
$\Psi_L \leftrightarrow \Psi_R$.

* **La paridad no conserva la quiralidad: la invierte.** Del transformado de $\Psi_L$ no queda nada a izquierdas,
  $P_L(\gamma^0\Psi_L) = \gamma^0 P_R P_L\Psi = 0$.
* **Solo cuenta $\gamma^0$.** El cambio ${\bf x} \to -{\bf x}$ actúa sobre el argumento, no sobre las componentes del
  espinor; y la fase global que puede acompañar a la paridad, $\Psi \to \eta_P\,\gamma^0\Psi(t, -{\bf x})$ con $|\eta_P| = 1$,
  tampoco cambia el autovalor de $\gamma^5$.
* **Una teoría que trate distinto las dos quiralidades viola la paridad.** Es lo que hace la interacción
  débil: la imagen especular de un $\nu_L$ sería un $\nu_R$, que no está en el Modelo Estándar.

### 2.4. La quiralidad de las antipartículas

Para los espinores $v$, los operadores dan el valor físico con el signo cambiado ([ext-dirac](./ext-dirac.ipynb),
sección 2.6), y la quiralidad no es una excepción: el antifermión a derechas es

$$
v_R = P_L\,v = \frac{1}{2}(I - \gamma^5)\,v .
$$

Por eso el mismo $P_L$ del vértice $V - A$ selecciona los fermiones a izquierdas **y** los antifermiones a
derechas, y la corriente del neutrino solo contiene $\nu_L$ y $\bar\nu_R$. [MT6.4]

## 3. La base quiral

### 3.1. La representación de Weyl

En Pauli-Dirac $\gamma^5$ no es diagonal. Cambiando de base con la matriz unitaria

$$
U = \frac{1}{\sqrt{2}}\begin{pmatrix} I & -I \\ I & I \end{pmatrix}, \qquad
\Psi_{W} = U\,\Psi_{PD}, \qquad \gamma^\mu_{W} = U\,\gamma^\mu_{PD}\,U^\dagger ,
$$

se obtiene la representación **de Weyl**, o **quiral**:

$$
\boxed{
\gamma^0 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix}, \qquad
\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ -\sigma_k & 0 \end{pmatrix}, \qquad
\gamma^5 = \begin{pmatrix} -I & 0 \\ 0 & I \end{pmatrix}
}
$$

y $P_L = \mathrm{diag}(I, 0)$, $P_R = \mathrm{diag}(0, I)$. Cumple la misma álgebra, $\{\gamma^\mu, \gamma^\nu\} = 2g^{\mu\nu}I$: la física no
cambia ([ext-dirac](./ext-dirac.ipynb), sección 1.5).

### 3.2. El espinor, partido en dos

En la base quiral, las dos componentes de arriba son la parte L y las dos de abajo la parte R:

$$
\Psi = \begin{pmatrix} \psi_L \\ \psi_R \end{pmatrix}, \qquad
\psi_L = \frac{1}{\sqrt{2}}(\psi_a - \psi_b), \quad \psi_R = \frac{1}{\sqrt{2}}(\psi_a + \psi_b),
$$

con $\psi_a$, $\psi_b$ los bloques de arriba y de abajo en Pauli-Dirac. Cada parte es un objeto de **dos componentes**.

* La **paridad**, $\gamma^0$, intercambia los dos bloques: $(\psi_L, \psi_R) \to (\psi_R, \psi_L)$.
* El **espín**, $\Sigma_k = \mathrm{diag}(\sigma_k, \sigma_k)$, es el mismo en las dos bases, y actúa igual sobre las dos partes.
* El **adjunto**, $\bar\Psi = \Psi^\dagger\gamma^0 = (\psi_R^\dagger, \psi_L^\dagger)$, cruza los bloques.

## 4. Fermiones sin masa: los espinores de Weyl

### 4.1. La ecuación de Dirac en la base quiral

Con $\not{p} = \gamma^0 E - \gamma^k p^k = \begin{pmatrix} 0 & E - \vec\sigma\cdot{\bf p} \\ E + \vec\sigma\cdot{\bf p} & 0 \end{pmatrix}$, la ecuación $(\not{p} - m)\,u = 0$ queda

$$
\begin{pmatrix} -m & E - \vec\sigma\cdot{\bf p} \\ E + \vec\sigma\cdot{\bf p} & -m \end{pmatrix}
\begin{pmatrix} u_L \\ u_R \end{pmatrix} = 0
\qquad \Longleftrightarrow \qquad
\begin{cases}
(E + \vec\sigma\cdot{\bf p})\,u_L = m\,u_R \\
(E - \vec\sigma\cdot{\bf p})\,u_R = m\,u_L
\end{cases}
$$

**La masa es lo único que acopla** $u_L$ con $u_R$.

### 4.2. Sin masa, las dos partes se separan

$$
(E + \vec\sigma\cdot{\bf p})\,u_L = 0, \qquad (E - \vec\sigma\cdot{\bf p})\,u_R = 0 :
$$

son las **ecuaciones de Weyl** (1929). Con $E = \mathrm{p}$ y ${\bf p} = \mathrm{p}\,\hat{\bf v}$:

$$
(\hat{\bf v}\cdot\vec\sigma)\,u_L = -u_L, \qquad (\hat{\bf v}\cdot\vec\sigma)\,u_R = +u_R .
$$

**Sin masa, la parte L tiene helicidad negativa y la R positiva.** Y son **dos partículas independientes**, de dos
componentes cada una: se puede tener una sin la otra. Es el neutrino del Modelo Estándar, que solo tiene $u_L$
(y su antipartícula, $\bar\nu_R$).

***[+] Mirador.*** *Weyl propuso su ecuación en 1929, y Pauli la descartó porque viola la paridad. Tras el
experimento de Wu (1957) resultó ser la del neutrino.*

## 5. Fermiones con masa

### 5.1. Las soluciones

Con masa, las dos ecuaciones de la sección 4.1 van juntas: dada $u_L$, $u_R = (E + \vec\sigma\cdot{\bf p})\,u_L/m$. Con ${\bf p}$ en $z$ y
$\sigma_3\chi_\pm = \pm\chi_\pm$, normalizados a $u^\dagger u = 2E$:

$$
\boxed{
u_\pm = \begin{pmatrix} \sqrt{E \mp \mathrm{p}}\;\chi_\pm \\ \sqrt{E \pm \mathrm{p}}\;\chi_\pm \end{pmatrix}
}
$$

(con $\sqrt{E - \mathrm{p}}\sqrt{E + \mathrm{p}} = m$). **Las dos partes están siempre presentes.** En reposo, $\mathrm{p} = 0$, pesan lo mismo:
$u = \sqrt{m}\,(\chi, \chi)$. Al aumentar la energía, en $u_+$ crece la parte R y en $u_-$ la L.

### 5.2. El término de masa

Con $\bar\Psi = (\psi_R^\dagger, \psi_L^\dagger)$,

$$
\bar\Psi\Psi = \psi_R^\dagger\psi_L + \psi_L^\dagger\psi_R = \bar\Psi_R\Psi_L + \bar\Psi_L\Psi_R :
$$

el término de masa del lagrangiano, $m\bar\Psi\Psi$, **acopla la izquierda con la derecha**. Un fermión con masa (de
Dirac) necesita las dos partes.

### 5.3. La quiralidad no se conserva al propagarse

El hamiltoniano de Dirac, $H = \vec\alpha\cdot{\bf p} + \beta m$, con $\alpha^k = \gamma^0\gamma^k$ y $\beta = \gamma^0$, en la base quiral es

$$
H = \begin{pmatrix} -\vec\sigma\cdot{\bf p} & m \\ m & \vec\sigma\cdot{\bf p} \end{pmatrix} :
$$

el término cinético es diagonal (no mezcla) y la masa está **fuera de la diagonal**. En general, $\gamma^5$ conmuta con
$\alpha^k$ (anticonmuta con cada factor) y anticonmuta con $\beta$:

$$
[\gamma^5, H] = 2m\,\gamma^5\gamma^0 \neq 0 .
$$

**Con masa, la quiralidad no se conserva**: un espinor que empieza siendo L desarrolla parte R al evolucionar.
Sin masa, sí se conserva.

## 6. Rotaciones y *boosts* en la base quiral

### 6.1. Las dos partes se transforman por separado

En [ext-dirac](./ext-dirac.ipynb) (secciones 3.2–3.6), los generadores son el espín, $\frac{1}{2}\Sigma_k$, para las rotaciones, y
$\frac{1}{2}\gamma^0\gamma^k$ para los *boosts*. En la base quiral, $\Sigma_k = \mathrm{diag}(\sigma_k, \sigma_k)$ y $\gamma^0\gamma^k = \mathrm{diag}(-\sigma_k, \sigma_k)$: **los dos son
diagonales por bloques**, y por tanto también $S(\Lambda)$:

$$
\boxed{
S = \begin{pmatrix} S_L & 0 \\ 0 & S_R \end{pmatrix}, \qquad
\begin{aligned}
&\text{rotación } \theta\,\hat{\bf n}: && S_L = S_R = e^{-i\,\theta\,\hat{\bf n}\cdot\vec\sigma/2} \\
&\textit{boost } \eta\,\hat{\bf n}: && S_L = e^{-\eta\,\hat{\bf n}\cdot\vec\sigma/2}, \quad S_R = e^{+\eta\,\hat{\bf n}\cdot\vec\sigma/2}
\end{aligned}
}
$$

* **Ninguna transformación de Lorentz mezcla L con R**: $\psi_L \to S_L\,\psi_L$ y $\psi_R \to S_R\,\psi_R$. Un espinor puramente L es
  puramente L para cualquier observador. **La quiralidad es invariante Lorentz** (equivalente a $[\gamma^5, S] = 0$,
  ext-dirac, sección 3.2).
* Las dos partes son **dos representaciones distintas** del grupo de Lorentz: se transforman igual bajo rotaciones,
  pero con el *boost* opuesto. Solo la paridad las une (sección 2.3).
* **Rotaciones:** la misma matriz unitaria en las dos partes: la proporción entre L y R no cambia.
* ***Boosts*:** no son unitarios y tienen exponentes opuestos: **reescalan** las dos partes de forma distinta. La
  proporción cambia, pero lo L sigue siendo L.

### 6.2. Ejemplo: el espinor en movimiento, sin resolver la ecuación

El espinor en reposo con espín arriba es $u(m) = \sqrt{m}\,(\chi_+, \chi_+)$. Un *boost* de rapidez $\eta$ en $z$ actúa sobre
$\chi_+$, con $\sigma_3 = +1$:

$$
S\,u(m) = \sqrt{m}\begin{pmatrix} e^{-\eta/2}\,\chi_+ \\ e^{+\eta/2}\,\chi_+ \end{pmatrix}
= \begin{pmatrix} \sqrt{E - \mathrm{p}}\;\chi_+ \\ \sqrt{E + \mathrm{p}}\;\chi_+ \end{pmatrix} = u_+ ,
$$

porque $m\,e^{\mp\eta} = m(\cosh\eta \mp \sinh\eta) = E \mp \mathrm{p}$. Es la solución de la sección 5.1. El *boost* no cambia la dirección
de ninguna de las dos partes: solo multiplica la L por $e^{-\eta/2}$ y la R por $e^{+\eta/2}$.

## 7. Helicidad y quiralidad

### 7.1. Un estado de helicidad, en sus dos quiralidades

De la sección 5.1, con $|u|^2 = 2E$, la fracción de cada quiralidad en un estado de helicidad es

| helicidad $\downarrow$ \ quiralidad $\rightarrow$ | R | L |
| :--: | :--: | :--: |
| $+$ | $\frac{E + \mathrm{p}}{2E} = \frac{1}{2}(1 + \beta)$ | $\frac{E - \mathrm{p}}{2E} = \frac{1}{2}(1 - \beta)$ |
| $-$ | $\frac{1}{2}(1 - \beta)$ | $\frac{1}{2}(1 + \beta)$ |

* **En reposo**, $\beta = 0$: mitad y mitad.
* **Sin masa**, $\beta = 1$: $u_+$ es puramente R y $u_-$ puramente L. **Helicidad y quiralidad coinciden.**
* **Con $E \gg m$**: la fracción «equivocada», $|P_L u_+|^2/|u_+|^2 = \frac{1}{2}(1 - \beta) \simeq m^2/4E^2$, es pequeña. La
  componente misma (su raíz, la amplitud) va como $m/2E$. La amplitud es la que entra en $M_{fi}$; la fracción,
  en las tasas.

### 7.2. En Pauli-Dirac

Es lo mismo escrito en la otra base. Con $u_+ = N(1, 0, \kappa, 0)$ y $u_- = N(0, 1, 0, -\kappa)$:

$$
P_R\,u_+ = \tfrac{1}{2}(1+\kappa)\,N\,(1, 0, 1, 0), \qquad P_L\,u_+ = \tfrac{1}{2}(1-\kappa)\,N\,(1, 0, -1, 0),
$$

$$
P_R\,u_- = \tfrac{1}{2}(1-\kappa)\,N\,(0, 1, 0, 1), \qquad P_L\,u_- = \tfrac{1}{2}(1+\kappa)\,N\,(0, 1, 0, -1) .
$$

Las **amplitudes** $\frac{1}{2}(1 \pm \kappa)$ son las que usa el tema de leptones; sus cuadrados, por la norma de los vectores,
$2N^2\cdot\frac{1}{4}(1 \pm \kappa)^2 = E \pm \mathrm{p}$, dan las fracciones de la tabla de arriba. En la desintegración a dos cuerpos
$\pi \to l\,\nu$, con $E_l$ y $\mathrm{p}_l$ del CM ([ext-cinematica](./ext-cinematica.ipynb), sección 4.3), $\kappa = (m_\pi - m_l)/(m_\pi + m_l)$ y
$\frac{1}{2}(1 - \kappa) = m_l/(m_\pi + m_l)$: la supresión por helicidad de [leptones](./leptones.ipynb).

## 8. Helicidad y quiralidad bajo transformaciones

### 8.1. Cada una tiene una virtud y un defecto

| | helicidad, $\vec\Sigma\cdot\hat{\bf p}$ | quiralidad, $\gamma^5$ |
| :-- | :-- | :-- |
| ¿se conserva al propagarse? (conmuta con $H$) | **sí** | **no**, si hay masa (sección 5.3) |
| ¿es la misma para todo observador? (conmuta con $S(\Lambda)$) | **no**, si hay masa | **sí** (sección 6.1) |
| ¿la tiene definida una partícula libre con masa? | **sí** | **no**: lleva las dos partes |
| sin masa | se conserva, es invariante y coincide con la quiralidad | ídem |

**La helicidad depende del observador** si hay masa: un observador que adelanta a la partícula la ve moverse hacia
atrás, con el espín igual, y su helicidad cambia de signo. Sin masa no se la puede adelantar.

**La quiralidad no depende del observador**, pero una partícula con masa no la tiene definida. Lo que es igual
para todos los observadores es la **separación** del espinor en sus dos partes: la parte L para uno es la parte L
para cualquier otro. Lo que cambia es **cuánto pesa** cada parte.

### 8.2. Ejemplo: adelantar a la partícula

Tomemos $u_+$, con rapidez $\vartheta$ en $z$. Sus dos partes están en la proporción (sección 6.2)

$$
\frac{\text{amplitud R}}{\text{amplitud L}} = \sqrt{\frac{E + \mathrm{p}}{E - \mathrm{p}}} = e^{\vartheta} .
$$

Un *boost* de rapidez $\eta$ en $z$ multiplica la R por $e^{\eta/2}$ y la L por $e^{-\eta/2}$: la proporción pasa a $e^{\vartheta + \eta}$.

* **$\eta = -\vartheta$**, el sistema en reposo: mitad y mitad.
* **$\eta = -2\vartheta$**, el observador que la adelanta: la proporción es $e^{-\vartheta}$, domina la parte L. La partícula
  se mueve hacia $-z$ con el espín hacia $+z$: helicidad negativa. Es el mismo estado, visto por otro observador.

**Ninguna parte ha cambiado de quiralidad**; ha cambiado su peso. Por eso la interacción débil se escribe con la
quiralidad: el $W$ se acopla a la parte L, y todos los observadores están de acuerdo en cuál es.

## 9. Las formas bilineales y la quiralidad

### 9.1. Los adjuntos de las partes quirales

Para separar las formas bilineales hacen falta los adjuntos de $\Psi_L$ y $\Psi_R$. Como $\gamma^5$ es hermítica y
anticonmuta con $\gamma^0$, $P_L\gamma^0 = \gamma^0 P_R$:

$$
\bar\Psi_L \equiv (P_L\Psi)^\dagger\gamma^0 = \Psi^\dagger P_L\gamma^0 = \Psi^\dagger\gamma^0 P_R = \bar\Psi\,P_R, \qquad \bar\Psi_R = \bar\Psi\,P_L .
$$

**El adjunto cambia el proyector.** Y como $\gamma^5$ anticonmuta con $\gamma^\mu$, al pasar un proyector a través de una $\gamma^\mu$
también cambia: $P_R\gamma^\mu = \gamma^\mu P_L$. Con estas dos reglas, y $P_L P_R = 0$, sale todo lo demás.

### 9.2. El escalar y el pseudoescalar mezclan las quiralidades

Escribiendo $\Psi = \Psi_L + \Psi_R$ y $\bar\Psi = \bar\Psi_L + \bar\Psi_R$, los términos con la misma quiralidad se anulan:

$$
\bar\Psi_L\Psi_L = \bar\Psi P_R\,P_L\Psi = 0, \qquad \bar\Psi_R\Psi_R = \bar\Psi P_L\,P_R\Psi = 0 ,
$$

y solo sobreviven los cruzados:

$$
\boxed{
\bar\Psi\Psi = \bar\Psi_L\Psi_R + \bar\Psi_R\Psi_L
}
\qquad
\bar\Psi\gamma^5\Psi = \bar\Psi_L\Psi_R - \bar\Psi_R\Psi_L .
$$

**El escalar une la izquierda con la derecha.** Es el término de masa, $m\bar\Psi\Psi$ (sección 5.2).

### 9.3. El vector y el axial no las mezclan

Ahora son los cruzados los que se anulan, porque la $\gamma^\mu$ cambia el proyector:

$$
\bar\Psi_L\gamma^\mu\Psi_R = \bar\Psi P_R\,\gamma^\mu P_R\Psi = \bar\Psi\gamma^\mu P_L P_R\Psi = 0 ,
$$

y lo mismo $\bar\Psi_R\gamma^\mu\Psi_L$. Sobreviven los de la misma quiralidad:

$$
\boxed{
\bar\Psi\gamma^\mu\Psi = \bar\Psi_L\gamma^\mu\Psi_L + \bar\Psi_R\gamma^\mu\Psi_R
}
\qquad
\bar\Psi\gamma^\mu\gamma^5\Psi = -\bar\Psi_L\gamma^\mu\Psi_L + \bar\Psi_R\gamma^\mu\Psi_R
$$

(en el axial, $\gamma^5\Psi_L = -\Psi_L$ y $\gamma^5\Psi_R = +\Psi_R$). Restando:

$$
\boxed{
\bar\Psi\gamma^\mu\,\frac{1}{2}(1 - \gamma^5)\,\Psi = \frac{1}{2}\left(V^\mu - A^\mu\right) = \bar\Psi_L\gamma^\mu\Psi_L
}
$$

**La regla:** un número **par** de matrices $\gamma$ (S, P y también el tensor, $\sigma^{\mu\nu}$) une L con R; un número **impar**
(V, A) une L con L y R con R.

| forma | en partes quirales | |
| :-- | :-- | :-- |
| $\bar\Psi\Psi$ | $\bar\Psi_L\Psi_R + \bar\Psi_R\Psi_L$ | **mezcla** L y R |
| $\bar\Psi\gamma^5\Psi$ | $\bar\Psi_L\Psi_R - \bar\Psi_R\Psi_L$ | mezcla |
| $\bar\Psi\gamma^\mu\Psi$ | $\bar\Psi_L\gamma^\mu\Psi_L + \bar\Psi_R\gamma^\mu\Psi_R$ | **no mezcla**; L y R por igual |
| $\bar\Psi\gamma^\mu\gamma^5\Psi$ | $-\bar\Psi_L\gamma^\mu\Psi_L + \bar\Psi_R\gamma^\mu\Psi_R$ | no mezcla; L y R con signo opuesto |
| $\frac{1}{2}(V - A)$ | $\bar\Psi_L\gamma^\mu\Psi_L$ | **solo L** |

### 9.4. En la base quiral

Con $\Psi = (\psi_L, \psi_R)$ y $\bar\Psi = (\psi_R^\dagger, \psi_L^\dagger)$ (sección 3.2), todo es explícito. Con $\gamma^0\gamma^0 = I$ y
$\gamma^0\gamma^k = \mathrm{diag}(-\sigma_k, \sigma_k)$, y llamando $\sigma^\mu = (I, \vec\sigma)$ y $\bar\sigma^\mu = (I, -\vec\sigma)$:

$$
\bar\Psi\Psi = \psi_R^\dagger\,\psi_L + \psi_L^\dagger\,\psi_R, \qquad
\bar\Psi\gamma^\mu\Psi = \psi_L^\dagger\,\bar\sigma^\mu\,\psi_L + \psi_R^\dagger\,\sigma^\mu\,\psi_R .
$$

El escalar está **fuera de la diagonal**: une los bloques de arriba y de abajo. La corriente es **diagonal**: cada parte
con la suya, y cada una con su propia matriz, $\bar\sigma^\mu$ o $\sigma^\mu$ (por eso las dos partes son representaciones
distintas, sección 6.1). La corriente $V - A$ es $\psi_L^\dagger\bar\sigma^\mu\psi_L$: **solo necesita el objeto de dos componentes** $\psi_L$.

### 9.5. Qué significa

* **Las interacciones vectoriales**, la electromagnética y la fuerte, acoplan la izquierda y la derecha **por igual
  y por separado**: no distinguen las manos, y conservan la paridad.
* **La corriente cargada de la débil**, $V - A$, solo acopla la izquierda: la parte derecha no participa.
* **Sin masa, la quiralidad se conserva en el vértice**: un electrón que entra L sale L, sea el vértice del fotón o del
  $W$. Como sin masa la quiralidad es la helicidad, a alta energía **la helicidad se conserva en los vértices**
  vectoriales.
* **La masa es lo único que une las dos partes.** Un fermión con solo la parte L, como el neutrino del Modelo
  Estándar, no puede tener un término $m\bar\Psi\Psi$: no tiene masa de Dirac.

***[+] Mirador.*** *En el Modelo Estándar, $e_L$ y $e_R$ son objetos distintos para la interacción débil: $e_L$ va con
$\nu_L$ y $e_R$ va solo. Un término $m\,\bar e_L e_R$ mezclaría dos objetos que la débil trata distinto, y no está permitido.
La masa del electrón la da su acoplo con el campo de Higgs (tema del Modelo Estándar).*

### 9.6. ¿Es la misma en todos los sistemas inerciales?

La corriente cargada une dos campos distintos, por ejemplo el electrón y su neutrino:

$$
j^\mu = \bar e\,\gamma^\mu\,\frac{1}{2}(1 - \gamma^5)\,\nu = \bar e_L\,\gamma^\mu\,\nu_L .
$$

Con una transformación de Lorentz, $\nu \to S\nu$, $\bar e \to \bar e\,S^{-1}$, y como $P_L$ conmuta con $S$:

$$
j'^\mu = \bar e\,S^{-1}\gamma^\mu P_L\,S\,\nu = \bar e\,(S^{-1}\gamma^\mu S)\,P_L\,\nu = \Lambda^\mu{}_\nu\;\bar e\,\gamma^\nu P_L\,\nu .
$$

* **Su forma es la misma en todos los sistemas inerciales**: sigue siendo una corriente entre partes L, $\bar e_L\gamma^\mu\nu_L$.
  Ningún observador ve intervenir la parte R.
* **Su valor no**: es un **cuadrivector**, y sus componentes cambian como las de cualquier cuadrivector. Lo que es
  igual para todos es su contracción con otro cuadrivector, $j^\mu W_\mu$ o $j_{lep}\cdot j_{had}$: un escalar, y con él $M_{fi}$.
* **Con la paridad, en cambio, la forma cambia**: $\gamma^0$ lleva $P_L$ a $P_R$ (sección 2.3), y la corriente se convierte en
  $\bar e_R\gamma^\mu\nu_R$, que no aparece en la teoría. Es la violación de la paridad.
